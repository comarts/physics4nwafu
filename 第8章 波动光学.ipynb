# 本章提要

**1. 光的发射、吸收、散射和色散**
**(1) 光的发射 激光原理**
① 发光的物理机制 发射光谱
光源发出的光是其中大量的分子或原子的能级状态发生变化时所辐射出的电磁波。
把光源物质从不同的激发态跃迁至基态时所发出的一条条离散的谱线，按照波长或频率的不同用一条条直线表示，就得到发射光谱。
② 发光过程的类型
按照基本发光单元激发方式的不同，可将发光过程分为热辐射、电致发光、光致发光和化学发光等。
按照光子的辐射方式的不同，可将发光过程分为自发辐射和受激辐射。
③ 激光原理
激光是受激辐射光放大的简称。要产生激光，必须具备有亚稳态能级结构的工作物质、激励系统和光学谐振腔。
(a) 粒子数反转
要实现并维持粒子数反转，必须具备以下两个条件：其一，必须从外界输入能量，使尽可能多的粒子跃迁到高能级上去，这一能量的供应过程称为激励或光泵浦；其二，工作物质内必须存在亚稳态能级。
(b) 光学谐振腔
光学谐振腔的作用有二：一是选择激光的方向；二是选择激光的频率。
④ 激光的特性
方向性好、强度高、单色性好、相干性好。

**(2) 光的吸收 吸收光谱**
光通过介质时要消耗一部分能量，导致透射光的强度减弱，这种现象称为介质对光的吸收。吸收光辐射或光能是光与物质的基本相互作用之一。
① 光吸收的类型
一般吸收和选择吸收。
② 朗伯定律 比尔定律
朗伯定律的数学形式：
I = I₀ e^(-μl)
比尔定律的数学形式：
I = I₀ e^(-μ₁ c l)
D = ε C l
③ 吸收光谱
产生连续光谱的光源所发出的光，通过具有选择吸收特性的物质后，用光谱仪可以观察到：在连续的发射光谱中，呈现与发生吸收的波长区域相对应的一些暗线或暗带，这就是吸收光谱。
④ 植物对光的吸收
植物对光的吸收主要靠色素系统来实现，这些色素对 300~750 nm 的可见光有不同的吸收率。不同波段的光具有不同的生理效应。

**(3) 光的散射**
光束在介质中传播时，部分光线偏离原方向分散传播的现象称为光的散射。
① 光散射的基本规律
仅考虑散射时，透射光遵守指数衰减规律：
I = I₀ e^(-μ_{s}l)
介质对光的吸收和散射同时存在，透射光的强度为
I = I₀ e^(-(μ_{a}+μ_{s})l)
② 光散射的基本类型
按介质不均匀性质的起因，散射可分为两类：丁铎尔散射和分子散射。
③ 瑞利散射与拉曼散射
散射光中与入射光的频率相同的散射称为瑞利散射。散射光的频率不同于入射光的频率的散射称为拉曼散射。
④ 散射现象的量子解释
入射光子与介质分子发生弹性碰撞产生瑞利散射。光子与分子之间的非弹性碰撞造成拉曼散射。

**(4) 光的色散**
物质的折射率随光的波长的不同而改变的现象称为色散。色散现象也是光和物质相互作用的结果。
① 色散率
ν = dn/dλ
② 正常色散与反常色散
折射率随着波长的减小而减小，此即反常色散；在吸收区域外，折射率随着波长的减小而增大，这是正常色散。

**2. 光的干涉**
**(1) 相干光**
同方向、同频率、相位差恒定的两束或多束光。
**(2) 光程及光程差**
介质的折射率 n 和光波经过的几何路程 r 的乘积称为光程。两束相干光的相位差 Δφ 和光程差 Δ 的关系是
Δφ = 2πΔ / λ
**(3) 半波损失**
光从光疏介质入射到光密介质时，反射光的相位会发生 π 的突变，相当于光程增加或减少 λ/2，这称为半波损失（光透过透镜时不会引起附加的光程差）。
**(4) 杨氏双缝干涉**
明条纹位置：
x_k = ± kDλ / (nd)   (k = 0, 1, 2, ...)
暗条纹位置：
x_k = ± (2k - 1)Dλ / (2nd)   (k = 1, 2, ...)
干涉条纹间距：
Δx = Dλ / (nd)
**(5) 劈尖薄膜的等厚干涉**
干涉加强和减弱条件：
明条纹 Δ = 2ne + λ/2 = kλ   (k = 1, 2, ...)
暗条纹 Δ = 2ne + λ/2 = (2k + 1)/2 λ   (k = 0, 1, 2, ...)
条纹间距：
l = λ / (2n sin θ)

**3. 光的衍射**
**(1) 惠更斯-菲涅耳原理**
介质中波动传播到任一波阵面上的各点，都可以看作发射子波的波源，任意时刻这些子波的包迹就是新的波前。从同一波阵面上各点发出的子波，经传播而在空间某点相遇时，该点的波的强度由各子波在该点的相干叠加决定。
**(2) 衍射现象**
光遇障碍物时偏离直线传播方向，绕过障碍物传播，形成亮度不等、间距不等的明暗相间的条纹的现象。
**(3) 夫琅禾费单缝衍射**
条纹特征：中央为明条纹的对称分布的明暗相间的条纹；中央明条纹宽度是其他明条纹宽度的 2 倍；从中央向两边条纹亮度逐渐减弱。中央明条纹半宽度为 Δx = λf / a。
衍射条纹加强及减弱的条件：
暗条纹 a sin θ = ± kλ   (k = 1, 2, ...)
明条纹 a sin θ = ± (2k + 1)/2 λ   (k = 1, 2, ...)
**(4) 光栅衍射**
条纹特征：在黑暗的背景上显现窄细明亮的谱线，且缝数越多，谱线越细越亮。
光栅方程：
(a + b) sin θ = ± kλ   (k = 0, 1, 2, ...)
**(5) 瑞利判据**
两个物点通过光学仪器形成两个像点，若一个像点的艾里斑的中心刚好与另一个像点衍射图样的第一级暗条纹重合，则这两个物点恰好能被分辨。
最小分辨角：
θ₀ = 1.22 λ / D   (其倒数为分辨率)
最小分辨距离：
δ = 1.22 λf / D
**(6) X 射线衍射**
布拉格公式：
2d sin θ = kλ   (k = 0, 1, 2, ...)

**4. 光的偏振**
**(1) 光的偏振态**
自然光、线偏振光（完全偏振光）、部分偏振光，以及圆偏振光等。
**(2) 线偏振光的获得方法**
① 吸收法；② 反射或折射法；③ 双折射法。
**(3) 布儒斯特定律**
tan i₀ = n₂ / n₁
**(4) 马吕斯定律**
I = I₀ cos² α
**(5) 双折射**
一束自然光沿非光轴方向射入各向异性晶体时分为两束光线，遵守折射定律的光线称为 o 光，不遵守折射定律的光线称为 e 光。o 光和 e 光都是线偏振光。
**(6) 旋光性**
旋光固体：
φ = α_λ l
旋光溶液：
φ = α_λ l C
**(7) 旋光色散**
线偏振光在光学活性物质中传播时，因 n_L ≠ n_R 而出现的旋光现象。
**(8) 圆二色性**
光学各向异性物质对左旋和右旋圆偏振光产生不同吸收的性质。

**5. 激光原理及生物学效应**
**(1) 激光的生物学效应**
激光的生物学效应一般是指激光作用于生物体可能产生的物理、化学或生物学的反应。按机理划分，激光的生物学效应大致有五种：热效应、光化学效应、机械效应、电磁场效应和刺激效应。
**(2) 激光在现代农业与生物科学中的应用**

激光荧光光谱技术、激光拉曼光谱技术、激光光声光谱技术等。

# 一、思考题解答（8.1 - 8.26）

**8.1 光与物质的相互作用表现在哪几个方面？什么是物质的吸收光谱？吸收光谱有哪些实际应用？**
*   **解题思路**：结合光与物质相互作用的物理机制作答。
*   **解答**：光与物质的相互作用主要表现在光的吸收、光的散射、光的色散、光的反射与折射以及光电效应等方面。物质的吸收光谱是指物质吸收特定波长光后形成的暗线光谱。实际应用包括：利用吸收光谱进行物质成分分析（光谱化学分析）、研究原子和分子的能级结构等。

**8.2 物质对光的吸收有哪两种类型？**
*   **解题思路**：根据吸收与波长的关系分类。
*   **解答**：分为普遍吸收（一般吸收）和选择吸收。普遍吸收是指物质对一定波长范围内各种波长的光吸收程度大致相同；选择吸收是指物质对某些特定波长的光有强烈的吸收。

**8.3 什么是光的散射？它分为哪两种类型？**
*   **解题思路**：散射的定义与分类。
*   **解答**：光的散射是指光在传播过程中，由于介质的不均匀性而使部分光线偏离原传播方向的现象。分为瑞利散射和拉曼散射（或分为弹性散射和非弹性散射）。

**8.4 什么是拉曼散射？激光拉曼散射光谱仪包括哪几部分，各有什么作用？**
*   **解题思路**：拉曼散射概念及仪器结构。
*   **解答**：拉曼散射是光通过介质时，光子与分子发生非弹性碰撞，导致散射光频率发生改变的现象。光谱仪主要包括：激光光源（提供单色光）、样品池（放置被测物）、单色仪（分光）、探测器（接收光信号）和数据处理系统。

**8.5 什么是光的色散？什么是正常色散和反常色散？**
*   **解题思路**：折射率随波长变化的规律。
*   **解答**：光的色散是指复色光分解为单色光的现象。正常色散是指折射率随波长增大而减小的现象；反常色散是指折射率随波长增大而增大的现象。

**8.6 为什么用两个独立的普通光源观察不到干涉图样？普通光源的相干条件是什么？一般用普通光源获得相干光的方法有哪些？**
*   **解题思路**：相干条件与获取相干光的方法。
*   **解答**：普通光源发光是原子的自发辐射，不同原子或同一原子不同时刻发出的光波相位差随机变化，不满足相干条件。相干条件：频率相同、振动方向相同、相位差恒定。获得相干光的方法：分波阵面法（如杨氏双缝）和分振幅法（如薄膜干涉）。

**8.7 为什么要引入光程的概念？如何理解光程？光程差和相位差的关系是什么？**
*   **解题思路**：光程的物理意义。
*   **解答**：引入光程是为了方便比较光在不同介质中传播时的相位变化。光程是指光在介质中传播的几何路程与介质折射率的乘积（$$L = nr$$）。光程差与相位差的关系为：$$\Delta \varphi = \frac{2\pi}{\lambda} \Delta L$$。

**8.8 杨氏双缝干涉实验中用白光作为光源，若在缝 $S_1$ 后放一片红色滤光片，缝 $S_2$ 后放一片绿色滤光片，问能否观察到干涉条纹？为什么？**
*   **解答**：不能。因为红色光和绿色光频率不同，不满足相干条件（频率相同），无法产生稳定的干涉图样。

**8.9 观察肥皂液膜的干涉时，刚吹起的肥皂泡没有颜色，吹到一定大小时会看到彩色，其颜色随泡的增大而改变，当彩色消失呈现黑色时，肥皂泡破裂，试解释此现象。**
*   **解答**：刚吹起时膜较厚，光程差远大于可见光波长，不产生干涉。随着泡增大，膜变薄，光程差减小，不同波长的光满足不同的干涉加强条件，呈现彩色。当膜极薄时，上下表面反射光产生半波损失抵消（或满足干涉相消条件），反射光减弱，透射光增强，呈现黑色，此时表面张力不足以维持，泡破裂。

**8.10 照相机镜头一般呈蓝色，试运用光学原理解释此现象。**
*   **解答**：照相机镜头上镀有增透膜。根据薄膜干涉原理，增透膜使反射光中特定波长（如黄绿色，人眼最敏感）的光干涉相消，从而增加透射光。由于反射光中蓝紫色光未被完全抵消，所以镜头看起来呈蓝紫色。

**8.11 如果形成空气劈尖的两块玻璃板内表面凹凸不平，那么空气的等厚干涉条纹还平行于棱边吗？为什么？**
*   **解答**：不平行。等厚干涉条纹是厚度相同的点连成的线。若表面凹凸不平，等厚线将随表面起伏而弯曲，不再平行于棱边。

**8.12 对于劈尖薄膜干涉，如果发生以下变化，其干涉条纹将如何变化？（1）仅倾角 $\theta$ 变化；（2）仅膜厚 $e$ 变化；（3）仅劈尖中介质的折射率 $n$ 变化。**
*   **解答**：
    (1) $\theta$ 变大，条纹间距 $$l = \frac{\lambda}{2n\sin\theta}$$ 变小，条纹变密。
    (2) 膜厚 $e$ 整体变化不影响条纹间距，但会引起条纹的移动。
    (3) $n$ 变大，条纹间距 $l$ 变小，条纹变密。

**8.13 单缝衍射和双缝干涉有什么相同点？又有什么区别？**
*   **解答**：相同点是两者都是光的波动性表现，都有明暗相间的条纹。区别在于：双缝干涉是两束光相干叠加，明纹亮度基本相同，间距均匀；单缝衍射是无数子波相干叠加，中央明纹最亮且宽，两侧明纹亮度递减，间距不等。

**8.14 单缝衍射实验中，当缝的宽度比光的波长大得多或小得多时，为什么都观察不到明显的衍射条纹？**
*   **解答**：缝宽远大于波长时，光近似直线传播，衍射现象不明显；缝宽远小于波长时，光强太弱，无法观察到清晰的衍射图样。

**8.15 隔着山可以收听到中波段的电台广播，而超短波广播、电视广播很容易被山或大型建筑物挡住，雷达电波几乎是直线传播的。试利用波动光学理论分析这些现象。**
*   **解答**：波长越长，衍射现象越显著。中波波长较长，容易绕过山丘；超短波、雷达波波长较短，衍射现象不明显，近似直线传播，容易被障碍物挡住。

**8.16 “隔着门缝看人——把人看扁了”，从波动光学的角度看，这句歇后语有道理吗？**
*   **解答**：没有道理。从波动光学看，门缝相当于单缝，光通过门缝发生衍射，人眼接收到的图像不会在垂直于门缝的方向上被“压扁”，而是会在垂直于缝的方向上发生展宽。

**8.17 在单缝衍射中，增大波长和改变缝宽对衍射图样有何影响？**
*   **解答**：增大波长，中央明纹半宽度 $$\Delta x = \frac{\lambda f}{a}$$ 增大，条纹变宽；增大缝宽 $a$，条纹变窄。

**8.18 回答下列问题：(1) 怎样减小最小分辨角 $\theta_0$（即增大分辨本领）？(2) 一台显微镜，用 550 nm 的黄绿光照射时的最小分辨角比用 270 nm 的紫光照射时的大多少倍？**
*   **解答**：
    (1) 根据 $$\theta_0 = 1.22 \frac{\lambda}{D}$$，可以通过增大透镜孔径 $D$ 或减小照射光波长 $\lambda$ 来减小最小分辨角。
    (2) $\theta_{550} / \theta_{270} = 550 / 270 \approx 2.037$ 倍。

**8.19 为什么天文望远镜的直径很大？用显微镜对物体进行显微摄影时，为什么用波长较短的光照射较好？**
*   **解答**：根据瑞利判据 $\theta_0 = 1.22 \lambda / D$，增大直径 $D$ 可以减小最小分辨角，提高分辨本领，看清更远的天体。显微镜用短波长光照射，也可以减小最小分辨角，提高分辨率。

**8.20 光栅衍射明条纹的条件是 $(a+b)\sin\theta = \pm k\lambda$；单缝衍射暗条纹的条件是 $a\sin\theta = \pm k\lambda$。上述两公式有矛盾吗？为什么？**
*   **解答**：不矛盾。单缝衍射暗纹条件是单缝内各子波干涉相消的结果，而光栅明纹条件是各缝之间干涉相长的结果。光栅衍射图样是单缝衍射和多缝干涉的总效果，两者物理机制不同。

**8.21 光的衍射效应为什么会限制光学仪器的分辨率？**
*   **解答**：由于衍射效应，物点通过光学仪器成的像不是一个几何点，而是一个有一定大小的艾里斑。当两个物点距离太近时，它们的艾里斑会重叠，导致无法分辨。

**8.22 自然光和线偏振光、部分偏振光有何区别？**
*   **解答**：自然光在垂直于传播方向的平面内各方向光矢量振幅相等；线偏振光光矢量只在一个固定方向振动；部分偏振光各方向光矢量振幅不等，但有一个优势方向。

**8.23 什么是双折射现象？双折射晶体的光轴是否只是一条线，或者是空间的一个方向？**
*   **解答**：双折射现象是指一束光射入各向异性晶体后分为两束折射光的现象。光轴是晶体内的一个特定方向，光沿此方向传播时不发生双折射。光轴是空间的一个方向，并非特指某一条线。

**8.24 车辆在夜间行驶时，司机最怕迎面驶来的车灯“眩光”，你能否用学过的知识设计一个方案，使司机既能克服“眩光”又不影响夜间的视线？**
*   **解答**：可在车灯和司机观察窗上安装偏振片，并使两者的偏振化方向互相垂直。这样迎面车灯发出的光无法通过司机前方的偏振片，从而消除眩光。同时，司机自己车灯发出的光经路面反射后，仍可被自己看到。

**8.25 什么是旋光效应？**
*   **解答**：旋光效应是指线偏振光通过某些物质（如石英、糖溶液）后，其振动面发生旋转的现象。

**8.26 什么是圆二色性？**
*   **解答**：圆二色性是指光学各向异性物质对左旋和右旋圆偏振光产生不同吸收的性质。

# 二、习题解答（8.1 - 8.31）

**8.1** 如图所示，$S_1$ 与 $S_2$ 是同一振荡器激发的两个有效的点状辐射源，这两个辐射源相距 4.0 m，初相位均为零，发射功率相同的电磁波，波长均为 1.0 m，试求：(1) 当把检测器沿 $x$ 轴移动时，由检测器依次接收到的极大信号的位置；(2) 离辐射源最近的信号极小的位置。
*   **解题思路**：利用波的干涉条件。波程差为波长整数倍时极大，半波长奇数倍时极小。
*   **解答**：设 $S_1$ 在原点，$S_2$ 在 $(4, 0)$。检测器在 $x$ 轴上。
    (1) 极大条件：$$\Delta r = |x - 4| - |x| = k\lambda = k \times 1.0$$（$k=0, \pm 1, \pm 2...$）。
    当 $x$ 在 $S_1, S_2$ 之间 ($0 \le x \le 4$) 时，$4 - 2x = k \Rightarrow x = 2 - k/2$。取 $k=0, \pm 1, \pm 2, \pm 3, \pm 4$，得 $x = 2, 2.5, 1.5, 3, 1, 3.5, 0.5, 4, 0$ m。
    (2) 极小条件：$$\Delta r = |x - 4| - |x| = (k + 0.5)\lambda$$。
    取 $k=0, 1, 2, 3$，得 $4 - 2x = k + 0.5 \Rightarrow x = 1.75 - k/2$。
    离辐射源最近的位置为 $x=0$ 或 $x=4$。代入 $x=0$，$\Delta r = 4 - 0 = 4$ m（非极小）。$x=4$，$\Delta r = 4$ m。最近极小位置：当 $k=3$ 时，$x=0.25$ m。

**8.2** 如图所示，光源经过狭缝 $S_1$ 和 $S_2$ 发出波长 $\lambda=500\text{ nm}$ 的相干光。在 $S_1$ 前放一块薄玻璃片 $G$，在屏上得到一组稳定的干涉条纹，然后在 $G$ 的一面均匀镀上透明介质薄膜，薄膜的折射率为 $n=2.23$。随着薄膜的厚度由零逐渐增加到 $d$，$P$ 点处由明条纹逐渐变为暗条纹，求 $d$ 的值。

**8.3** 在杨氏双缝干涉实验中，波长为 500 nm 的单色光垂直入射到间距为 $d=0.5\text{ mm}$ 的双缝上，屏到双缝中心的距离为 $D=1.0\text{ m}$。(1) 求第 10 级明条纹中心的位置；(2) 求条纹宽度；(3) 用一云母片 ($n=1.58$) 覆盖其中一缝，中央明条纹移到原来第 8 级明条纹中心处，问云母片的厚度是多少？
*   **解题思路**：(1)(2) 直接使用杨氏双缝干涉公式。(3) 云母片引入附加光程差，引起条纹移动。
*   **解答**：

    (2) 条纹宽度：$$\Delta x = \frac{D\lambda}{d} = \frac{1.0 \times 500 \times 10^{-9}}{0.5 \times 10^{-3}} = 1.0\text{ mm}$$。
    (3) 中央明纹移动意味着光程差为零的点移动，附加光程差为 $(n-1)e = 8\lambda$。

**8.4** 一双缝装置的一个缝被折射率为 1.40 的薄玻璃片遮盖，另一个缝被折射率为 1.70 的薄玻璃片遮盖。在玻璃片插入以后，屏上原来的中央极大所在处，变为第 5 级明条纹。假设 $\lambda=480\text{ nm}$，且玻璃片厚度均为 $d$，求 $d$。

*   **解答**：附加光程差 $$\Delta = (n_2 - 1)d - (n_1 - 1)d = (n_2 - n_1)d = 5\lambda$$。

**8.6** 白光垂直照射在空气中厚度为 0.40 μm、折射率为 1.50 的玻璃片上。试问在可见光范围内 (380~750 nm)：(1) 哪些波长的光在反射中增强？(2) 哪些波长的光在透射中增强？
*   **解题思路**：薄膜干涉。反射增强条件 $2ne + \lambda/2 = k\lambda$；透射增强条件 $2ne = k\lambda$。
*   **解答**：$2ne = 2 \times 1.50 \times 0.40 \times 10^{-6} = 1.2 \times 10^{-6}\text{ m} = 1200\text{ nm}$。
    (1) 反射增强：$$2ne + \frac{\lambda}{2} = k\lambda \Rightarrow \lambda = \frac{2400}{2k-1}\text{ nm}$$。
    $k=2, \lambda=800\text{ nm}$（不可见）；$k=3, \lambda=480\text{ nm}$（可见）；$k=4, \lambda=342.8\text{ nm}$（不可见）。所以反射增强波长为 480 nm。
    (2) 透射增强：$$2ne = k\lambda \Rightarrow \lambda = \frac{1200}{k}\text{ nm}$$。
    $k=2, \lambda=600\text{ nm}$（可见）；$k=3, \lambda=400\text{ nm}$（可见）；$k=4, \lambda=300\text{ nm}$（不可见）。所以透射增强波长为 400 nm 和 600 nm。

**8.7** 一平面单色光垂直照射在厚度均匀的薄膜上，膜覆盖在玻璃板上，膜的折射率为 1.30，玻璃的折射率为 1.50。若单色光的波长由光源连续可调，可观察到 700 nm 与 500 nm 这两个波长的单色光在反射中消失，试求薄膜的最小厚度。
*   **解题思路**：反射消失即干涉相消。满足 $2ne = (k+1/2)\lambda$。
*   **解答**：设膜厚为 $e$。反射光消失条件：$$2ne = (k_1 + \frac{1}{2})\lambda_1 = (k_2 + \frac{1}{2})\lambda_2$$。
    代入得：$$2ne = \frac{2k_1+1}{2} \times 700 = \frac{2k_2+1}{2} \times 500 \Rightarrow 7(2k_1+1) = 5(2k_2+1)$$。
    取最小整数解，令 $2k_1+1 = 5, 2k_2+1 = 7$，即 $k_1=2, k_2=3$。
    最小厚度：$$e = \frac{5 \times 700 \times 10^{-9}}{4 \times 1.30} = \frac{3.5 \times 10^{-6}}{5.2} \approx 673.1\text{ nm}$$。

**8.14** 用单色光观察牛顿环，测得某一明环的直径为 3.0 mm，而其外第五个明环的直径为 4.60 mm。若平凸透镜凸面的曲率半径为 1.03 m，试求该单色光的波长。
*   **解题思路**：牛顿环明环半径公式 $$r_k^2 = (k - 0.5)R\lambda$$（或 $r_k^2 = kR\lambda$，视中心是暗斑还是亮斑而定，通常中心为暗斑，亮环半径平方差 $\Delta r^2 = 5R\lambda$）。

*   **解答**：
    (1) 角分辨率 $$\theta = \frac{d}{H} = \frac{0.05}{160 \times 10^3} = 3.125 \times 10^{-7}\text{ rad}$$。
    (2) 孔径 $$D = 1.22 \frac{\lambda}{\theta} = 1.22 \times \frac{500 \times 10^{-9}}{3.125 \times 10^{-7}} \approx 1.952\text{ m}$$。

**8.23** 美国波多黎各阿雷西博谷地的无线电天文望远镜的“物镜”镜面孔径为 300 m，曲率半径也是 300 m，其工作的最短波长是 4 cm。对于此波长，这台望远镜的角分辨率是多少？
*   **解题思路**：直接套用瑞利判据。
*   **解答**：$$\theta = 1.22 \frac{\lambda}{D} = 1.22 \times \frac{0.04}{300} \approx 1.627 \times 10^{-4}\text{ rad}$$。

**8.24** 已知天空中两颗星相对于一望远镜的角距离为 $4.8 \times 10^{-6}\text{ rad}$，它们都发出波长为 $\lambda=550\text{ nm}$ 的光。问望远镜的口径至少要多大，才能分辨出这两颗星？
*   **解题思路**：由 $$\theta = 1.22 \lambda / D$$ 求解 $D$。
*   **解答**：$$D = 1.22 \frac{\lambda}{\theta} = 1.22 \times \frac{550 \times 10^{-9}}{4.8 \times 10^{-6}} \approx 0.1398\text{ m} = 13.98\text{ cm}$$。

**8.25** 水的折射率为 1.33，玻璃的折射率为 1.50。当光由水射向玻璃而反射时，起偏角为多少？当光由玻璃射向水而反射时，起偏角又为多少？
*   **解题思路**：布儒斯特定律 $$\tan i_0 = n_2 / n_1$$。
*   **解答**：
    (1) 水射向玻璃：$$\tan i_0 = \frac{1.50}{1.33} \approx 1.128 \Rightarrow i_0 \approx 48.44^\circ$$。
    (2) 玻璃射向水：$$\tan i_0 = \frac{1.33}{1.50} \approx 0.887 \Rightarrow i_0 \approx 41.56^\circ$$。

**8.26** 光在某种介质界面上的临界角是 45°，它在界面同一侧的起偏角是多少？
*   **解题思路**：临界角 $C$ 满足 $$\sin C = 1/n$$。起偏角满足 $$\tan i_0 = n$$。
*   **解答**：由 $$\sin 45^\circ = 1/n \Rightarrow n = \sqrt{2} \approx 1.414$$。
    $$\tan i_0 = \sqrt{2} \Rightarrow i_0 = \arctan(\sqrt{2}) \approx 54.74^\circ$$。

**8.27** 使自然光通过两个偏振化方向相交 60° 的偏振片，透射光强为 $I_1$，今在这两个偏振片之间再插入另一偏振片，它的偏振化方向与前两个偏振片均成 30°，则透射光强为多少？
*   **解题思路**：马吕斯定律 $$I = I_0 \cos^2 \alpha$$，自然光通过第一偏振片光强减半。
*   **解答**：设自然光强为 $I_0$。通过第一偏振片后光强 $$I_a = I_0 / 2$$。
    通过第二偏振片后：$$I_1 = I_a \cos^2 60^\circ = \frac{I_0}{2} \times \frac{1}{4} = \frac{I_0}{8}$$。
    插入第三偏振片（与前两个均成 30°）：
    通过第三偏振片光强 $$I_b = I_a \cos^2 30^\circ = \frac{I_0}{2} \times \frac{3}{4} = \frac{3I_0}{8}$$。
    通过第二偏振片光强 $$I_2 = I_b \cos^2 30^\circ = \frac{3I_0}{8} \times \frac{3}{4} = \frac{9I_0}{32}$$。

**8.28** 一束光强为 $I_0$ 的自然光，相继通过两个偏振片 $P_1, P_2$ 后出射光强为 $I_0/4$。若以入射光线为轴旋转 $P_2$，要使出射光强为零，$P_2$ 至少应转过的角度是多少？
*   **解题思路**：自然光通过 $P_1$ 变为 $I_0/2$，通过 $P_2$ 光强 $I_0/2 \cos^2 \theta = I_0/4 \Rightarrow \theta = 45^\circ$ 或 $135^\circ$。
*   **解答**：当前 $P_1$ 与 $P_2$ 偏振化方向夹角为 $45^\circ$。要使出射光强为零，两偏振化方向需垂直（夹角 $90^\circ$）。
    因此 $P_2$ 至少应转过 $$90^\circ - 45^\circ = 45^\circ$$。

**8.29** 两个偏振片 $P_1, P_2$ 叠放在一起，其偏振化方向之间的夹角为 30°。一束光强为 $I_0$ 的光垂直入射到偏振片上，已知该入射光由光强相同的自然光和线偏振光混合而成，现测得透过偏振片后的出射光强与入射光强之比为 9/16，试求入射光中线偏振光的光矢量的振动方向（以 $P_1$ 的偏振化方向为基准）。
*   **解题思路**：设自然光强为 $I_n$，线偏振光强为 $I_l$，且 $I_n = I_l = I_0/2$。
*   **解答**：通过 $P_1$ 后光强 $$I_1 = \frac{I_n}{2} + I_l \cos^2 \alpha$$（$\alpha$ 为线偏振光与 $P_1$ 的夹角）。
    通过 $P_2$ 后光强 $$I_2 = I_1 \cos^2 30^\circ = \left(\frac{I_n}{2} + I_l \cos^2 \alpha\right) \times \frac{3}{4}$$。
    已知 $I_n = I_l = I_0/2$，代入：
    $$I_2 = \left(\frac{I_0}{4} + \frac{I_0}{2} \cos^2 \alpha\right) \times \frac{3}{4} = \frac{3I_0}{16} (1 + 2\cos^2 \alpha)$$。
    由题意 $I_2 / I_0 = 9/16 \Rightarrow \frac{3}{16} (1 + 2\cos^2 \alpha) = \frac{9}{16} \Rightarrow 1 + 2\cos^2 \alpha = 3 \Rightarrow \cos^2 \alpha = 1$。
    所以 $\alpha = 0$，即线偏振光的光矢量振动方向与 $P_1$ 的偏振化方向平行。

**8.30** 两偏振片的偏振化方向成 30° 角，透射光强为 $I_1$，若入射光不变而使两偏振片偏振化方向的夹角变为 45°，则透射光强将如何变化？
*   **解题思路**：马吕斯定律。
*   **解答**：$$I_1 = I \cos^2 30^\circ = \frac{3}{4} I$$。
    $$I_2 = I \cos^2 45^\circ = \frac{1}{2} I$$。
    $$\frac{I_2}{I_1} = \frac{1/2}{3/4} = \frac{2}{3}$$。透射光强变为原来的 $\frac{2}{3}$。

**8.31** 一未知浓度的葡萄糖水溶液装满于 12.9 m 长的玻璃管中，当一单色线偏振光垂直于管端面并沿管的中心轴线通过时，由检偏器测得光的振动面旋转了 1.23°。已知葡萄糖溶液的旋光率为 $20.5^\circ \cdot \text{cm}^3 / (\text{dm} \cdot \text{g})$，求该葡萄糖溶液的浓度。
*   **解题思路**：旋光公式 $$\varphi = \alpha_\lambda l C$$。
*   **解答**：已知 $\varphi = 1.23^\circ$，$l = 12.9\text{ m} = 129\text{ dm}$，$\alpha_\lambda = 20.5^\circ \cdot \text{cm}^3 / (\text{dm} \cdot \text{g})$。
    $$C = \frac{\varphi}{\alpha_\lambda l} = \frac{1.23}{20.5 \times 129} \approx 4.65 \times 10^{-4}\text{ g/cm}^3$$。

## 思考题
### 8.1 光与物质的相互作用表现在哪几个方面？
**思路**：教材光和物质作用分类
答：主要表现为**光的吸收、光的散射、光的色散**三类。

### 8.2 物质对光的吸收有哪两种类型？什么是物质的吸收光谱？吸收光谱有哪些实际应用？
**思路**：吸收分为普遍吸收、选择吸收。
- 两种类型：
  1. 普遍吸收：物质对所有波长光几乎同等吸收；
  2. 选择吸收：物质只选择性吸收某些特定波长的光。
- 吸收光谱：连续谱光源透过物质后，被吸收波长位置出现暗线/暗带的光谱。
- 应用：光谱分析、物质成分检测、定量分析、红外检测、大气成分探测等。

### 8.3 什么是光的散射？它分为哪两种类型？
**思路**：散射定义；瑞利散射、拉曼散射
答：光在介质传播时，介质不均匀，部分光偏离原传播方向向四周发射，叫光的散射。
分为**瑞利散射**（弹性散射，散射光波长不变）、**拉曼散射**（非弹性散射，散射光波长改变）。

### 8.4 什么是拉曼散射？激光拉曼散射光谱仪包括哪几部分，各有什么作用？
**思路**：拉曼是非弹性光散射；仪器构成
答：拉曼散射：光子与介质分子发生能量交换，散射光频率（波长）不同于入射光的非弹性散射。
仪器组成：
1. 激光光源：高强度单色入射激发光；
2. 样品池：放置待测样品；
3. 单色仪：分光，分离不同波长散射光；
4. 探测器：接收光信号转为电信号；
5. 信号处理系统：采集、输出拉曼光谱。

### 8.5 什么是光的色散？什么是正常色散和反常色散？
**思路**：折射率随波长变化
答：色散：介质折射率$n$随入射光波长$\lambda$改变，不同色光折射角不同。
- 正常色散：波长增大，折射率减小（可见光区大多数透明介质）；
- 反常色散：波长增大，折射率增大，发生在介质吸收带附近。

### 8.6 为什么用两个独立的普通光源观察不到干涉图样？普通光源的相干条件是什么？一般用普通光源获得相干光的方法有哪些？
**思路**：普通光源发光是大量原子随机独立发光，相位无稳定差。
- 原因：独立光源原子发光随机，相位差随时间无规则快速变化，干涉条纹极速抖动，人眼看不到。
- 相干条件：①频率相同；②振动方向相同；③相位差恒定。
- 获得相干光方法：**分波面法（杨氏双缝）、分振幅法（薄膜干涉）**。

### 8.7 为什么要引入光程的概念？如何理解光程？光程差和相位差的关系是什么？
**思路**：不同介质光速不同，直接几何路程不能算相位变化，引入光程统一折算到真空中。
- 光程定义：$\(L=nr\)$，介质折射率×几何路程，把介质中路程等效成真空中光走过的距离。
- 目的：统一计算不同介质中传播产生的相位变化。
- 关系：
$$\Delta\varphi=\frac{2\pi}{\lambda}\Delta$$
$\Delta$为光程差，$\lambda$是真空中波长。

### 8.8 杨氏双缝干涉实验中用白光作为光源，若在缝$S_1$后放一片红色滤光片，缝$S_2$后放一片绿色滤光片，问能否观察到干涉条纹？为什么？
**思路**：相干条件第一条：频率必须相同。
答：**不能**。红光和绿光频率不同，不满足相干条件，无法产生稳定干涉。

### 8.9 观察肥皂液膜的干涉时，刚吹起的肥皂泡没有颜色，吹到一定大小时会看到彩色，其颜色随泡的增大而改变，当彩色消失呈现黑色时，肥皂泡破裂，试解释此现象。
**思路**：薄膜干涉+半波损失；膜厚变化。
1. 刚开始：膜很厚，光程差远大于相干长度，不干涉，无彩色；
2. 慢慢变薄，膜厚和可见光波长相当，薄膜上下表面反射光发生干涉，出现彩色；膜不断变薄，光程差改变，颜色随之改变；
3. 膜极薄时，膜厚$e\to0$，上下表面反射：上表面空气→肥皂（光疏到密，有$\lambda/2$半波损失）；下表面肥皂→空气（光密到疏，无半波损失）。\(I_{\(\frac{I_{out}}{I_{总}}=\frac{\left(\frac12+\cos^2\theta\right)\cdot\frac34}{2}=\frac9{16}\)}=I_{nat}+I_{lin}=I+I=2I\)光程差≈$\lambda/2$，所有可见光都满足相消干涉，看起来黑色；膜强度不足随即破裂。

### 8.10 照相机镜头一般呈蓝色，试用光学原理解释此现象。
**思路**：增透膜。镜头镀一层薄膜，使人眼最敏感黄绿光透射增强，反射光中黄绿光相消，反射剩下红蓝，视觉呈蓝紫色。

### 8.11 如果形成空气劈尖的两块玻璃板内表面凹凸不平，那么空气的等厚干涉条纹还平行于棱边吗？为什么？
**思路**：等厚条纹特点：**同一根条纹对应薄膜同一厚度**。
答：不再平行于棱边。等厚条纹上每一点空气膜厚度相同；玻璃板凹凸，等厚位置偏离平直棱边，条纹随凹凸变形。

### 8.12 对于劈尖薄膜干涉，如果发生以下变化，其干涉条纹将如何变化？(1)仅倾角$\theta$变化；(2)仅膜厚$e$变化；(3)仅劈尖中介质的折射率$n$变化。
劈尖条纹间距：$l=\frac{\lambda}{2n\theta}$
1. $\theta$增大：条纹变密，向棱边靠拢；$\theta$减小，条纹变疏；
2. 整体膜厚均匀增加：条纹整体向棱边平移，条纹间距不变；
3. $n$增大，$l$减小，条纹变密；$n$减小，条纹变疏。

### 8.13 单缝衍射和双缝干涉有什么相同点？又有什么区别？
**思路**：干涉是多光束叠加；衍射是同一波面子波叠加。
相同：都是光波相干叠加，都出现明暗相间条纹，都满足相位差导致相长/相消。
区别：
1. 成因：双缝干涉是**两个独立子波源**的光叠加；单缝衍射是**单缝波面上无数子波**叠加；
2. 条纹：双缝条纹等宽等间距；单缝衍射中央明纹宽，两侧明纹窄、亮度衰减；
3. 公式：双缝 $d\sin\theta=k\lambda$；单缝暗纹 $b\sin\theta=k\lambda$。

### 8.14 单缝衍射实验中，当缝的宽度比光的波长大得多或小得多时，为什么都观察不到明显的衍射条纹？
- $b\gg\lambda$：衍射角极小，条纹挤在中央，近似直线传播，看不到衍射；
- $b\ll\lambda$：整个屏幕只有一片亮区，明暗条纹无法分辨，没有明暗交替。
衍射明显条件：缝宽与波长数量级相近。

### 8.15 隔着山可以收听到中波段的电台广播，而超短波广播、电视广播很容易被山或大型建筑物挡住，雷达电波几乎是直线传播的。试利用波动光学理论分析这些现象。
**思路**：衍射条件：障碍物尺寸与波长可比，衍射显著。
中波波长很长，山体尺寸和波长可比，**衍射明显**，可以绕山传播；
超短波、电视、雷达波波长很短，远小于山体，衍射效应极弱，近似直线传播，无法绕过大山。

### 8.16 “隔着门缝看人——把人看扁了”，从波动光学的角度看，这句歇后语有道理吗？
**思路**：门缝衍射。
答：没有道理。门缝衍射，竖直方向门缝宽度大，竖直方向衍射不明显；水平缝窄，水平方向衍射展宽，只会让人在水平方向展宽，不是压扁。

### 8.17 在单缝衍射中，增大波长和改变缝宽对衍射图样有何影响？
公式：中央明纹宽度$\Delta x_0=\frac{2\lambda D}{b}$
- $\lambda$增大：衍射角变大，条纹整体拉宽；
- 缝宽$b$增大：衍射角变小，条纹收缩变密；$b$减小，条纹展宽。

### 8.18 回答下列问题：(1)怎样减少最小分辨角$\theta_0$(即增大分辨本领)？(2)一台显微镜，用550 nm的黄绿光照射时的最小分辨角比用270 nm的紫光照射时的大多少倍？
瑞利判据：$\theta_0=1.22\frac{\lambda}{D}$
(1) 增大仪器孔径$D$，或减小入射光波长$\lambda$，减小最小分辨角，提高分辨本领；
(2)
$$\frac{\theta_{01}}{\theta_{02}}=\frac{\lambda_1}{\lambda_2}=\frac{550}{270}\approx 2.04$$
黄绿光最小分辨角是紫光的2.04倍。

### 8.19 为什么天文望远镜的直径很大？用显微镜对物体进行显微摄影时，为什么用波长较短的光照射较好？
思路：瑞利判据。
望远镜：$D$越大，最小分辨角越小，分辨本领越高，可以分辨遥远天体；
显微镜：波长越短，最小分辨角越小，分辨本领越高，看清微小细节。

### 8.20 光栅衍射明条纹的条件是$(a+b)\sin\theta=\pm k\lambda$；单缝衍射暗条纹的条件是$a\sin\theta=\pm k\lambda$。上述两公式有矛盾吗？为什么？
**无矛盾**
光栅方程：**多缝干涉**，相邻缝之间光程差；
单缝衍射暗纹：**单缝内部子波叠加相消**。
光栅衍射是单缝衍射调制多缝干涉；当光栅某级干涉主极大刚好落在单缝衍射暗纹位置，该级条纹消失，叫缺级。两个式子描述的物理对象不一样。

### 8.21 光的衍射效应为什么会限制光学仪器的分辨率？
**思路**：圆孔衍射+瑞利判据。
透镜圆孔会产生圆孔衍射，点光源成像不是点，是爱里斑。两个靠近物点对应的爱里斑重叠到一定程度，人眼无法区分两个像。衍射决定最小分辨角，限制分辨极限。

### 8.22 自然光和线偏振光、部分偏振光有何区别？
- 自然光：垂直传播方向，各个方向光振动均匀对称，各方向振幅相同；
- 线偏振光（完全偏振光）：光振动只在一个固定方向；
- 部分偏振光：不同方向振动强弱不等，可看成自然光叠加线偏振光。

### 8.23 什么是双折射现象？双折射晶体的光轴是否只是一条线，或者是空间的一个方向？
自然光入射各向异性晶体，分成o光、e光两束折射光叫双折射。
**光轴不是一条线，是晶体内部一个特定方向**；沿光轴方向传播o光e光折射率相同，不发生双折射。

### 8.24 车辆在夜间行驶时，司机最怕迎面驶来的车灯“眩光”，你能否用学过的知识设计一个方案，使司机既能克服“眩光”又不影响夜间的视线？
思路：偏振片。
方案：所有汽车车灯和挡风玻璃，都装上偏振片，偏振化方向统一倾斜45°。本车车灯出射光偏振方向与挡风偏振片平行，自己能看清路面；对面来车车灯偏振方向与本车挡风偏振方向垂直，眩光被消光，消除强光。

### 8.25 什么是旋光效应？
线偏振光透过某些物质，振动面会绕传播方向旋转，这个现象叫旋光效应。旋转角度和物质厚度、浓度、波长有关。

### 8.26 什么是圆二色性？
光学各向异性物质，对左旋圆偏振光、右旋圆偏振光吸收系数不同，这个性质叫圆二色性。

自然光部分：过P1，光强$\(I/2\)$；线偏振入射，与P1夹角$\theta$，过P1后$I\cos^2\theta$。
到达P2的光强：$\left(\frac I2+I\cos^2\theta\right)$，P1-P2夹角30°，马吕斯：
$$I_{out}=\left(\frac I2+I\cos^2\theta\right)\cos^230^\circ$$
$$\frac{I_{out}}{I_{\(I_{\(\frac{I_{out}}{I_{总}}=\frac{\left(\frac12+\cos^2\theta\right)\cdot\frac34}{2}=\frac9{16}\)}=I_{nat}+I_{lin}=I+I=2I\)}}=\frac{\left(\frac12+\cos^2\theta\right)\cdot\frac34}{2}=\frac9{16}$$

## 8.30 偏振片夹角30°透射$I_1$；变成45°，求新光强比值。
入射自然光，第一偏振片后$I=I_0/2$
$$I_1=\frac{I_0}{2}\cos^230^\circ=\frac{I_0}{2}\cdot\frac34=\frac{3I_0}{8}$$
$$I_2=\frac{I_0}{2}\cos^245^\circ=\frac{I_0}{2}\cdot\frac12=\frac{I_0}{4}$$
$$\frac{I_2}{I_1}=\frac{\frac{I_0}{4}}{\frac{3I_0}{8}}=\frac23$$
新透射光强$I_2=\frac23 I_1$
