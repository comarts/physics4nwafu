# 本章提要


**1. 简谐振动**

(1) 运动方程
$$x=A\cos(\omega t+\varphi)$$

(2) 振动的速度和加速度
$$v=\frac{dx}{dt}=-A\omega\sin(\omega t+\varphi)$$
$$a=\frac{dv}{dt}=\frac{d^2x}{dt^2}=-\omega^2A\cos(\omega t+\varphi)$$

由初始条件求 $A$、$\varphi$：
$$A=\sqrt{x_0^2+\frac{v_0^2}{\omega^2}}$$, $$\varphi=\arctan\left(-\frac{v_0}{\omega x_0}\right)$$

(3) 旋转矢量
以匀角速度 $\omega$ 沿逆时针方向转动的矢量 $A$ 来分析简谐振动。

**2. 简谐振动的特征**

(1) 运动学特征
$$a=-\omega^2x$$

(2) 动力学特征
$$F=-kx$$

(3) 能量特征
$$E=E_k+E_p=\frac{1}{2}kA^2\sin^2(\omega t+\varphi)+\frac{1}{2}kA^2\cos^2(\omega t+\varphi)=\frac{1}{2}kA^2$$

**3. 简谐振动的合成**

(1) 两个同频率同方向简谐振动的合运动仍是角频率为 $\omega$ 的简谐振动。其振幅 $A$、初相位 $\varphi$ 为
$$A=\sqrt{A_1^2+A_2^2+2A_1A_2\cos(\varphi_2-\varphi_1)}$$
$$\varphi=\arctan\frac{A_1\sin\varphi_1+A_2\sin\varphi_2}{A_1\cos\varphi_1+A_2\cos\varphi_2}$$

(2) 同频率相互垂直的两个简谐振动的合成，一般为椭圆运动。

(3) 具有不同频率的相互垂直的简谐振动的合成比较复杂，当其频率为整数比时，轨迹是闭合的，运动是周期性的，这种图形称为李萨如图形。

**4. 相平面 相空间**

在以速度和位置构建的坐标系中得到的平面或空间，称为相平面或相空间。“相”指物体的运动状态。

**5. 波动的基本概念**

(1) 波线与波面
自波源沿各传播方向画的一些带箭头的线称为波线；介质中振动相位相同的各点组成的曲面称为同相面或波面。在某一时刻，最前面的波面称为波前或波阵面。波线与波面相互垂直。

(2) 波速 $u$、波长 $\lambda$、周期 $T$、频率 $\nu$ 之间的关系
$$u=\frac{\lambda}{T}=\lambda\nu$$

**6. 平面简谐波的波动方程**
$$y=A\cos\left[\omega\left(t-\frac{x}{u}\right)+\varphi\right]$$

**7. 波的传播特性**

(1) 波的叠加原理
几列波可以保持各自的特点通过同一介质，好像没有其他波一样，在它们相互重叠的区域内，每一点的振动都是各列波单独在该点产生的振动的矢量和。

(2) 波的干涉
由频率相同、振动方向相同、相位相同或相位差恒定的两个波源所发出的波，在空间相遇时，出现某些点振动始终加强，某些点振动始终减弱或完全抵消的现象。

### 一、 思考题解答（对应教材第225页）

**7.1 什么是简谐振动？简谐振动中振幅、周期、频率和相位各有什么物理意义？**
*   **思路**：考察简谐振动的定义及基本特征量的物理含义。
*   **解答**：简谐振动是指物体所受的回复力与位移成正比且方向相反（$F=-kx$）的振动，其运动方程为 $x=A\cos(\omega t+\varphi)$。
    *   振幅 $A$：物体离开平衡位置的最大位移的绝对值，反映振动的能量大小。
    *   周期 $T$：完成一次完整振动所需的时间。
    *   频率 $\nu$：单位时间内完成全振动的次数，$\nu = 1/T$。
    *   相位 $(\omega t+\varphi)$：决定振动物体在某一时刻位移和速度的大小和方向的物理量。初相 $\varphi$ 决定初始时刻的运动状态。

**7.2 如何由简谐振动的基本方程来确定做简谐振动的物体的初始运动状态？**
*   **思路**：利用初始条件 $t=0$ 时的位移 $x_0$ 和速度 $v_0$。
*   **解答**：将 $t=0$ 代入运动方程和速度方程，得到：
    $$x_0 = A\cos\varphi$$
    $$v_0 = -A\omega\sin\varphi$$
    联立可解得振幅和初相：
    $$A = \sqrt{x_0^2 + \frac{v_0^2}{\omega^2}}$$
    $$\varphi = \arctan\left(-\frac{v_0}{\omega x_0}\right)$$
    需结合 $x_0$ 和 $v_0$ 的正负符号来确定初相 $\varphi$ 所在的具体象限。

**7.3 如何用旋转矢量法分析简谐振动和振动的合成？**
*   **思路**：利用匀速圆周运动在坐标轴上的投影来模拟简谐振动。
*   **解答**：旋转矢量法是指以一个长度为 $A$ 的矢量绕原点以角速度 $\omega$ 逆时针匀速旋转，该矢量在 $x$ 轴上的投影 $x=A\cos(\omega t+\varphi)$ 即为简谐振动方程。对于振动的合成，将各个分振动的旋转矢量按矢量加法（如平行四边形法则或首尾相接法）求和，所得合矢量在 $x$ 轴上的投影即为合振动。

**7.4 做简谐振动的物体的动能、势能和总能量随时间如何变化？三者之间的关系如何？何时其动能最大，何时其势能最大，何时动能和势能相等？**
*   **思路**：考察简谐振动的能量公式。
*   **解答**：动能 $E_k = \frac{1}{2}mv^2 = \frac{1}{2}kA^2\sin^2(\omega t+\varphi)$，势能 $E_p = \frac{1}{2}kx^2 = \frac{1}{2}kA^2\cos^2(\omega t+\varphi)$。总能量 $E = E_k + E_p = \frac{1}{2}kA^2$，保持不变。
    三者关系：动能和势能相互转化，总能量守恒。
    *   当物体通过平衡位置（$x=0$）时，动能最大，势能为零。
    *   当物体处于最大位移处（$x=\pm A$）时，势能最大，动能为零。
    *   当 $x = \pm \frac{A}{\sqrt{2}}$ 时，动能和势能相等。

**7.5 简述同方向同频率简谐振动合成的特点。**
*   **思路**：考察同方向同频率简谐振动合成的结论。
*   **解答**：合成运动仍然是同方向、同频率的简谐振动。合振幅 $A = \sqrt{A_1^2+A_2^2+2A_1A_2\cos(\varphi_2-\varphi_1)}$。合振动的初相 $\varphi = \arctan\frac{A_1\sin\varphi_1+A_2\sin\varphi_2}{A_1\cos\varphi_1+A_2\cos\varphi_2}$。当相位差为 $2k\pi$ 时合振幅最大（$A_1+A_2$）；当相位差为 $(2k+1)\pi$ 时合振幅最小（$|A_1-A_2|$）。

**7.6 什么是自激振动？它有何特点？**
*   **思路**：区分受迫振动与自激振动。
*   **解答**：自激振动是指系统在恒定能源的作用下，通过系统自身的内部机制产生周期性交变力，从而维持的稳定振动。特点：振动频率由系统本身的固有参数决定，与外部能源的周期无关；振幅由系统的非线性因素限制而趋于稳定。

**7.7 什么是波动？试比较纵波和横波在传播中的差异。**
*   **思路**：考察波的定义及分类。
*   **解答**：波动是振动状态（相位）在介质中的传播过程。
    差异：横波中质点的振动方向与波的传播方向垂直，存在波峰和波谷，只能在固体中传播；纵波中质点的振动方向与波的传播方向平行，存在密部和疏部，可以在固体、液体和气体中传播。

**7.8 什么是简谐波？简谐波的波动方程有哪几种表达形式？简述简谐波波动方程的物理意义。**
*   **思路**：考察波动方程的形式及物理意义。
*   **解答**：简谐波是简谐振动在介质中的传播。
    波动方程的常见形式有：
    $$y = A\cos\left[\omega\left(t - \frac{x}{u}\right) + \varphi\right]$$
    $$y = A\cos\left(\omega t - \frac{2\pi}{\lambda}x + \varphi\right)$$
    物理意义：表示介质中任一质点 $x$ 在任意时刻 $t$ 的位移；描述了波形的传播，体现了时间和空间的周期性。

**7.9 简述波的叠加原理。什么是波的干涉？**
*   **思路**：考察波的独立传播原理和干涉定义。
*   **解答**：波的叠加原理：几列波在同一介质中相遇时，各自保持原有的特性（频率、波长、振动方向等）继续传播，在相遇区域内，任一质点的位移是各列波单独在该点引起的位移的矢量和。
    波的干涉：频率相同、振动方向相同、相位差恒定的两列波相遇时，在空间某些点振动始终加强，某些点振动始终减弱或完全抵消的现象。

**7.10 什么是波程差？利用波程差讨论两列波相干加强、减弱的条件。**
*   **思路**：波程差与相位差的关系。
*   **解答**：波程差 $\Delta r = r_2 - r_1$ 是两列波从各自波源传到空间某点的几何路程之差。
    相位差 $\Delta\varphi = \frac{2\pi}{\lambda}\Delta r + (\varphi_2 - \varphi_1)$。
    当 $\Delta\varphi = 2k\pi$ 时，干涉加强；
    当 $\Delta\varphi = (2k+1)\pi$ 时，干涉减弱。

**7.11 声强级是如何定义的？声强级和人耳感觉的响度的关系如何？**
*   **思路**：考察声学基本物理量。
*   **解答**：声强级 $L = 10\log_{10}\frac{I}{I_0}$，单位为分贝（dB），其中 $I_0 = 10^{-12}$ W/m$^2$ 为基准声强。响度是人耳对声音强弱的主观感受，它不仅与声强级有关，还与声音的频率有关。

**7.12 什么是超声波？简述其在生产实践中的应用。**
*   **思路**：考察超声波的定义和应用。
*   **解答**：超声波是指频率高于 20 kHz 的机械波。应用：无损检测（探伤）、医学成像（B超）、超声波清洗、焊接、测距、测速等。

**7.1 简谐振动的表达式和简谐波的表达式有什么不同？又有什么联系？**
*   **思路**：振动方程描述单质点，波动方程描述多质点。
*   **解答**：不同：简谐振动表达式 $x=A\cos(\omega t+\varphi)$ 描述的是一个质点位移随时间的变化规律；简谐波表达式 $y=A\cos[\omega(t-x/u)+\varphi]$ 描述的是介质中所有质点位移随时间和空间的变化规律。

**7.4 设一简谐振动的振动方程为 $x=0.1\cos\left(20\pi t+\frac{\pi}{4}\right)$ (SI单位)。求：(1) 振动的振幅、频率和初相位；(2) $t=2.0$ s 时的位移、速度。**
*   **思路**：对比标准振动方程 $x=A\cos(\omega t+\varphi)$。
*   **解答**：
    (1) 对比可知：振幅 $A=0.1$ m，角频率 $\omega=20\pi$ rad/s，频率 $\nu=\omega/2\pi=10$ Hz，初相 $\varphi=\pi/4$。
    (2) $t=2.0$ s 时，代入方程：
    位移 $x = 0.1\cos\left(20\pi \times 2 + \frac{\pi}{4}\right) = 0.1\cos\left(\frac{\pi}{4}\right) = 0.1 \times \frac{\sqrt{2}}{2} \approx 0.0707$ m。
    速度 $v = \frac{dx}{dt} = -0.1 \times 20\pi\sin\left(20\pi t + \frac{\pi}{4}\right)$。
    代入 $t=2.0$ s：$v = -2\pi\sin\left(40\pi + \frac{\pi}{4}\right) = -2\pi \times \frac{\sqrt{2}}{2} = -\sqrt{2}\pi \approx -4.44$ m/s。

**7.5 已知物体的振动方程为 $x=0.2\cos 3t$ (SI单位)，试求：(1) 物体离平衡位置 $0.05$ m 时的速度和加速度；(2) $x=0$ 时的速度和加速度。**
*   **思路**：利用 $v = -\omega A\sin(\omega t)$ 和 $a = -\omega^2 x$。
*   **解答**：
    (1) $x=0.05$ m 时，由 $0.2\cos 3t = 0.05$ 得 $\cos 3t = 0.25$，则 $\sin 3t = \pm\frac{\sqrt{15}}{4}$。
    速度 $v = -0.2 \times 3\sin 3t = -0.6 \times \left(\pm\frac{\sqrt{15}}{4}\right) = \mp 0.15\sqrt{15} \approx \mp 0.58$ m/s。
    加速度 $a = -\omega^2 x = -3^2 \times 0.05 = -0.45$ m/s$^2$。
    (2) $x=0$ 时，$\cos 3t = 0$，则 $\sin 3t = \pm 1$。
    速度 $v = -0.6 \times (\pm 1) = \mp 0.6$ m/s。
    加速度 $a = -\omega^2 x = 0$ m/s$^2$。

**7.6 质量为 $20$ g 的小球与轻质弹簧构成弹簧振子，此系统按方程 $x=0.004\cos\left(4\pi t-\frac{\pi}{4}\right)$ (SI单位)振动。求：(1) 振动的角频率、周期、振幅、初相位、速度及加速度的最大值；(2) $t=2.5$ s 时的相位。**
*   **思路**：对比标准方程，利用极值条件求最大值。
*   **解答**：
    (1) 角频率 $\omega = 4\pi$ rad/s；周期 $T = 2\pi/\omega = 0.5$ s；振幅 $A = 0.004$ m；初相 $\varphi = -\pi/4$。
    速度最大值 $v_{max} = A\omega = 0.004 \times 4\pi = 0.016\pi \approx 0.05$ m/s。
    加速度最大值 $a_{max} = A\omega^2 = 0.004 \times (4\pi)^2 = 0.064\pi^2 \approx 0.63$ m/s$^2$。
    (2) $t=2.5$ s 时的相位为 $4\pi \times 2.5 - \frac{\pi}{4} = 10\pi - \frac{\pi}{4} = \frac{39\pi}{4}$（或表示为 $-\frac{\pi}{4}$）。

**7.7 一个 $2\times10^{-3}$ kg 的质点做简谐振动，其振动方程为 $x=6\times10^{-2}\cos\left(5t-\frac{\pi}{2}\right)$ (SI单位)，求：(1) 振幅和周期；(2) 起始位移和速度；(3) $\frac{\pi}{2}$ s 末的位移、速度、加速度、动能和势能。**
*   **思路**：基本振动公式及能量公式 $E_k = \frac{1}{2}mv^2$，$E_p = \frac{1}{2}kx^2$。
*   **解答**：
    (1) 振幅 $A = 6\times10^{-2}$ m，角频率 $\omega=5$ rad/s，周期 $T = 2\pi/5 = 0.4\pi$ s。
    (2) $t=0$ 时，位移 $x_0 = 0.06\cos(-\pi/2) = 0$ m。
    速度 $v_0 = -0.06 \times 5\sin(-\pi/2) = 0.3$ m/s。
    (3) $t=\pi/2$ s 时，相位为 $5 \times \frac{\pi}{2} - \frac{\pi}{2} = 2\pi$。
    位移 $x = 0.06\cos(2\pi) = 0.06$ m。
    速度 $v = -0.3\sin(2\pi) = 0$ m/s。
    加速度 $a = -\omega^2 x = -25 \times 0.06 = -1.5$ m/s$^2$。
    动能 $E_k = \frac{1}{2}mv^2 = 0$ J。
    势能 $E_p = \frac{1}{2}kx^2 = \frac{1}{2}m\omega^2x^2 = \frac{1}{2} \times 0.002 \times 25 \times 0.06^2 = 9\times10^{-5}$ J。

**7.8 两质点在同一直线上做同频率、同振幅的简谐振动，且平衡位置相同。振动过程中，每当它们经过振幅一半的地方时会相遇，而运动方向相反，求它们的相位差，并用旋转矢量表示。**
*   **思路**：利用旋转矢量图分析相遇时的相位状态。
*   **解答**：设两质点振动方程分别为 $x_1=A\cos(\omega t+\varphi_1)$ 和 $x_2=A\cos(\omega t+\varphi_2)$。
    当它们相遇在 $x=A/2$ 时，相位为 $\pm\pi/3$。由于运动方向相反，一个相位为 $\pi/3$（向负方向运动），另一个为 $-\pi/3$（向正方向运动）。
    相位差 $\Delta\varphi = \frac{\pi}{3} - \left(-\frac{\pi}{3}\right) = \frac{2\pi}{3}$。
    旋转矢量表示：两个长度为 $A$ 的矢量，夹角为 $2\pi/3$，在 $x$ 轴上的投影均为 $A/2$，且旋转方向相反（一个向上，一个向下）。

**7.9 有两个同方向同频率的简谐振动：$x_1=0.4\cos(2\pi t+\pi/6)$，$x_2=0.2\cos(2\pi t+\varphi)$ (SI单位)，试问：(1) $\varphi$ 为何值时合振动的振幅最小？求此振幅值；(2) $\varphi$ 为何值时合振动的振幅最大？求此振幅值。**
*   **思路**：合振幅公式 $A = \sqrt{A_1^2+A_2^2+2A_1A_2\cos(\varphi_2-\varphi_1)}$。
*   **解答**：
    (1) 振幅最小条件：相位差 $\varphi - \pi/6 = (2k+1)\pi$，即 $\varphi = \frac{7\pi}{6} + 2k\pi$ ($k=0, \pm1, \dots$)。
    最小振幅 $A_{min} = |A_1 - A_2| = |0.4 - 0.2| = 0.2$。
    (2) 振幅最大条件：相位差 $\varphi - \pi/6 = 2k\pi$，即 $\varphi = \frac{\pi}{6} + 2k\pi$ ($k=0, \pm1, \dots$)。
    最大振幅 $A_{max} = A_1 + A_2 = 0.4 + 0.2 = 0.6$。

**7.10 已知两个同方向同频率的简谐振动的振动方程分别为 $x_1=5\cos\left(10t+\frac{3}{5}\pi\right)$，$x_2=6\cos\left(10t+\frac{1}{5}\pi\right)$ (SI单位)。(1) 求合振动的振幅及初相；(2) 若有另一同方向同频率的简谐振动 $x_3=7\cos(10t+\varphi_3)$ (SI单位)，则 $\varphi_3$ 为多少时，$x_1+x_3$ 的振幅最大？又 $\varphi_3$ 为多少时，$x_2+x_3$ 的振幅最小？**
*   **思路**：矢量合成法或公式法。
*   **解答**：
    (1) 相位差 $\Delta\varphi = \frac{1}{5}\pi - \frac{3}{5}\pi = -\frac{2}{5}\pi$。
    合振幅 $A = \sqrt{5^2 + 6^2 + 2 \times 5 \times 6 \times \cos\left(-\frac{2}{5}\pi\right)} = \sqrt{61 + 60\cos(72^\circ)} \approx \sqrt{61 + 18.54} \approx 8.92$。
    初相 $\tan\varphi = \frac{5\sin(3\pi/5) + 6\sin(\pi/5)}{5\cos(3\pi/5) + 6\cos(\pi/5)} \approx \frac{4.755 + 3.528}{-1.545 + 4.854} \approx 2.503 \implies \varphi \approx 1.19$ rad。
    (2) 使 $x_1+x_3$ 振幅最大：$\varphi_3 - \frac{3}{5}\pi = 2k\pi \implies \varphi_3 = \frac{3}{5}\pi + 2k\pi$。
    使 $x_2+x_3$ 振幅最小：$\varphi_3 - \frac{1}{5}\pi = (2k+1)\pi \implies \varphi_3 = \frac{6}{5}\pi + 2k\pi$。

**7.11 若日光灯电路中灯管两端电压和镇流器两端电压分别为 $u_1=90\sqrt{2}\cos 100\pi t$，$u_2=200\sqrt{2}\cos(100\pi t+\pi/2)$ (SI单位)，试求日光灯电路两端的总电压 $U=u_1+u_2$ 的表达式。**
*   **思路**：同频率简谐振动的合成，利用矢量合成。
*   **解答**：总电压振幅 $U_m = \sqrt{(90\sqrt{2})^2 + (200\sqrt{2})^2} = \sqrt{16200 + 80000} = 10\sqrt{962} \approx 310.16$ V。
    初相 $\varphi$ 满足 $\tan\varphi = \frac{200\sqrt{2}}{90\sqrt{2}} = \frac{20}{9} \implies \varphi \approx 1.15$ rad。
    总电压表达式为 $U = 310.16\cos(100\pi t + 1.15)$ V。

**7.12 当简谐振动在介质中传播时，问：(1) 波源振动的周期（或频率）与波动的周期（或频率）数值是否相同？(2) 波源振动的速度与波速是否相同？**
*   **思路**：区分振动速度与波速的物理概念。
*   **解答**：
    (1) 相同。波动的周期和频率取决于波源的振动周期和频率。
    (2) 不同。波源振动的速度是介质质点做简谐运动的速度，大小和方向随时间周期性变化；波速是振动状态（相位）在介质中的传播速度，由介质的性质决定，在同一种均匀介质中是恒定的。

**7.13 一声波在空气中的波长是 $0.25$ m，速度是 $340$ m/s。当它进入另一个介质时，波长变为 $0.79$ m，求它在这种介质中的传播速度。**
*   **思路**：频率不变。
*   **解答**：波的频率 $\nu = \frac{u_1}{\lambda_1} = \frac{340}{0.25} = 1360$ Hz。
    在新介质中的波速 $u_2 = \nu \lambda_2 = 1360 \times 0.79 = 1074.4$ m/s。

**7.14 在一平板下装有弹簧，平板上放一质量为 $1.0$ kg 的重物。若平板在竖直方向上做简谐振动，周期为 $0.5$ s，振幅为 $0.20$ m，(1) 求平板到最低点时，重物对平板的作用力；(2) 若频率不变，则平板以多大的振幅振动时，重物跳离平板？(3) 若振幅不变，则平板以多大的频率振动时，重物跳离平板？**
*   **思路**：利用牛顿第二定律，重物跳离平板的临界条件是支持力 $N=0$。
*   **解答**：
    (1) 在最低点，加速度向上，$a = A\omega^2 = A\left(\frac{2\pi}{T}\right)^2 = 0.20 \times \left(\frac{2\pi}{0.5}\right)^2 \approx 31.58$ m/s$^2$。
    由牛顿第二定律 $N - mg = ma \implies N = m(g+a) = 1.0 \times (9.8 + 31.58) = 41.38$ N。
    (2) 重物跳离平板的条件是在最高点时支持力为零，即 $mg = mA\omega^2 \implies g = A\left(\frac{2\pi}{T}\right)^2$。
    $A = \frac{gT^2}{4\pi^2} = \frac{9.8 \times 0.5^2}{4\pi^2} \approx 0.062$ m。
    (3) 振幅不变，$g = A(2\pi f)^2 \implies f = \frac{1}{2\pi}\sqrt{\frac{g}{A}} = \frac{1}{2\pi}\sqrt{\frac{9.8}{0.20}} = \frac{7}{2\pi} \approx 1.11$ Hz。

**7.15 某机器内的活塞按正弦方式上下振动，振幅为 $10$ cm。设有一个垫圈遗落在活塞上。假定机器运转速率不断增大，问活塞振动频率达到何值时，垫圈就不再平稳地停留在活塞上面？**
*   **思路**：与7.14同理，临界条件为最高点支持力为零。
*   **解答**：跳离条件 $g = A\omega^2 = A(2\pi f)^2 \implies f = \frac{1}{2\pi}\sqrt{\frac{g}{A}} = \frac{1}{2\pi}\sqrt{\frac{9.8}{0.10}} = \frac{7\sqrt{2}}{2\pi} \approx 1.58$ Hz。

**7.16 波源做简谐振动，其坐标原点的振动表达式为 $y=4.0\times10^{-2}\cos 240\pi t$ (SI单位)。它所形成的波以 $30$ m/s 的速度沿 $x$ 轴正方向传播。试写出简谐波的表达式。**
*   **思路**：利用波动方程标准形式 $y = A\cos[\omega(t - x/u)]$。
*   **解答**：由题意，$A = 4.0\times10^{-2}$ m，$\omega = 240\pi$ rad/s，$u = 30$ m/s。
    波动表达式为 $y = 4.0\times10^{-2}\cos\left[240\pi\left(t - \frac{x}{30}\right)\right]$ m。

**7.17 太平洋上有一洋波速度为 $740$ km/h，波长为 $300$ km。这种洋波的频率是多少？横渡太平洋 $8000$ km 需要多长时间？**
*   **思路**：利用 $u = \lambda\nu$ 和 $t = s/u$。
*   **解答**：频率 $\nu = \frac{u}{\lambda} = \frac{740}{300} \approx 2.47$ h$^{-1}$（或 $6.85\times10^{-4}$ Hz）。
    时间 $t = \frac{s}{u} = \frac{8000}{740} \approx 10.81$ h。

**7.18 一波源做简谐振动，其振幅为 $1$ m，周期为 $0.01$ s，经平衡位置向 $y$ 轴正方向运动时为计时起点，设此振动以 $u=400$ m/s 的速度沿 $x$ 轴正方向传播，求：(1) 波动方程；(2) 距波源为 $16$ m 和 $20$ m 处两质点的振动方程和初相位；(3) 距波源为 $15$ m 和 $16$ m 处的两质点的相位差。**
*   **思路**：先确定初相，再写出波动方程。
*   **解答**：
    (1) 由题意，$A=1$ m，$T=0.01$ s，$\omega = \frac{2\pi}{T} = 200\pi$ rad/s。$t=0$ 时经平衡位置向 $y$ 轴正方向运动，则初相 $\varphi = -\frac{\pi}{2}$。
    波动方程 $y = 1\cos\left[200\pi\left(t - \frac{x}{400}\right) - \frac{\pi}{2}\right]$ m。
    (2) $x=16$ m 处，振动方程 $y = \cos\left[200\pi t - 200\pi\left(\frac{16}{400}\right) - \frac{\pi}{2}\right] = \cos\left(200\pi t - 8\pi - \frac{\pi}{2}\right) = \cos\left(200\pi t - \frac{\pi}{2}\right)$，初相 $-\frac{\pi}{2}$。
    $x=20$ m 处，振动方程 $y = \cos\left[200\pi t - 200\pi\left(\frac{20}{400}\right) - \frac{\pi}{2}\right] = \cos\left(200\pi t - 10\pi - \frac{\pi}{2}\right) = \cos\left(200\pi t - \frac{\pi}{2}\right)$，初相 $-\frac{\pi}{2}$。
    (3) 相位差 $\Delta\varphi = \frac{2\pi}{\lambda}(x_2 - x_1) = \frac{2\pi}{uT}(16 - 15) = \frac{2\pi}{4}(1) = \frac{\pi}{2}$。

**7.19 已知一列平面简谐波的波动方程为 $y=2\cos 2\pi\left(10t-\frac{x}{5}\right)$ (SI单位)。(1) 求 $t=0.25$ s 时距原点最近的波峰的位置；(2) 此波峰在何时通过原点？**
*   **思路**：令相位等于 $2k\pi$ 求波峰位置。
*   **解答**：
    (1) 波峰条件：相位 $20\pi t - \frac{2\pi x}{5} = 2k\pi$。
    当 $t=0.25$ s 时，$5\pi - \frac{2\pi x}{5} = 2k\pi \implies 5 - \frac{2x}{5} = 2k \implies x = 12.5 - 5k$。
    离原点最近（$x>0$），取 $k=2$，得 $x = 2.5$ m。
    (2) 当波峰通过原点 $x=0$ 时，相位 $20\pi t = 2k\pi \implies t = 0.1k$。
    该波峰对应 $k=2$（由 $x=2.5$ 时 $k=2$），故 $t = 0.1 \times 2 = 0.2$ s。

**7.20 两相干波源 $S_1, S_2$ 的振幅相同，频率为 $100$ Hz，相位差为 $\pi$。若 $S_1, S_2$ 两点相距 $20$ m，两波在同一介质中的传播速度为 $800$ m/s，试求 $S_1, S_2$ 连线上因干涉而静止的各点的位置。**
*   **思路**：干涉静止条件为相位差为奇数倍 $\pi$。
*   **解答**：波长 $\lambda = \frac{u}{\nu} = \frac{800}{100} = 8$ m。
    设 $S_1$ 在原点，$S_2$ 在 $x=20$ 处，点 $P$ 距 $S_1$ 为 $x$，距 $S_2$ 为 $20-x$。
    相位差 $\Delta\varphi = \frac{2\pi}{\lambda}(20-x - x) + \pi = \frac{2\pi}{8}(20-2x) + \pi$。
    静止条件：$\Delta\varphi = (2k+1)\pi$。
    $\frac{\pi}{4}(20-2x) = 2k\pi \implies 20 - 2x = 8k \implies x = 10 - 4k$。
    由于 $0 \le x \le 20$，$k$ 可取 $-2, -1, 0, 1, 2$。
    对应静止点的位置为：$x = 2, 6, 10, 14, 18$ m。

**7.21 如图所示，两波源 $A,B$ 振动方向相同，相位相同，相距 $0.07$ m，它们发出的波的频率为 $30$ Hz，波速为 $0.5$ m/s。求在与 $AB$ 连线成 $30^\circ$ 夹角的直线上距 $A$ 处 $3$ m 的 $P$ 处，两列波的相位差。**
*   **思路**：利用余弦定理求路径差，再求相位差。
*   **解答**：波长 $\lambda = \frac{u}{\nu} = \frac{0.5}{30} = \frac{1}{60}$ m。
    在 $\triangle PAB$ 中，由余弦定理：
    $PB = \sqrt{PA^2 + AB^2 - 2 \cdot PA \cdot AB \cos 30^\circ} = \sqrt{3^2 + 0.07^2 - 2 \times 3 \times 0.07 \times \frac{\sqrt{3}}{2}} \approx 2.9396$ m。
    路径差 $\Delta r = PA - PB = 3 - 2.9396 = 0.0604$ m。
    相位差 $\Delta\varphi = \frac{2\pi}{\lambda}\Delta r = 2\pi \times 60 \times 0.0604 \approx 7.248\pi$。

**7.22 如图所示，两相干波源分别在 $P,Q$ 两处，它们相距 $\frac{3}{2}\lambda$。由 $P,Q$ 发出初相相同、频率为 $\nu$、波长为 $\lambda$ 的两列相干波，$R$ 为 $PQ$ 连线上的一点，求：(1) 自 $PQ$ 发出的两列波在 $R$ 处的相位差；(2) 两列波在 $R$ 处干涉时的合振幅。**
*   **思路**：波的干涉条件。
*   **解答**：
    (1) 由于 $R$ 在 $PQ$ 延长线上，路径差 $\Delta r = PR - QR = PQ = \frac{3}{2}\lambda$。
    相位差 $\Delta\varphi = \frac{2\pi}{\lambda}\Delta r = \frac{2\pi}{\lambda} \times \frac{3}{2}\lambda = 3\pi$。
    (2) 合振幅 $A = \sqrt{A_1^2 + A_2^2 + 2A_1A_2\cos(3\pi)} = |A_1 - A_2|$。若两波源振幅相同，则合振幅为 $0$。

**7.23 如图所示，$A,B$ 两点为同一介质中的两相干波源，其振幅皆为 $5$ cm，频率皆为 $100$ Hz，但当 $A$ 点为波峰时，$B$ 点恰为波谷。设介质中的波速为 $10$ m/s，试写出由 $A,B$ 发出的两列波传到 $P$ 点时干涉的结果。**
*   **思路**：计算总相位差。
*   **解答**：波长 $\lambda = \frac{u}{\nu} = \frac{10}{100} = 0.1$ m。
    由图可知 $PA = 15$ m，$PB = 20$ m，路径差 $\Delta r = 20 - 15 = 5$ m。
    $A$ 为波峰时 $B$ 为波谷，说明两波源相位差 $\Delta\varphi_0 = \pi$。
    $P$ 点总相位差 $\Delta\varphi = \frac{2\pi}{\lambda}\Delta r + \Delta\varphi_0 = \frac{2\pi}{0.1} \times 5 + \pi = 100\pi + \pi = 101\pi$。
    因为 $101\pi$ 是 $\pi$ 的奇数倍，所以 $P$ 点干涉相消，合振幅为 $|5 - 5| = 0$ cm。

**7.24 $A,B$ 为两相干波源，相距 $\lambda/4$ ($\lambda$ 为波长)，$A$ 较 $B$ 的相位超前 $\pi/2$。问在 $AB$ 连线上，$A$ 外侧各点的合振幅如何？$B$ 外侧各点的合振幅又如何？**
*   **思路**：分析外侧点的路径差和相位差。
*   **解答**：设波源振幅为 $A_0$。
    (1) 在 $A$ 外侧（$A$ 左侧），点距 $A$ 为 $r_A$，距 $B$ 为 $r_B = r_A + \lambda/4$。
    相位差 $\Delta\varphi = \varphi_A - \varphi_B + \frac{2\pi}{\lambda}(r_B - r_A) = \frac{\pi}{2} + \frac{2\pi}{\lambda}\left(\frac{\lambda}{4}\right) = \pi$。
    干涉相消，合振幅为 $0$。
    (2) 在 $B$ 外侧（$B$ 右侧），点距 $A$ 为 $r_A$，距 $B$ 为 $r_B = r_A - \lambda/4$。
    相位差 $\Delta\varphi = \varphi_A - \varphi_B + \frac{2\pi}{\lambda}(r_B - r_A) = \frac{\pi}{2} + \frac{2\pi}{\lambda}\left(-\frac{\lambda}{4}\right) = 0$。
    干涉加强，合振幅为 $2A_0$。

**7.25 有三个同方向同频率的简谐振动，它们的振幅均为 $A$，相位差依次为 $\frac{2}{3}\pi$，求合振动的振幅。若相位差依次为 $\frac{1}{3}\pi$，合振幅又是多大？（用振幅矢量法计算。）**
*   **思路**：利用旋转矢量合成。
*   **解答**：
    (1) 相位差依次为 $\frac{2}{3}\pi$ 时，三个旋转矢量首尾相接构成一个闭合的等边三角形，合矢量为零，合振幅为 $0$。
    (2) 相位差依次为 $\frac{1}{3}\pi$ 时，三个矢量夹角为 $\frac{\pi}{3}$。
    水平分量：$A + A\cos\frac{\pi}{3} + A\cos\frac{2\pi}{3} = A + \frac{A}{2} - \frac{A}{2} = A$。
    垂直分量：$0 + A\sin\frac{\pi}{3} + A\sin\frac{2\pi}{3} = \frac{\sqrt{3}}{2}A + \frac{\sqrt{3}}{2}A = \sqrt{3}A$。
    合振幅 $A_{total} = \sqrt{A^2 + (\sqrt{3}A)^2} = 2A$。

**7.26 《三国演义》中有张飞喝断当阳桥的故事。设张飞大喝一声的声强级为 $140$ dB，频率为 $400$ Hz。问：(1) 张飞喝声的声压振幅和振幅各是多少？(2) 如果一个士兵喝声的声强级为 $90$ dB，那么张飞一喝相当于多少士兵同时大喝一声？**
*   **思路**：声强级公式 $L = 10\log_{10}(I/I_0)$，声强与声压、振幅的关系。
*   **解答**：
    (1) 由 $L = 10\log_{10}\frac{I}{I_0} = 140 \implies I = I_0 \times 10^{14} = 10^{-12} \times 10^{14} = 100$ W/m$^2$。
    声压振幅 $p_m = \sqrt{2\rho u I}$，取空气密度 $\rho \approx 1.29$ kg/m$^3$，声速 $u \approx 340$ m/s，则 $\rho u \approx 438.6$ kg/(m$^2$s)。
    $p_m = \sqrt{2 \times 438.6 \times 100} \approx 296.2$ Pa。
    位移振幅 $A = \frac{p_m}{\rho u \omega} = \frac{p_m}{\rho u 2\pi\nu} = \frac{296.2}{438.6 \times 2\pi \times 400} \approx 2.69\times10^{-4}$ m。
    (2) 士兵的声强级 $L' = 90$ dB，对应声强 $I' = 10^{-12} \times 10^9 = 10^{-3}$ W/m$^2$。
    人数 $n = \frac{I}{I'} = \frac{100}{10^{-3}} = 100,000$ 人。

# 第7章 振动和波动 思考题+习题解答
> 说明：
> 简谐振动标准式：$x=A\cos(\omega t+\varphi)$
> 角频率：$\omega=\frac{2\pi}{T}=2\pi \nu$
> 平面简谐波沿$\(+x\)$传播：$y=A\cos\left[\omega\left(t-\frac{x}{u}\right)+\varphi_\(\boldsymbol{0}\)\right]$
> 同方向同频率振动合成：$A=\sqrt{A_1^2+A_2^2+2A_1A_2\cos\Delta\varphi}$
> 波的相位差：$\Delta\varphi=\varphi_2-\varphi_1-\frac{2\pi}{\lambda}(r_2-r_1)$
> 声强级：$L_I=1\(\boldsymbol{0}\)\lg\frac{I}{I_\(\boldsymbol{0}\)},\quad I_\(\boldsymbol{0}\)=1\(\boldsymbol{0}\)^{-12}\mathrm{W/m^2}$

---
## 一、思考题解答
### 7.1
**定义**：物体在回复力 $F=-kx$ 作用下的振动称为简谐振动，微分方程 $\displaystyle \frac{d^2x}{dt^2}+\omega^2 \(x=\(\boldsymbol{0}\)\)$，解为 $x=A\cos(\omega t+\varphi)$。
- 振幅$A$：振动物体离开平衡位置的**最大位移大小**，表征振动强弱；
- 周期$T$：完成一次全振动所需时间；
- 频率$\nu$：单位时间内全振动次数，$\nu=\frac1T$；
- 相位$\omega t+\varphi$：描述**任意时刻振子的运动状态（位移、速度）**；$\varphi$为初相位，$\(t=\(\boldsymbol{0}\)\)$时刻的相位。

### 7.2
简谐振动方程 $x=A\cos(\omega t+\varphi)$。
$\(t=\(\boldsymbol{0}\)\)$时：
位移：$x_\(\boldsymbol{0}\)=A\cos\varphi$；速度：$v_\(\boldsymbol{0}\)=-A\omega\sin\varphi$。
由初始位移$x_\(\boldsymbol{0}\)$、初速度$v_\(\boldsymbol{0}\)$，联立求出$A=\sqrt{x_\(\boldsymbol{0}\)^2+\left(\frac{v_\(\boldsymbol{0}\)}{\omega}\right)^2}$和初相$\varphi$，即可确定初始时刻位移、速度，即初始运动状态。

### 7.3
旋转矢量：以角速度$\omega$逆时针旋转的矢量，矢量长度=振幅$A$；矢量与$x$轴夹角=相位$\omega t+\varphi$；矢量端点在x轴投影就是振子位移$x$。
1. 分析振动：矢量转动，投影随时间做余弦振动，直观判断相位、速度方向；
2. 振动合成：同方向同频率简谐振动合成，将两个旋转矢量做矢量相加，合矢量就是合振动的旋转矢量，合矢量的模为合振幅，合矢量初角为合初相。

### 7.4
简谐振动：
势能 $E_p=\frac12 kx^2=\frac12kA^2\cos^2(\omega t+\varphi)$
动能 $E_k=\frac12 mv^2=\frac12kA^2\sin^2(\omega t+\varphi)$
总能量 $E=E_k+E_p=\frac12kA^2=\text{常数}$，**总能量守恒，与时间无关**。
- 最大位移处：$x=\pm A$，势能最大，动能=\(\boldsymbol{0}\)；
- 平衡位置：$\(x=\(\boldsymbol{0}\)\)$，动能最大，势能=\(\boldsymbol{0}\)；
- $|x|=\frac{A}{\sqrt2}$ 时，$E_k=E_p=\frac14kA^2$，动能势能相等。

### 7.5
同方向、同频率简谐振动合成后，**仍然是同频率的简谐振动**。
合振幅：
$$A=\sqrt{A_1^2+A_2^2+2A_1A_2\cos(\varphi_2-\varphi_1)}$$
合初相：
$$\tan\varphi=\frac{A_1\sin\varphi_1+A_2\sin\varphi_2}{A_1\cos\varphi_1+A_2\cos\varphi_2}$$
相位差$\Delta\varphi=\varphi_2-\varphi_1=2k\pi$，合振幅最大 $A=A_1+A_2$；
相位差$\Delta\varphi=(2k+1)\pi$，合振幅最小 $A=|A_1-A_2|$。

### 7.6
**自激振动**：系统依靠自身运动从恒定外部能源获取能量，维持持续振动。
特点：不需要周期性外力驱动；振动频率由系统本身参数决定；存在稳定的振幅；如钟摆、弦乐器。

### 7.7
**波动**：介质中质点在平衡位置附近振动，振动形态、能量由近及远传播；质点本身不随波迁移。
- 横波：质点振动方向$\perp$波传播方向；有波峰、波谷；固体可以传播横波，液体气体不能传播横波。
- 纵波：质点振动方向**平行**波传播方向；有密部、疏部；固、液、气体都可以传播纵波（声波是纵波）。

### 7.8
**简谐波**：波源做简谐振动，介质中各质点都做同频率简谐振动形成的波。
三种表达形式（沿$\(+x\)$）：
$$
\begin{align*}
y&=A\cos\left[\omega\left(t-\frac{x}{u}\right)+\varphi_\(\boldsymbol{0}\)\right]\\
y&=A\cos\left(\omega t-\frac{2\pi x}{\lambda}+\varphi_\(\boldsymbol{0}\)\right)\\
y&=A\cos\left[2\pi\left(\frac{t}{T}-\frac{x}{\lambda}\right)+\varphi_\(\boldsymbol{0}\)\right]
\end{align*}
$$
物理意义：
固定$x=x_\(\boldsymbol{0}\)$：得到$x_\(\boldsymbol{0}\)$处质点振动方程，描述该质点位移随时间变化；
固定$t=t_\(\boldsymbol{0}\)$：得到$t_\(\boldsymbol{0}\)$时刻空间波形，描述同一时刻，不同位置质点位移；
$\(x,t\)$同时变化：描述波形随时间向前传播。

### 7.9
**波的叠加原理**：几列波相遇时，各自保持原有特性独立传播；重叠区域，介质质点位移等于各波单独引起位移矢量之和。
**波的干涉**：**频率相同、振动方向相同、相位差恒定**的两列波相遇，空间出现稳定的强弱分布现象。

### 7.1\(\boldsymbol{0}\)
**波程差$\delta$**：两相干波源到空间某点的路程之差 $\delta=r_2-r_1$。
相位差：$\displaystyle \Delta\varphi=\varphi_2-\varphi_1-\frac{2\pi}{\lambda}\delta$
若两波源初相位相同 $\varphi_1=\varphi_2$：
- 干涉加强：$\delta=k\lambda,\ k=\(\boldsymbol{0}\),\pm1,\pm2\cdots$；
- 干涉减弱：$\delta=(2k+1)\frac{\lambda}{2},\ k=\(\boldsymbol{0}\),\pm1,\pm2\cdots$。

### 7.11
声强级定义：
$$L_I=1\(\boldsymbol{0}\)\lg\frac{I}{I_\(\boldsymbol{0}\)}\quad(\text{单位：dB})$$
$I_\(\boldsymbol{0}\)=1\(\boldsymbol{0}\)^{-12}\mathrm{W/m^2}$，是人耳听阈。
响度是人耳主观感觉；**声强级是客观物理量，响度和声强级不是线性关系，人耳对响度感知是对数关系**；声强级越大，人感觉越响，但响度还和频率有关。

###7.12
**超声波**：频率高于2\(\boldsymbol{0}\)\(\boldsymbol{0}\)\(\boldsymbol{0}\)\(\boldsymbol{0}\)Hz，超出人耳听觉上限的声波。
应用：超声探伤、B超医学成像、超声清洗、超声测距、水下声呐。
