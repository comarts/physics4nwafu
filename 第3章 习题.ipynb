# 本章提要

**1. 表面张力系数**
$$\alpha = \frac{F}{L} = \frac{\Delta A}{\Delta S} = \frac{\Delta E}{\Delta S}$$

**2. 弯曲液面内的压强**

凸液面内的压强：
$$p = p_{\scriptscriptstyle{0}} + p_s$$
凹液面内的压强：
$$p = p_{\scriptscriptstyle{0}} - p_s$$

**3. 球形液面的附加压强**

$$p_s = \frac{2\alpha}{R}$$

**4. 液体在毛细管中上升（或下降）的高度**
$$h = \frac{2\alpha \cos \theta}{\rho g r}$$

# 思考题
## 思考题3.1

液体表面层中的分子受到哪个方向的作用力？表面张力的方向如何？

解题思路：液体表面层分子间距较大，分子间表现为引力。液体内部分子对表面层分子的吸引力大于外部气体分子的吸引力，因此表面层分子受到指向液体内部的合力。表面张力是沿着液体表面切线方向，垂直于分界线的拉力。

解答：

液体表面层分子：受到液体内部分子向内的吸引力大于气体分子向外的吸引力，分子受到指向液体内部的合力。

表面张力方向：沿液体表面切线方向，垂直于分界线，使液面有收缩趋势。

## 思考题3.2

为什么有些昆虫能在水上行走而人不能？

解题思路：昆虫能否在水面行走取决于其重力与水面表面张力提供的支持力的大小关系。表面张力大小与接触周长有关。

解答：

昆虫的体重很小，其脚部与水接触时，水面发生形变产生的表面张力（沿切线方向）的垂直分量足以支撑其重力。而人的体重远大于水的表面张力能提供的最大支持力，因此人无法在水面行走。

## 思考题3.3

吹肥皂泡时，当管的一端不断吹气时，为什么在管的另一端吹出的肥皂泡会慢慢缩小？

解题思路：根据球形液面的附加压强公式 $p_s = \dfrac{4\alpha}{R}$（肥皂泡有内外两个表面）。泡内压强 $p = p_0 + \dfrac{4\alpha}{R}$。半径越小，内部压强越大。

解答：

因为肥皂泡内的压强比外部大气压高，且附加压强与半径成反比。随着泡内气体通过管子向外流出，泡的半径减小，导致泡内压强增大，从而使泡内气体进一步向外排出，最终导致肥皂泡慢慢缩小。

## 思考题3.4

一滴水珠掉在荷叶上，分成许多小水滴，水的总能量发生什么变化？为什么？

解题思路：表面能 $E = \alpha \cdot \Delta S$。水滴分裂后，总体积不变，但总表面积增大。

解答：

一滴大水珠分成很多小水滴，总表面积变大；表面能 $E=\alpha S$，$\alpha$ 不变，表面积 $S$ 增大，总表面能增加。（分裂过程需要外界做功）


## 思考题3.5

将大小不同的两个肥皂泡用玻璃管连通，它们将发生什么变化？为什么？

解题思路：利用附加压强公式 $p = p_0 + \dfrac{4\alpha}{R}$。小泡半径小，内部压强大于大泡内部压强。

🏍️解答：

小肥皂泡不断缩小，大肥皂泡不断变大，直到小泡收缩到只剩一层薄膜贴在管口。
原因：肥皂泡附加压强 $\Delta p=\dfrac{4\alpha}{R}$，半径越小，泡内压强越大；气体从小泡（高压）流向大泡（低压）。


## 思考题3.6

扩大液体表面和拉开一张橡皮膜，都要做功。所做的功转化成了什么能量？

解题思路：做功与能量转化。橡皮膜是弹性形变，液体表面是分子势能的增加。

解答：

扩大液体表面做功，转化为液体的表面能；

拉开橡皮膜做功转化为橡皮膜的弹性势能。二者本质不同：液体表面能来源于分子间作用力，橡皮膜是固体弹性形变。


## 思考题3.7

造成液面弯曲的原因是什么？

解题思路：接触角与润湿现象。液面弯曲是由液体与固体接触处的润湿情况决定的（接触角 $\theta$）。

解答：

造成液面弯曲的原因是液体与固体接触处的附着层内分子受力情况不同，导致液体润湿或不润湿固体（即接触角 $\theta$ 不同）。


## 思考题3.8

为什么雾天多发生在秋冬两季？

解题思路：雾的形成需要水汽达到饱和并凝结。秋冬季夜间地面辐射冷却快，近地面气温容易降至露点以下。

🏍️解答：

雾是小液滴。根据弯曲液面附加压强，小液滴饱和蒸气压比平面液体高；秋冬气温低，饱和蒸气压下降，空气中水汽容易以凝结核为基础凝结成小液滴，形成雾。


## 思考题3.9

何谓接触角？何谓润湿与不润湿？从微观上加以说明。

解题思路：接触角定义及微观解释。

🏍️解答：

在液体与固体接触处，过接触点作液面的切线，该切线与固体表面（经过液体内部）的夹角称为接触角 $\theta$。当 $\theta < 90^\circ$ 时，液体润湿固体；当 $\theta > 90^\circ$ 时，液体不润湿固体。微观上，润湿是附着力大于内聚力（$\theta < 90^\circ$）；不润湿是内聚力大于附着力（$\theta > 90^\circ$）。

## 思考题3.10

何谓气体栓塞现象？该现象是如何产生的？

解题思路：毛细管中气泡阻碍液体流动的现象。

解答：

气体栓塞是指当液体在细管中流动时，如果混入气泡，气泡会阻碍液体流动甚至使其停止的现象。

润湿液体在毛细管，气泡两端液面曲率不同，产生方向相反的附加压强；多个气泡附加压强叠加，阻碍液体流动。输液、输油管要避免气泡。

## 思考题3.11

微观上看，不同弯曲液面(凸、凹)下，分子逃出何种液面较为容易？

解题思路：弯曲液面上的饱和蒸气压。凸液面分子受到的束缚较小，容易逃逸。

解答：

分子从凸液面逃出较为容易。因为凸液面曲率半径 $R>0$，表面层分子受到指向液体内部的合力较小，分子容易脱离液面进入气相。

## 习题3.1

一矩形框被可移动的横杆分成两部分，这两部分分别覆以表面张力系数为 $40\times10^{-3}\ \mathrm{N\cdot m^{-1}}$ 和 $70\times10^{-3}\ \mathrm{N\cdot m^{-1}}$ 的液膜，横杆与框的一边平行，长度为 $0.1 \text{m}$，求横杆所受的力。

解题思路：横杆受到两侧液膜的表面张力。液膜有两个表面。

🏍️解答：

左侧拉力 $F_1 = 2\alpha_1 L = 2 \times 40\times 10^{-3} \times 0.1 = 8\times 10^{-3} \text{ N}$

右侧拉力 $F_2 = 2\alpha_2 L = 2 \times 70\times 10^{-3} \times 0.1 = 14\times 10^{-3} \text{ N}$

合力 $F = F_2 - F_1 = 6\times 10^{-3} \text{ N}$，方向指向表面张力系数较大的一侧。

## 习题3.2

有一根毛细管，完全润湿毛细管管壁且水能在管内升高到 $h$。假如把这根毛细管插入水中，露出水面的高度为 $h/2$，你能不能由此得到人造喷泉？为什么？水在管中上升的高度是多少，这时接触角是多大？

解题思路：毛细现象。水完全润湿管壁，接触角 $\theta = 0^\circ$。液柱上升高度取决于附加压强与静水压强平衡。

解答：

不能形成喷泉。

液体上升到管口后，液面曲率半径自动变大，$\cos\theta$ 减小，附加压强降低，平衡重力。

液面停在管口，上升高度就是 $(h/2)$。

设原来 $\theta_0=0,\cos\theta_0=1,\ h=\dfrac{2\alpha}{\rho g r}$

现在：$\dfrac h2=\dfrac{2\alpha\cos\theta}{\rho g r}=h\cos\theta\implies \cos\theta=\dfrac12, \theta=60^\circ$

## 习题3.4

等温地吹出一个直径为$d=5.0\times 10^{-2}\ \text{m}$ 的肥皂泡需做多少功？这时肥皂泡内外压强差是多少？已知肥皂泡的表面张力系数为$\alpha=2.5\times 10^{-2}\ \mathrm{N\cdot m^{-1}}$。

## 📌习题3.5

一个人先吹起了一个半径为 $r_0$ 的肥皂泡，接着他又在刚才那个肥皂泡内吹起了一个半径为 $r_1$ 的肥皂泡。

(1) 试求此时外面那个肥皂泡的半径与$r_0、r_1$ 之间的关系；

(2) 此人在吹第二个肥皂泡时所做的功为多少？假设吹肥皂泡的过程很缓慢，周围环境的压强为$p_0$，温度为$T_0$，肥皂泡表面张力系数为$\alpha$。

解题思路：(1) 质量守恒，气体的 $pV$ 乘积近似守恒（等温过程）。(2) 做功等于表面能变化量。

解答：

设最终外面肥皂泡的半径为 $R$

(1) 外面肥皂泡半径 $R$ 满足的关系

肥皂泡有内外两个表面，所以泡内压强比泡外高
$$\Delta p=\frac{4\alpha}{r}$$

初始外面肥皂泡内空气压强为
$$p_0+\frac{4\alpha}{r_0}$$
体积为
$$V_0=\frac43\pi r_0^3$$

吹入内层肥皂泡后，原来外面泡内的空气被挤到两泡之间的夹层中。夹层中空气压强为
$$p_0+\frac{4\alpha}{R}$$
体积为
$$V=\frac43\pi\bigl(R^3-r_1^3\bigr)$$

整个过程缓慢，温度不变，夹层中原来外面泡内空气的物质的量不变，因此由理想气体状态方程：
$$\left(p_0+\frac{4\alpha}{r_0}\right)r_0^3=\left(p_0+\frac{4\alpha}{R}\right)\bigl(R^3-r_1^3\bigr)$$

所以外面肥皂泡最终半径 $R$ 满足
$$\left(p_0+\frac{4\alpha}{r_0}\right)r_0^3=
\left(p_0+\frac{4\alpha}{R}\right)\bigl(R^3-r_1^3\bigr)$$
解此方程即可得到 $R$。

(2) 吹第二个肥皂泡时所做的功

肥皂泡有两个表面，表面能为
$$E=2\cdot 4\pi r^2\alpha=8\pi\alpha r^2$$

初始外面泡表面能为
$$E_0=8\pi\alpha r_0^2$$

最终外面泡表面能为
$$E_R=8\pi\alpha R^2$$
内层泡表面能为
$$E_1=8\pi\alpha r_1^2$$

缓慢等温吹气时，理想气体内能不变，大气压的体积功相互抵消，人做的净功等于肥皂膜表面能的增加量，因此
$$W=\Delta E=8\pi\alpha R^2+8\pi\alpha r_1^2-8\pi\alpha r_0^2$$

即
$$W=8\pi\alpha\bigl(R^2+r_1^2-r_0^2\bigr)$$
其中 $R$ 由第 (1) 问中的方程确定。

## 习题3.6

如图所示，在水中平行而竖直地插入两块间距为 $d=0.5\ \text{mm}$ 玻璃板，求板间水面上升的高度 $h$。已知接触角 $\theta=0$，水的 $\alpha=7.3\times10^{-2}\ \mathrm{N\cdot m^{-1}}$。

解题思路：两平行玻璃板间的毛细上升高度。

$

## 习题3.7

一根内直径为 $1.0\ \text{mm}$ 上端封口的玻璃管，竖直插入水银中，平衡后，管内气压比大气压低 $30\times10^{2}\ \text{Pa}$，求水银在管内变化的高度。（接触角 $140^\circ$，$\alpha=47\times10^{-2}\ mathrm{N\cdot m^{-1}}$）

解题思路：

用到**弯曲液面附加压强定义**：弯曲液面两侧压强差 $p_{\text{气}}-p_{\text{液}}=\dfrac{2\alpha\cos\theta}{r}$；**流体静力学压强公式**：高度差$h$对应的静压强$\Delta p=\rho g h$。设管内水银面比管外高$h$（向上为正），写出管内气体压强、管内液面下液体压强，联立附加压强方程求解$h$。

解答：

已知：
$d=1.0\ \text{mm}$，$r=\dfrac{d}{2}=5.0\times10^{-4}\ \text{m}$，
$p_0-p_i=30\times10^2\ \text{Pa}=3.0\times10^3\ \text{Pa}$，
$\theta=140^\circ$，$\alpha=47\times10^{-2}\ \mathrm{N\cdot m^{-1}}$，
$\rho=13.6\times10^3\ \mathrm{kg/m^3}$，$g=9.8\ \mathrm{m/s^2}$，
$\cos140^\circ\approx -0.7660$。

管内气体压强：
$$p_i=p_0-3.0\times10^3$$
管内液面处液体压强：
$$p_l=p_0-\rho g h$$
附加压强关系：
$$p_i-p_l=\frac{2\alpha\cos\theta}{r}$$

代入数值计算右端：
$$
p_i-p_l=\frac{2\times 47\times10^{-2}\times(-0.7660)}{5.0\times10^{-4}}\approx -1.44\times10^3\ \text{Pa}
$$

将$p_i,p_l$代入附加压强式：
$$
(p_0-3.0\times10^3)-(p_0-\rho g h)=-1.44\times10^3
$$
化简：
$$
\rho g h -3.0\times10^3 = -1.44\times10^3
$$
$$
\rho g h = 1.56\times10^3
$$

$$
h=\frac{1.56\times10^3}{\rho g}=\frac{1.56\times10^3}{13.6\times10^3\times9.8}\approx 1.17\times10^{-2}\ \text{m}
$$

## 习题3.8

有一株高 $5\ \text{m}$ 的树，外层木质管子(树液输管)为均匀毛细管，半径为$0.02\ \mathrm{mm}$。设树液的表面张力系数为$\alpha=0.05\ \mathrm{N\cdot m^{-1}}$，接触角为$45^\circ$，问树的根部最小压强为多少时，才能使树液上升到树的顶端？(树液的密度近似取水的密度。)

解题思路：

用到**弯曲液面附加压强公式** $p_s=\dfrac{2\alpha\cos\theta}{r}$ 和**流体静力学压强公式** $\Delta p=\rho g h$。
树液上升依靠毛细管附加压强克服树液柱重力压强；根部压强与顶端液面压强的差值，需要支撑高度$h$的液柱并提供毛细附加压强。取顶端液面处压强为大气压$p_0$，求根部最小压强$p_{\text{根}}$。

解答：

已知：
$h=5\ \text{m},\ r=0.02\ \text{mm}=0.02\times10^{-3}\ \text{m}=2\times10^{-5}\ \text{m}$，
$\alpha=0.05\ \mathrm{N\cdot m^{-1}},\ \theta=45^\circ,\ \cos45^\circ=\dfrac{\sqrt{2}}{2}\approx0.7071$，
树液密度 $\rho=1.0\times10^3\ \mathrm{kg/m^3},\ g=9.8\ \mathrm{m/s^2}$。

顶端毛细管液面附加压强：液面内侧（液体）压强比外侧气体压强低
$$
p_0-p_{\text{顶液}}=\frac{2\alpha\cos\theta}{r}
$$
$$
p_{\text{顶液}}=p_0-\frac{2\alpha\cos\theta}{r}
$$

静力学关系：根部压强 = 顶端液体压强 + 液柱压强
$$
p_{\text{根}} = p_{\text{顶液}}+\rho g h
$$
联立得：
$$
p_{\text{根}}=p_0-\frac{2\alpha\cos\theta}{r}+\rho g h
$$

代入数值计算毛细项：
$$
\frac{2\alpha\cos\theta}{r}=\frac{2\times0.05\times0.7071}{2\times10^{-5}}=3535.5\ \text{Pa}
$$

液柱压强：
$$
\rho g h=1.0\times10^3\times9.8\times5=49000\ \text{Pa}
$$

$$
p_{\text{根}}=p_0 - 3535.5 + 49000 = p_0+45464.5\ \text{Pa}
$$

相对于大气压的**根部超出压强（表压）**：
$$
\Delta p_{\text{根}}=4.55\times10^4\ \text{Pa}
$$

## 习题3.9

在面积为$50\ \mathrm{km^2}$的湖面上，下了一场降雨量为$50\ \text{mm}$的大雨，雨滴半径$r=1.0\ \text{mm}$，设温度不变，水的表面张力系数为$\alpha=7.3\times10^{-2}\ \mathrm{N\cdot m^{-1}}$，求释放出来的能量。

解题思路：雨水总体积$V=S_0\cdot H$；分裂成$N$个小液滴，总表面积减去原来湖面面积，表面能变化。

解答：

总降雨体积 $V = S \cdot d = 50\times 10^6 \times 50\times 10^{-3} = 2.5\times 10^6 \text{ m}^3$

雨滴体积 $v=\frac43\pi r^3,\ r=10^{-3}\mathrm{m}$

$N=\frac{V}{v},\quad S_{\text{总液滴}}=N\cdot4\pi r^2=\dfrac{3V}{r}$

原来水面面积$S_0$远小于液滴总表面积，可以忽略。

释放能量 $E = \alpha \cdot S_{总} = \alpha \cdot \dfrac{3V}{r} = 7.3\times 10^{-2} \times \dfrac{3 \times 2.5\times 10^6}{1.0\times 10^{-3}} = 5.475\times 10^8 \text{ J}$

## 习题3.10

深为 $2.0\ \text{m}$ 的 水池底部产生的直径为 $d=5.0\times10^{-5}\ \text{m}$ 的气泡，等温上升到水面上时，直径为多大？水的表面张力系数为$\alpha=7.3\times10^{-2}\ \mathrm{N\cdot m^{-1}}$。

解题思路:

用到：**玻意耳定律（等温过程理想气体）**，**弯曲液面附加压强公式**。
气泡内部压强 = 外部液体静压强 + 附加压强；球形气泡（液膜，内部气体，外侧水）附加压强：$\Delta p=\frac{4\alpha}{d}$。
- 池底：深度 $h$，液体静压强 $p_0+\rho g h$；气泡内压强 $p_1 = p_0+\rho g h+\dfrac{4\alpha}{d_1}$
- 水面：外部压强为大气压 $p_0$；气泡内压强 $p_2 = p_0+\dfrac{4\alpha}{d_2}$
等温过程：$p_1 V_1=p_2 V_2$；球体积 $V=\dfrac{4}{3}\pi\left(\dfrac{d}{2}\right)^3=\dfrac{\pi d^3}{6}$。


解答：

等温过程 $p_1 V_1=p_2 V_2$，代入球体体积：
$$
p_1 \cdot \frac{\pi d_1^3}{6}=p_2 \cdot \frac{\pi d_2^3}{6} \implies p_1 d_1^3 = p_2 d_2^3
$$
$$
d_2 = d_1 \cdot \sqrt[3]{\frac{p_1}{p_2}}
$$

计算池底气泡内压强 $p_1$：
$$
\begin{aligned}
p_1 &= p_0+\rho g h+\frac{4\alpha}{d_1}\\
&=1.013\times10^5 + 1000\times9.8\times2.0 + \frac{4\times 7.3\times10^{-2}}{5.0\times10^{-5}}\\
&=1.013\times10^5 + 19600 + 5840\\
&=126740\ \text{Pa}
\end{aligned}
$$

水面处气泡内压强 $p_2$：
$$
\begin{aligned}
p_2 &= p_0+\frac{4\alpha}{d_2}
\end{aligned}
$$
> 注意：这里方程含有 $d_2$，是隐式方程。
> 先观察：附加压强相比大气压很小，可先做零级近似，忽略表面张力项估算 $d_2$，再校验；也可直接数值求解。

先做**近似估算（暂忽略表面张力项）**：
$p_1 \approx p_0+\rho g h = 120900\ \text{Pa},\ p_2\approx p_0=1.013\times10^5\ \text{Pa}$
$$
d_2 \approx d_1 \sqrt[3]{\frac{p_0+\rho g h}{p_0}}
=5.0\times10^{-5}\cdot \sqrt[3]{\frac{120900}{101300}}
\approx5.0\times10^{-5}\cdot \sqrt[3]{1.1935}\approx5.0\times10^{-5}\times1.0607 \approx 5.30\times10^{-5}\ \text{m}
$$

现在把附加压强纳入完整方程：
$$
\left(p_0+\rho g h+\frac{4\alpha}{d_1}\right) d_1^3
=\left(p_0+\frac{4\alpha}{d_2}\right) d_2^3
$$
代入数值：
$$
126740 \times (5.0\times10^{-5})^3
=\left(1.013\times10^5+\frac{0.292}{d_2}\right) d_2^3
$$
$$
1.58425\times10^{-8}=1.013\times10^5\,d_2^3 + 0.292\,d_2^2
$$

解得 $d_2\approx 5.29\times10^{-5}\ \text{m}$





## 习题3.11

半径为 $1.0\times10^{-2}\ \text{m}$ 的球形泡在压强为 $1.136\times10^{5}\ \text{Pa}$ 的大气中吹成，若泡膜的表面张力系数为$\alpha=5.0\times10^{-2}\ \mathrm{N\cdot m^{-1}}$，问周围的大气压强为多大，才可能使泡的半径增加为 $2.0\times10^{-2}\ \text{m}$？设这种变化是在等温情况下进行的。

解题思路：

球形泡膜有内外两个表面，因此表面张力产生的附加压强为  
$$
\Delta p=\frac{4\alpha}{r}
$$
泡内气体压强为  
$$
p_{\text{内}}=p_{\text{外}}+\frac{4\alpha}{r}
$$

等温变化中，泡内气体满足玻意耳定律：  

$$p_{\text{内}}V=\text{常量}$$

球形泡体积为  

$$V=\frac{4}{3}\pi r^3$$

所以  

$$p_{1\text{内}}V_1=p_{2\text{内}}V_2$$

即  

$$\left(p_1+\frac{4\alpha}{r_1}\right)r_1^3=
\left(p_2+\frac{4\alpha}{r_2}\right)r_2^3$$

解答：

已知  
$$
r_1=1.0\times10^{-2}\ \text{m}
$$
$$
r_2=2.0\times10^{-2}\ \text{m}
$$
$$
p_1=1.136\times10^5\ \text{Pa}
$$
$$
\alpha=5.0\times10^{-2}\ \text{N/m}
$$

先计算  
$$
4\alpha=4\times5.0\times10^{-2}=0.20\ \text{N/m}
$$

初始附加压强  
$$\frac{4\alpha}{r_1}=\frac{0.20}{1.0\times10^{-2}}=20\ \text{Pa}$$

最终附加压强  
$$\frac{4\alpha}{r_2}=\frac{0.20}{2.0\times10^{-2}}=10\ \text{Pa}$$

由等温过程  

$$\left(p_1+\frac{4\alpha}{r_1}\right)r_1^3=
\left(p_2+\frac{4\alpha}{r_2}\right)r_2^3$$

得  

$$p_2=
\left(p_1+\frac{4\alpha}{r_1}\right)
\left(\frac{r_1}{r_2}\right)^3-
\frac{4\alpha}{r_2}$$

代入数据  

$$p_2=\left(1.136\times10^5+20\right)
\left(\frac{1.0\times10^{-2}}{2.0\times10^{-2}}\right)^3-10$$

$$p_2=1.1362\times10^5\times\frac{1}{8}-10$$

$$p_2=1.42025\times10^4-10=1.41925\times10^4\ \text{Pa}$$

水和油边界的表面张力系数为 $\alpha=18\times10^{-3}\ \mathrm{N\cdot m^{-1}}$，为了使质量为 $0.2\times10^{-3}\ \text{kg}$ 的油在水内散布成小油滴(当作球形)，需做的功为 $24\times10^{-2}\ \text{J}$，问小油滴的半径有多大？散布过程可认为是等温的，油的密度 $\rho=900\ \mathrm{kg\cdot m^{-3}}$。

    $W = \alpha \cdot \frac{3V}{r} \Rightarrow r = \frac{3\alpha V}{W} = \frac{3 \times 18\times 10^{-3} \times 2.22\times 10^{-7}}{24\times 10^{-2}} \approx 5.0\times 10^{-7} \text{ m}$。

> $$
> W=\alpha\frac{3V}{r}=\frac{3\alpha m}{\rho r}
> $$
> $$
> r=\frac{3\alpha m}{\rho W}=\frac{3\times18\times10^{-3}\times0.2\times10^{-3}}{900\times 24\times10^{-2}}=5\times10^{-8}\mathrm{m}
> $$
> **答案：$r=5\times10^{-8}\mathrm{m}$**

## ❗习题3.23

一根横截面是圆形的管子，外半径是 $0.14\ \text{cm}$，一端封闭，在管的封闭端加一重物，并使管的封闭端竖直向下地浮在水里。管和重物的总质量为 $0.2\ \text{g}$，如果接触角为零，问管的封闭端低于水面多少？已知 $\alpha_水=7.3\times10^{-2}\ \mathrm{N\cdot m^{-1}}$。

解题思路：管漂浮，受力平衡。浮力 = 重力 + 表面张力向上的分量。

解答：设管子浸入水中的深度为 $h$。排开水的体积 $V = \pi R^2 h$。

    浮力 $F_b = \rho g \pi R^2 h$。
    表面张力 $F_s = 2\pi R \alpha \cos\theta = 2\pi R \alpha$。
    平衡方程：$F_b = mg + F_s \Rightarrow \rho g \pi R^2 h = mg + 2\pi R \alpha$。
    $h = \frac{mg + 2\pi R \alpha}{\rho g \pi R^2}$。
    代入数据：$m = 0.2 \text{ g} = 2\times 10^{-4} \text{ kg}$，$R = 0.14 \text{ cm} = 1.4\times 10^{-3} \text{ m}$，$\alpha = 7.3\times 10^{-2} \text{ N/m}$。
    计算得 $h \approx 0.11 \text{ m} = 11 \text{ cm}$。

> 圆管外半径$R=0.14\mathrm{cm}$；封闭端向下浮在水中；总质量$m=0.2\mathrm{g}$；$\theta=0$；$\alpha=7.3\times10^{-2}\mathrm{N/m}$。求封闭端低于水面深度。
> 受力平衡：重力=浮力+表面张力向下拉力（润湿，表面张力拉管子向下）
> $$
> mg+\alpha\cdot 2\pi R=\rho g \cdot \pi R^2 h
> $$
> $$
> h=\frac{mg+2\pi R\alpha}{\rho g\pi R^2}
> $$
> 代入数值：$m=0.2\times10^{-3}\mathrm{kg},\ R=0.14\times10^{-2}\mathrm{m}$
> $$
> mg=1.96\times10^{-3}\mathrm{N},\quad 2\pi R\alpha\approx6.41\times10^{-4}\mathrm{N}
> $$
> $$
> h=\frac{1.96\times10^{-3}+6.41\times10^{-4}}{1000\times9.8\times\pi\times(0.14\times10^{-2})^2}\approx 0.428\mathrm{m}=42.8\mathrm{\ cm}
> $$
