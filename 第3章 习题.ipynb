## 思考题
### 思考题3.1

液体表面层中的分子受到哪个方向的作用力？表面张力的方向如何？

🏍️解答：

液体表面层分子：受到液体内部分子向内的吸引力大于气体分子向外的吸引力，分子受到指向液体内部的合力。

表面张力方向：沿液体表面切线方向，垂直于分界线，使液面有收缩趋势。

### 思考题3.2

为什么有些昆虫能在水上行走而人不能？

🏍️解答：

昆虫体重很小，脚压在水面，表面张力的向上分力可以平衡昆虫重力；
人的体重巨大，重力远大于水表面张力能提供的最大支撑力，所以人不能在水上行走。

### 思考题3.3

吹肥皂泡时，当管的一端不断吹气时，为什么在管的另一端吹出的肥皂泡会慢慢缩小？

🏍️解答：

肥皂泡有内外两层液膜，附加压强 $\Delta p=\dfrac{4\alpha}{R}$，泡内部压强 = 外界大气压 $+\dfrac{4\alpha}{R}$。
停止吹气时，泡内压强大于管外大气压，气体就从泡内沿管子流出，肥皂泡体积缩小。


### 思考题3.4

一滴水珠掉在荷叶上，分成许多小水滴，水的总能量发生什么变化？为什么？

🏍️解答：

总表面能**增大**。

一滴大水珠分成很多小水滴，总表面积变大；表面能 $E=\alpha S$，$\alpha$ 不变，表面积 $S$ 增大，总表面能增加。（分裂过程需要外界做功）

### 思考题3.5

将大小不同的两个肥皂泡用玻璃管连通，它们将发生什么变化？为什么？

🏍️解答：

小肥皂泡不断缩小，大肥皂泡不断变大，直到小泡收缩到只剩一层薄膜贴在管口。
原因：肥皂泡附加压强 $\Delta p=\dfrac{4\alpha}{R}$，半径越小，泡内压强越大；气体从小泡（高压）流向大泡（低压）。

### 思考题3.6

扩大液体表面和拉开一张橡皮膜，都要做功。所做的功转化成了什么能量？

🏍️解答：

扩大液体表面做功，转化为液体的表面能；

拉开橡皮膜做功转化为橡皮膜的弹性势能。二者本质不同：液体表面能来源于分子间作用力，橡皮膜是固体弹性形变。

### 思考题3.7

造成液面弯曲的原因是什么？

🏍️解答：

液体表面层分子受力不对称，存在表面张力；当固体和液体分子间附着力与液体内聚力不相等时，表面张力使液面发生弯曲。

### 思考题3.8

为什么雾天多发生在秋冬两季？

🏍️解答：

雾是小液滴。根据弯曲液面附加压强，小液滴饱和蒸气压比平面液体高；秋冬气温低，饱和蒸气压下降，空气中水汽容易以凝结核为基础凝结成小液滴，形成雾。


### 思考题3.9

何谓接触角？何谓润湿与不润湿？从微观上加以说明。

🏍️解答：

- 接触角$\theta$：固‑液接触面的切线与液体表面切线，在液体内部测量的夹角。
- 润湿：$\theta< 90^\circ$；不润湿：$\theta>90^\circ$。

微观：附着力（固‑液分子引力）>内聚力（液‑液分子引力）→润湿；附着力<内聚力→不润湿。

### 思考题3.10

何谓气体栓塞现象？该现象是如何产生的？

🏍️解答：

气体栓塞：液体细管中，气泡阻碍液体流动的现象。

产生：润湿液体在毛细管，气泡两端液面曲率不同，产生方向相反的附加压强；多个气泡附加压强叠加，阻碍液体流动。输液、输油管要避免气泡。

### 思考题3.11

微观上看，不同弯曲液面(凸\(r_0\(a_1\(r_1、r_2\)a_2\)r_1\)凹)下，分子逃出何种液面较为容易？

🏍️解答：

凸液面分子更容易逃出。

凸液面分子周围分子少，受到的向内吸引力小，束缚弱，更容易脱离液面蒸发；凹液面分子受到向内的引力大，难逃出。

### 思考题解答

**3.1 液体表面层中的分子受到哪个方向的作用力？表面张力的方向如何？**
*   **解题思路**：液体表面层分子间距较大，分子间表现为引力。液体内部分子对表面层分子的吸引力大于外部气体分子的吸引力，因此表面层分子受到指向液体内部的合力。表面张力是沿着液体表面切线方向，垂直于分界线的拉力。
*   **答案**：液体表面层中的分子受到指向液体内部的合力。表面张力的方向沿液体表面切线方向，且垂直于分界线。

**3.2 为什么有些昆虫能在水面上行走而人不能？**
*   **解题思路**：昆虫能否在水面行走取决于其重力与水面表面张力提供的支持力的大小关系。表面张力大小与接触周长有关。
*   **答案**：昆虫的体重很小，其脚部与水接触时，水面发生形变产生的表面张力（沿切线方向）的垂直分量足以支撑其重力。而人的体重远大于水的表面张力能提供的最大支持力，因此人无法在水面行走。

**3.3 吹肥皂泡时，当管的一端不断吹气时，为什么在管的另一端吹出的肥皂泡会慢慢缩小？**
*   **解题思路**：根据球形液面的附加压强公式 $p_s = \frac{4\alpha}{R}$（肥皂泡有内外两个表面）。泡内压强 $p = p_0 + \frac{4\alpha}{R}$。半径越小，内部压强越大。
*   **答案**：因为肥皂泡内的压强比外部大气压高，且附加压强与半径成反比。随着泡内气体通过管子向外流出，泡的半径减小，导致泡内压强增大，从而使泡内气体进一步向外排出，最终导致肥皂泡慢慢缩小。

**3.4 一滴水掉在荷叶上，分成许多小水滴，水的总能量发生什么变化？为什么？**
*   **解题思路**：表面能 $E = \alpha \cdot \Delta S$。水滴分裂后，总体积不变，但总表面积增大。
*   **答案**：水的总能量增加（表面能增加）。因为水滴分裂成许多小水滴后，总表面积增大，而表面张力系数 $\alpha$ 不变，所以表面能 $E = \alpha \Delta S$ 增加。

**3.5 将大小不同的两个肥皂泡用玻璃管连通，它们将发生什么变化？为什么？**
*   **解题思路**：利用附加压强公式 $p = p_0 + \frac{4\alpha}{R}$。小泡半径小，内部压强大于大泡内部压强。
*   **答案**：小肥皂泡会变得更小，大肥皂泡会变得更大。因为小泡内压强大于大泡内压强，气体将从压强高的小泡流向压强低的大泡，导致小泡收缩，大泡膨胀。

**3.6 扩大液体表面和拉开一张橡皮膜，都要做功。所做的功转化成了什么能量？**
*   **解题思路**：做功与能量转化。橡皮膜是弹性形变，液体表面是分子势能的增加。
*   **答案**：拉开橡皮膜所做的功转化为橡皮膜的弹性势能；扩大液体表面所做的功转化为液体的表面能（分子势能）。

**3.7 造成液面弯曲的原因是什么？**
*   **解题思路**：接触角与润湿现象。液面弯曲是由液体与固体接触处的润湿情况决定的（接触角 $\theta$）。
*   **答案**：造成液面弯曲的原因是液体与固体接触处的附着层内分子受力情况不同，导致液体润湿或不润湿固体（即接触角 $\theta$ 不同）。

**3.8 为什么雾天多发生在秋、冬两季？**
*   **解题思路**：雾的形成需要水汽达到饱和并凝结。秋冬季夜间地面辐射冷却快，近地面气温容易降至露点以下。
*   **答案**：秋、冬两季昼夜温差大，夜间地面辐射冷却强烈，近地面空气温度容易降到露点以下，使得空气中的水蒸气达到饱和并凝结成小水滴，形成雾。

**3.9 何谓接触角？何谓润湿与不润湿？从微观上加以说明。**
*   **解题思路**：接触角定义及微观解释。
*   **答案**：在液体与固体接触处，过接触点作液面的切线，该切线与固体表面（经过液体内部）的夹角称为接触角 $\theta$。当 $\theta < 90^\circ$ 时，液体润湿固体；当 $\theta > 90^\circ$ 时，液体不润湿固体。微观上，润湿是附着力大于内聚力（$\theta < 90^\circ$）；不润湿是内聚力大于附着力（$\theta > 90^\circ$）。

**3.10 何谓气体栓塞现象？该现象是如何产生的？**
*   **解题思路**：毛细管中气泡阻碍液体流动的现象。
*   **答案**：气体栓塞是指当液体在细管中流动时，如果混入气泡，气泡会阻碍液体流动甚至使其停止的现象。这是因为气泡在细管中形成弯曲液面，产生附加压强，且由于管径或气泡大小不均匀，导致气泡两端压强差抵消了外部压强差。

**3.11 微观上看，不同弯曲液面（凸、凹）下，分子逃出何种液面较为容易？**
*   **解题思路**：弯曲液面上的饱和蒸气压。凸液面分子受到的束缚较小，容易逃逸。
*   **答案**：分子从凸液面逃出较为容易。因为凸液面曲率半径 $R>0$，表面层分子受到指向液体内部的合力较小，分子容易脱离液面进入气相。

# 第3章 液体的表面现象 思考题+习题解答

> 
> 本章核心公式：
> 表面张力系数：$\alpha=\frac{F}{L}=\frac{\Delta A}{\Delta S}=\frac{\Delta E}{\Delta S}$
> 球面附加压强：$p_s=\frac{2\alpha}{R}$（液滴，单液面）；肥皂泡 $p_s=\frac{4\alpha}{R}$（内外两个液膜表面）
> 毛细管上升：$h=\frac{2\alpha\cos\theta}{\rho g r}$

---

## 思考题解答

### 3.1 液体表面层中的分子受到哪个方向的作用力？表面张力的方向如何？

**思路**：区分分子受力（引力合力）与宏观表面张力。
液体表面层分子，受到内部分子吸引力 > 外部气体分子吸引力，**分子受到指向液体内部的合力**。
表面张力是液面内的宏观拉力：**方向沿液面切线方向，垂直于所取的线段，指向液面收缩方向**。

### 3.2 为什么有些昆虫能在水上行走而人不能？

**思路**：表面张力大小与物体重力对比。
水的表面张力可以支撑质量很小的昆虫。昆虫脚压在水面，使水面下凹，表面张力沿液面切线向上，合力平衡昆虫重力。
人的重力远大于水面能提供的最大表面张力，人会破坏液面浸入水中。

### 3.3 吹肥皂泡时，当管的一端不断吹气时，为什么在管的另一端吹出的肥皂泡会慢慢缩小？

**思路**：附加压强，肥皂泡内压强 $p=p_0+\frac{4\alpha}{R}$，半径越小，内部压强越大。
连通的两端，一端大泡，一端小泡；小泡半径小，内部气压更大。气体从小泡压向大泡，小泡不断缩小，大泡变大。

### 3.4 一滴水掉在荷叶上，分成许多小水滴，水的总能量发生什么变化？为什么？

**思路**：表面能 $E=\alpha S$。
水滴分裂，**总表面积增大，总表面能增加**。分裂过程需要外界做功，转化为表面能。

### 3.5 将大小不同的两个肥皂泡用玻璃管连通，它们将发生什么变化？为什么？

**思路**：肥皂泡附加压强 $p_s=\frac{4\alpha}{R}$。
小肥皂泡半径 $R$ 小，泡内压强更高；气体从小泡流向大泡。**小泡不断缩小，直到只剩下管口处一个液膜；大泡半径变大**。

### 3.6 扩大液体表面和拉开一张橡皮膜，都要做功。所做的功转化成了什么能量？

拉开橡皮膜：外力做功转化为**弹性势能**。
扩大液体表面：外力做功转化为液体的**表面能（表面势能）**。

### 3.7 造成液面弯曲的原因是什么？

**思路**：内聚力与附着力的竞争。
液体分子之间的吸引力（内聚力）、液体分子与固体壁分子的吸引力（附着力）大小不等。

- 附着力 > 内聚力：润湿，液面凹弯；
- 附着力 < 内聚力：不润湿，液面凸弯。

### 3.8 为什么雾天多发生在秋、冬两季？

**思路**：凝结，饱和蒸气压随温度降低下降。
秋冬昼夜温差大，夜间气温下降，空气容纳水汽能力下降；空气水汽达到饱和，水汽以微小尘埃为凝结核，凝结成微小水滴悬浮在空气中，形成雾。

### 3.9 何谓接触角？何谓润湿与不润湿？从微观上加以说明。

**接触角**：固、液接触点，液体表面切线与固体表面之间，在液体一侧的夹角 $\theta$。

- **润湿**：$\theta<90^\circ$。液体与固体附着力 > 液体内聚力，液体尽量铺展在固体表面。
- **不润湿**：$\theta>90^\circ$。附着力 < 液体内聚力，液体收缩，尽量减小与固体接触面积。

### 3.10 何谓气体栓塞现象？该现象是如何产生的？

**气体栓塞**：毛细管内液体流动时，如果管内出现气泡，液体流动受阻，严重时堵塞管道。
产生：弯曲液面存在附加压强。一段液体两端有气泡，两个弯曲液面附加压强反向，抵消外加压强差，液体无法流动。血管、植物导管都可能发生。

### 3.11 微观上看，不同弯曲液面（凸、凹）下，分子逃出何种液面较为容易？

凸液面：液面分子受到向内的拉力更小，分子更容易脱离液面蒸发；
凹液面：分子受到指向液体内部的吸引力更强，分子更难逃出。
**凸液面分子更容易逃出液面。**

### 3.2

> 
> 毛细管完全润湿，上升高度 $h$；露出水面高度只有 $(h/2)$。能不能人造喷泉？求管内液面高度、接触角。
> **思路**：毛细管上升公式 $h=\frac{2\alpha\cos\theta}{\rho g r}$。
> 不能形成喷泉。液体上升到管口后，液面曲率半径自动变大，$\cos\theta$ 减小，附加压强降低，平衡重力。
> 液面停在管口，上升高度就是 $(h/2)$。
> 设原来 $\theta_0=0,\cos\theta_0=1,\ h=\frac{2\alpha}{\rho g r}$。
> 现在：$\frac h2=\frac{2\alpha\cos\theta}{\rho g r}=h\cos\theta\implies \cos\theta=\frac12,\ \boldsymbol{\theta=60^\circ}$。
> **答案：不能喷泉；液面高度$(h/2)$；接触角$60^\circ$**

### 3.3

- A点：液面上方，$p_A=p_0$
- B点：凹液面下方紧邻液面：$p_B=p_0-\frac{2\alpha}{r}$
- C点：和外部水槽液面同一水平面：$p_C=p_0$
- D点：水槽内，与C等高：$p_D=p_0$

> &=8\times\pi\times2.5\times10^{-2}\times(2.5\times10^{-2})^2\approx 3.93\times10^{-4}\ \mathrm{J}\
> \Delta p&=\frac{4\alpha}{R}=\frac{4\times 2.5\times10^{-2}}{2.5\times10^{-2}}=4\ \mathrm{Pa}
> \end{align*}
> $$
> **答案：做功$W\approx3.93\times10^{-4}\mathrm{J}$；内外压差$4\mathrm{Pa}$**
