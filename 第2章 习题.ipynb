# 思考题
### 思考题2.1
伯努利方程成立的条件是什么？泊肃叶方程成立的条件又是什么？

🏍️解答：
伯努利方程成立条件：理想流体、定常流动、同一流管。

泊肃叶方程成立条件：黏性流体、定常层流、水平圆管。

### 思考题2.2
如何解释“巷小风大”？

🏍️解答：

巷道狭窄，空气流通截面积小，根据连续性方程 $S_1v_1=S_2v_2$，截面积变小流速增大；对应伯努利原理，流速大处压强低。

### 思考题2.3

如何解释“河道宽处水流缓，河道窄处水流急”？

🏍️解答：

河水近似不可压缩流体，连续性方程 $S v=\text{常量}$。河道宽处横截面积$S$大，则流速$v$小；河道窄处横截面积$S$小，则流速$v$大。

### 思考题2.4

定常流动是理想流体所特有的吗？黏性流体有无定常流动？

🏍️解答：

不是理想流体特有。黏性流体也可以发生定常流动：空间各点流速不随时间变化，只是层内存在内摩擦力，泊肃叶公式描述的圆管层流就是黏性流体的定常流动。

### 思考题2.5

平行行驶的两艘船为什么要保持一定距离？

🏍️解答：

两船内侧水流通道变窄，水流流速大，压强小；船外侧水流流速小、压强大。内外压强差会把两船相互推向一起，容易发生碰撞，因此必须保持间距。

### 思考题2.6

液体在管中流动时，从连续性原理来看，管子越粗，流速越小，而从泊肃叶方程来看，管子越粗，流量越大，这两者是否矛盾，为什么？

🏍️解答：

不矛盾。

连续性方程 $\(Sv=Q\)$：流量$Q$固定不变前提下，截面积$S$越大流速$v$越小。

泊肃叶公式 $Q=\dfrac{\pi R^4\Delta p}{8\eta l}$：管道两端压强差$\Delta p$固定不变前提下，半径越大流量$Q$越大。

两个公式的前提条件不一样。

### 思考题2.7

水平的玻璃管，由截面均匀而截面积不等的A、B、C三段串联而成。水从A段进入，从C段流出。在三段管壁上各打一小孔，A段上小孔有水射出，B段小孔有气泡出现，C段小孔不射水，也无气泡，设水为理想流体。问三段管子哪段内径最大，哪段内径最小，为什么？

🏍️解答：

- B段内径最小：流速最大，压强低于大气压，所以空气被压入形成气泡。
- A段内径最大：流速小，压强大于大气压，小孔向外喷水。
- C段压强等于大气压，流速介于两者之间。


### 思考题2.8

生物流体同一般流体有何区别？

🏍️解答：

1. 很多生物流体（血液）属于非牛顿流体，黏度会随流速/切变率改变，不服从牛顿黏性定律；
2. 生物流体成分复杂，含有细胞、大分子颗粒；
3. 生物管道是弹性可形变管道，不是刚性管壁；
4. 流动多为脉动非定常流动。

### 习题2.2

变截面管宽部分的截面积$S_1=0.08\ \mathrm{cm^2}$，窄部分的截面积$S_2=0.04\ \mathrm{cm^2}$，两处间的压强差为25 Pa，求管中宽部流体流动的速度。已知液体密度为$1059.5\ \mathrm{kg\cdot m^{-3}}$。

🏍️解答：

$S_1=0.08\ \mathrm{cm^2},\ S_2=0.04\ \mathrm{cm^2}$，$\Delta p=25\ \mathrm{Pa}$，$\rho=1059.5\ \mathrm{kg/m^3}$

连续性：$S_1 v_1=S_2 v_2 \Rightarrow v_2=2v_1$

$$p_1+\frac12\rho v_1^2 = p_2+\frac12\rho v_2^2$$
$$\Delta p=\frac12\rho(v_2^2-v_1^2)=\frac12\rho(4v_1^2-v_1^2)=\frac32\rho v_1^2$$
$$v_1=\sqrt{\dfrac{2\Delta p}{3\rho}}=\sqrt{\dfrac{2\times 25}{3\times 1059.5}}\approx 0.126\ \mathrm{m/s}$$

### 习题2.12

在水管的某一点$A$，水的流速为$1\ \mathrm{m\cdot s^{-1}}$，计示压强为$3\times10^5\ \mathrm{Pa}$。如果沿水管到另一点$B$，$B$点比$A$点高度降低了20 cm，$B$点水管的横截面积为$A$点的二分之一，求$B$点的计示压强。

🏍️解答：

已知条件

$v_A=1\ \mathrm{m/s}$，A点计压强 $p_{A示}=3\times 10^5\ \mathrm{Pa}$

B比A降低 $h_A-h_B=0.20\ \mathrm{m}$

$S_B=\frac12 S_A$，水密度 $\rho=1000\ \mathrm{kg/m^3}$

步骤1：连续性方程求$v_B$

$$S_A v_A = S_B v_B$$

$$v_B = v_A\cdot \frac{S_A}{S_B}=1\cdot \frac{S_A}{\frac12 S_A}= 2\ \mathrm{m/s}$$


步骤2：伯努利方程（计示压强，大气压$p_0$抵消）

$$p_{A示} \frac12\rho v_A^2+\rho g h_A = p_{B示} + \frac12\rho v_B^2+\rho g h_B$$
移项：
$$
p_{B示} = p_{A示} + \frac12\rho(v_A^2-v_B^2)+\rho g(h_A-h_B)
$$

代入数值：
$$
\begin{aligned}
\frac12\rho(v_A^2-v_B^2)&=0.5\times 1000\times(1^2-2^2)=500\times(-3)=-1500\ \mathrm{Pa}\\
\rho g(h_A-h_B)&=1000\times 9.8\times 0.20 = 1960\ \mathrm{Pa}
\end{aligned}
$$

$$
\begin{aligned}
p_{B示}&=3\times10^5 -1500 +1960 \\
&=300000 + 460 \\
&=3.0046\times 10^{5}\ \mathrm{Pa}
\end{aligned}
$$

一个顶部开口的圆筒容器，高为20 cm，直径为10 cm，在圆筒底部中心开一个横截面积为$1\ \mathrm{cm^2}$的小圆孔，水从圆筒顶部以$140\ \mathrm{cm^3\cdot s^{-1}}$的流量由水管注入圆筒，问圆筒中的水面可以升到多高？

### 习题2.16

一个四壁竖直的大开口水槽，其中盛水，水深为$H$。在槽的一侧水面下$h$深处开一小孔。问：(1)射出的水流到地面时，距槽底边的距离是多少？(2)在槽壁上多高处再开一小孔，能使射出的水流有相同的射程？(3)要想得到最大射程，小孔要开在水面以下多深处？最大射程为多少？

🏍️解答：

水槽水深$H$，小孔水下深度$h$；小孔离地高度 $y=H-h$。托里拆利$v=\sqrt{2gh}$。

1\. 平抛运动：竖直 $y=\dfrac12 g t^2 \Rightarrow t=\sqrt{\dfrac{2(H-h)}{g}}$
水平射程
$$x=v\cdot t=\sqrt{2gh}\cdot\sqrt{\frac{2(H-h)}{g}}=\boldsymbol{2\sqrt{h(H-h)}}$$

2\. 相同射程：$h'$满足$h(H-h)=h'(H-h')$，解为$h'=H-h$；小孔距离水面深度为$\boldsymbol{H-h}$。

3\. 求$x=2\sqrt{h(H-h)}$最大值；$f(h)=h(H-h)$，$h=\dfrac H2$取最大。
小孔水下深度 $\boldsymbol{h=\dfrac H2}$；最大射程$\boldsymbol{x_\text{max}=H}$。

### 思考题解答

**2.1 伯努利方程成立的条件是什么？泊肃叶方程成立的条件又是什么？**
*   **解题思路**：回顾课本中这两个方程推导时的假设前提。
*   **解答**：
    *   伯努利方程成立的条件：理想流体（不可压缩、无黏性）、作定常流动、在同一流管中。
    *   泊肃叶方程成立的条件：黏性流体在水平圆管中作层流流动（即雷诺数 $Re < 2000$ 左右）。

**2.2 如何解释“巷小风大”？**
*   **解题思路**：利用连续性原理，空气可视为不可压缩流体。
*   **解答**：根据连续性原理，不可压缩流体作定常流动时，同一流管任意截面处流速与截面积的乘积是一个常量，即 $Sv = \text{常量}$。巷子小的地方截面积 $S$ 小，因此流速 $v$ 必然大，即“巷小风大”。

**2.3 如何解释“河道宽处水流缓，河道窄处水流急”？**
*   **解题思路**：同2.2，应用连续性原理。
*   **解答**：由于水的流量守恒，根据连续性方程 $Sv = \text{常量}$，河道宽处（截面积 $S$ 大）流速 $v$ 小，水流缓；河道窄处（截面积 $S$ 小）流速 $v$ 大，水流急。

**2.4 定常流动是理想流体所特有的吗？黏性流体有无定常流动？**
*   **解题思路**：明确“定常流动”是对流体运动状态的时间描述，与流体自身的物理性质（是否有黏性）无关。
*   **解答**：不是。定常流动是指流场中任一点的流速、压强等物理量不随时间改变，它适用于任何流体（理想或黏性）。黏性流体也可以有定常流动，只要满足流动状态不随时间变化即可。

**2.5 平行行驶的两艘船为什么要保持一定距离？**
*   **解题思路**：结合连续性方程和伯努利方程分析。
*   **解答**：两船平行行驶时，它们之间的水流通道变窄，根据连续性方程 $Sv = \text{常量}$，两船间水的流速会变大。根据伯努利方程 $p + \frac{1}{2}\rho v^2 + \rho gh = \text{常量}$，流速大的地方压强小。因此两船内侧的压强小于外侧，产生向内的压力差，会导致两船相互吸引而碰撞，所以必须保持安全距离。

**2.6 液体在管中流动时，从连续性原理来看，管子越粗，流速越小，而从泊肃叶方程来看，管子越粗，流量越大，这两者是否矛盾，为什么？**
*   **解题思路**：注意连续性方程讨论的是流速 $v$ 与面积 $S$ 的关系，而泊肃叶方程讨论的是流量 $Q$ 与半径 $R$ 的关系。
*   **解答**：不矛盾。连续性方程 $Q = Sv$ 说明，在流量 $Q$ 恒定的条件下，截面积 $S$ 越大流速 $v$ 越小。而泊肃叶方程 $Q = \frac{\pi R^4 \Delta p}{8\eta l}$ 说明，在压强差 $\Delta p$ 恒定的条件下，管道半径 $R$ 越大，流量 $Q$ 越大。两者是从不同变量控制角度描述流体运动，并不冲突。

**2.7 水平的玻璃管，由截面均匀而截面积不等的 $A, B, C$ 三段串联而成。水从 $A$ 段进入，从 $C$ 段流出。在三段管壁上各打一小孔，$A$ 段上小孔有水射出，$B$ 段小孔有气泡出现，$C$ 段小孔不射水，也无气泡，设水为理想流体。问三段管子哪段内径最大，哪段内径最小，为什么？**
*   **解题思路**：利用伯努利方程判断压强大小，再结合连续性方程判断截面积。
*   **解答**：
    *   $B$ 段有气泡出现，说明 $B$ 段压强 $p_B < p_0$（大气压）。
    *   $A$ 段有水射出，说明 $A$ 段压强 $p_A > p_0$。
    *   $C$ 段无气泡也无水射出，说明 $C$ 段压强 $p_C = p_0$。
    *   根据伯努利方程（水平管 $\rho gh$ 不变），压强越小的地方流速越大。因此流速 $v_B > v_C > v_A$。
    *   由连续性方程 $Sv = \text{常量}$，流速越大，截面积越小。所以 $S_B < S_C < S_A$。
    *   **结论：$A$ 段内径最大，$B$ 段内径最小。**

**2.8 生物流体同一般流体有何区别？**
*   **解题思路**：结合课本上方文本和生物学知识。
*   **解答**：生物流体（如血液）大多是非牛顿流体，具有各向异性的非线性黏弹性，且具有多层次微结构。它们的性质对环境因素（温度、pH值、取样方法等）相当敏感，不能用简单的连续介质假设理论完全描述。

**2.1 在如图所示的装置中，液体在水平管道中流动，截面 $B$ 与大气相通。试求盆中液体能够被吸上时，$h$ 的表达式（设 $S_A, S_B$ 分别为管道 $A, B$ 处的截面积，$Q$ 为秒流量，$C$ 点与大气相通，$p_C = p_0$）。**
*   **解题思路**：利用连续性方程求 $A$ 点流速，再根据伯努利方程求 $A$ 点压强，最后根据静力学平衡求 $h$。
*   **解答**：
    由连续性方程 $Q = S_A v_A = S_B v_B$，得 $v_A = \frac{Q}{S_A}$，$v_B = \frac{Q}{S_B}$。
    对 $A, B$ 两点列伯努利方程（水平管）：$p_A + \frac{1}{2}\rho v_A^2 = p_B + \frac{1}{2}\rho v_B^2$。
    已知 $p_B = p_0$（大气压），故 $p_A = p_0 + \frac{1}{2}\rho (v_B^2 - v_A^2) = p_0 + \frac{1}{2}\rho \left[ \left(\frac{Q}{S_B}\right)^2 - \left(\frac{Q}{S_A}\right)^2 \right]$。
    要使盆中液体被吸上，需满足 $p_0 - p_A \ge \rho g h$。
    即 $\rho g h \le -\frac{1}{2}\rho \left[ \left(\frac{Q}{S_B}\right)^2 - \left(\frac{Q}{S_A}\right)^2 \right] = \frac{\rho Q^2}{2} \left( \frac{1}{S_A^2} - \frac{1}{S_B^2} \right)$。
    解得：$h \le \frac{Q^2}{2g} \left( \frac{1}{S_A^2} - \frac{1}{S_B^2} \right)$。
    即 $h$ 的表达式为：$h = \frac{Q^2}{2g} \left( \frac{1}{S_A^2} - \frac{1}{S_B^2} \right)$。

**2.2 变截面管宽部分的截面积 $S_1 = 0.08 \text{ cm}^2$，窄部分的截面积 $S_2 = 0.04 \text{ cm}^2$，两处间的压强差为 $25 \text{ Pa}$，求管中宽部流体流动的速度。已知液体密度为 $1059.5 \text{ kg} \cdot \text{m}^{-3}$。**
*   **解题思路**：利用连续性方程和伯努利方程联立求解。
*   **解答**：
    由连续性方程：$S_1 v_1 = S_2 v_2 \Rightarrow v_2 = \frac{S_1}{S_2} v_1 = \frac{0.08}{0.04} v_1 = 2v_1$。
    由伯努利方程（水平管）：$p_1 + \frac{1}{2}\rho v_1^2 = p_2 + \frac{1}{2}\rho v_2^2$。
    整理得：$p_1 - p_2 = \frac{1}{2}\rho (v_2^2 - v_1^2)$。
    代入已知数据：$25 = \frac{1}{2} \times 1059.5 \times ((2v_1)^2 - v_1^2) = \frac{1}{2} \times 1059.5 \times 3v_1^2$。
    $25 = 1589.25 v_1^2 \Rightarrow v_1^2 = \frac{25}{1589.25} \approx 0.01573$。
    $v_1 \approx 0.125 \text{ m/s}$。
    **答：管中宽部流体流动的速度为 $0.125 \text{ m/s}$。**

**2.3 如图所示，水平管的横截面积在粗处为 $40 \text{ cm}^2$，在细处为 $10 \text{ cm}^2$，水的流量为 $3 \times 10^{-3} \text{ m}^3 \cdot \text{s}^{-1}$。水平管下面装有一个U形管，管内下部盛有水银，已知水银密度为 $13.6 \times 10^3 \text{ kg} \cdot \text{m}^{-3}$。求：(1) 水在粗处和细处的流速；(2) 两处的压强差；(3) U形管中水银柱的高度差。**
*   **解题思路**：(1) 由 $Q=Sv$ 直接求解；(2) 由伯努利方程求解；(3) 由压强差和静力学公式 $\Delta p = (\rho_{Hg} - \rho_{water})gh$ 求解。
*   **解答**：
    (1) 粗处流速 $v_1 = \frac{Q}{S_1} = \frac{3 \times 10^{-3}}{40 \times 10^{-4}} = 0.75 \text{ m/s}$。
    细处流速 $v_2 = \frac{Q}{S_2} = \frac{3 \times 10^{-3}}{10 \times 10^{-4}} = 3 \text{ m/s}$。
    (2) 由伯努利方程（水平管，取水密度 $\rho = 1000 \text{ kg/m}^3$）：
    $p_1 - p_2 = \frac{1}{2}\rho (v_2^2 - v_1^2) = \frac{1}{2} \times 1000 \times (3^2 - 0.75^2) = 500 \times (9 - 0.5625) = 4218.75 \text{ Pa}$。
    (3) U形管中水银柱高度差 $h$ 满足 $\Delta p = (\rho_{Hg} - \rho_{water})gh$。
    $4218.75 = (13600 - 1000) \times 9.8 \times h = 12600 \times 9.8 \times h$。
    $h = \frac{4218.75}{12600 \times 9.8} \approx 0.0342 \text{ m} = 3.42 \text{ cm}$。

**2.4 如图所示，水定常地流过直径为 $0.1 \text{ m}$ 的管子，经喷管射入大气，喷管直径为 $0.05 \text{ m}$。如果截面 $b$ 上的水速为 $20 \text{ m} \cdot \text{s}^{-1}$，求截面 $a$ 上流体的压强。**
*   **解题思路**：利用连续性方程求出 $a$ 处流速，再对 $a, b$ 两点列伯努利方程（$b$ 处为大气压，且高度差为 $4\text{m}$）。
*   **解答**：
    $S_a = \pi (\frac{0.1}{2})^2 = 0.0025\pi \text{ m}^2$，$S_b = \pi (\frac{0.05}{2})^2 = 0.000625\pi \text{ m}^2$。
    由连续性方程 $v_a S_a = v_b S_b$，得 $v_a = v_b \frac{S_b}{S_a} = 20 \times \frac{0.000625\pi}{0.0025\pi} = 20 \times \frac{1}{4} = 5 \text{ m/s}$。
    对 $a, b$ 两点列伯努利方程，取 $b$ 处为基准面（高度 $0$），则 $a$ 处高度 $h_a = 4\text{ m}$，$b$ 处压强 $p_b = p_0 = 1.013 \times 10^5 \text{ Pa}$：
    $p_a + \frac{1}{2}\rho v_a^2 + \rho g h_a = p_b + \frac{1}{2}\rho v_b^2 + \rho g h_b$
    $p_a + \frac{1}{2} \times 1000 \times 5^2 + 1000 \times 9.8 \times 4 = 1.013 \times 10^5 + \frac{1}{2} \times 1000 \times 20^2 + 0$
    $p_a + 12500 + 39200 = 101300 + 200000$
    $p_a = 301300 - 51700 = 249600 \text{ Pa} \approx 2.50 \times 10^5 \text{ Pa}$。
    **答：截面 $a$ 上流体的压强约为 $2.50 \times 10^5 \text{ Pa}$。**

**2.5 半径为 $0.02 \text{ m}$ 的水管以 $0.01 \text{ m}^3 \cdot \text{s}^{-1}$ 的流量输送水，水温为 $20 \text{ ℃}$。问：(1) 水的平均流速是多少？(2) 流动是层流还是湍流？(3) 要确定管中流体流动的最大速度，这些数据是否足够？**
*   **解题思路**：(1) 由 $Q=Sv$ 求平均流速；(2) 计算雷诺数 $Re$ 判断；(3) 看是否能通过已知量求出最大速度（层流时 $v_{max}=2v_{avg}$，湍流则不可）。
*   **解答**：
    (1) 截面积 $S = \pi r^2 = \pi \times (0.02)^2 \approx 1.2566 \times 10^{-3} \text{ m}^2$。
    平均流速 $v = \frac{Q}{S} = \frac{0.01}{1.2566 \times 10^{-3}} \approx 7.96 \text{ m/s}$。
    (2) $20\text{℃}$ 水的黏度 $\eta \approx 1.005 \times 10^{-3} \text{ Pa}\cdot\text{s}$，密度 $\rho = 998 \text{ kg/m}^3$。
    $Re = \frac{\rho v d}{\eta} = \frac{998 \times 7.96 \times 0.04}{1.005 \times 10^{-3}} \approx 3.16 \times 10^5 > 2000$，故为**湍流**。
    (3) **不够**。因为流动是湍流，湍流的速度分布不再是简单的抛物线分布，无法仅凭平均流速和管径准确计算出最大速度（通常需要知道湍流的具体模型或摩擦系数等）。

**2.6 考虑理想不可压缩流体在收缩管道中的一维定常流动，如果管道截面积沿流动方向（$x$ 方向）的变化规律为 $A(x)$，如图所示，由测量得知收缩管进口截面积为 $A_0$，进口流速为 $v_0$，试求收缩管中任意截面处的压强梯度 $\frac{dp}{dx}$ 的表达式。设流体密度为 $\rho$，管道水平放置。**
*   **解题思路**：先由连续性方程求出 $v(x)$，再对 $x$ 求导得到 $\frac{dv}{dx}$，最后结合伯努利方程对 $x$ 求导得到 $\frac{dp}{dx}$。
*   **解答**：
    由连续性方程：$A(x)v(x) = A_0 v_0 \Rightarrow v(x) = \frac{A_0 v_0}{A(x)}$。
    求导得流速梯度：$\frac{dv}{dx} = A_0 v_0 \cdot \left( -\frac{1}{A^2(x)} \right) \cdot \frac{dA}{dx} = -\frac{A_0 v_0}{A^2(x)} \frac{dA}{dx}$。
    对伯努利方程 $p + \frac{1}{2}\rho v^2 = \text{常量}$ 两边对 $x$ 求导：
    $\frac{dp}{dx} + \rho v \frac{dv}{dx} = 0 \Rightarrow \frac{dp}{dx} = -\rho v \frac{dv}{dx}$。
    将 $v(x)$ 和 $\frac{dv}{dx}$ 代入：
    $\frac{dp}{dx} = -\rho \left( \frac{A_0 v_0}{A(x)} \right) \left( -\frac{A_0 v_0}{A^2(x)} \frac{dA}{dx} \right) = \frac{\rho A_0^2 v_0^2}{A^3(x)} \frac{dA}{dx}$。

**2.7 由于飞机机翼形状的关系，在机翼上面的气流速度大于下面的速度，在机翼上下两面间就形成压强差，因而产生使机翼上升的力。假定空气流过机翼是定常流动，并假定空气的密度不变，等于 $1.29 \times 10^{-3} \text{ g} \cdot \text{cm}^{-3}$。如果机翼下面的气流速率为 $100 \text{ m} \cdot \text{s}^{-1}$，问要得到 $1000 \text{ Pa}$ 的压强差，机翼上面的气流速率应该是多少？**

*   **解答**：
    密度 $\rho = 1.29 \times 10^{-3} \text{ g/cm}^3 = 1.29 \text{ kg/m}^3$。
    由伯努利方程：$p_2 + \frac{1}{2}\rho v_2^2 = p_1 + \frac{1}{2}\rho v_1^2$ （下标1为下表面，2为上表面）。
    压强差 $\Delta p = p_1 - p_2 = \frac{1}{2}\rho (v_2^2 - v_1^2)$。
    $1000 = \frac{1}{2} \times 1.29 \times (v_2^2 - 100^2)$。
    $2000 / 1.29 = v_2^2 - 10000 \Rightarrow v_2^2 \approx 1550.39 + 10000 = 11550.39$。
    $v_2 \approx 107.47 \text{ m/s}$。
    **答：机翼上面的气流速率应该是约 $107.47 \text{ m/s}$。**

**2.8 如图所示，风平行地吹过一栋密封的房子，当风速为多大时房子内外的压强差为 10%？假如房子密封得足够好，这一压强差会使得房子爆裂。已知空气密度为 $1.293 \text{ kg} \cdot \text{m}^{-3}$。**

*   **解答**：
    室外：$p + \frac{1}{2}\rho v^2 = \text{常量}$；室内：$v=0$，$p_{in} = p_0$。
    由伯努利方程，室外压强 $p_{out} = p_0 - \frac{1}{2}\rho v^2$。
    内外压强差 $\Delta p = p_{in} - p_{out} = \frac{1}{2}\rho v^2$。
    题目已知 $\Delta p = 10\% \times p_0 = 0.1 \times 1.013 \times 10^5 = 10130 \text{ Pa}$。
    $10130 = \frac{1}{2} \times 1.293 \times v^2 \Rightarrow v^2 = \frac{20260}{1.293} \approx 15668.99$。
    $v \approx 125.18 \text{ m/s}$。
    **答：当风速约为 $125.18 \text{ m/s}$ 时，压强差为10%。**

**2.9 主管与细管间的压强差为 $10^6 \text{ Pa}$，主管和细管的横截面积分别为 $0.2 \text{ m}^2$ 和 $0.1 \text{ m}^2$，问管子中水的流量是多少？**
*   **解题思路**：联立连续性方程和伯努利方程。
*   **解答**：
    $S_1 = 0.2 \text{ m}^2, S_2 = 0.1 \text{ m}^2$。由连续性方程 $S_1 v_1 = S_2 v_2 \Rightarrow v_2 = 2v_1$。
    由伯努利方程：$p_1 - p_2 = \frac{1}{2}\rho (v_2^2 - v_1^2) = \frac{1}{2} \times 1000 \times (4v_1^2 - v_1^2) = 1500 v_1^2$。
    已知 $p_1 - p_2 = 10^6 \text{ Pa}$，则 $10^6 = 1500 v_1^2 \Rightarrow v_1^2 = \frac{1000000}{1500} \approx 666.67$。
    $v_1 \approx 25.82 \text{ m/s}$。

**2.10 一根长水管直径为 $15 \text{ cm}$，其中充满水，水管的狭细部分直径为 $7.5 \text{ cm}$，如果在 $15 \text{ cm}$ 直径部分水流速为 $1.2 \text{ cm} \cdot \text{s}^{-1}$。(1) 求狭细部分水的流速；(2) 以 $\text{m}^3 \cdot \text{s}^{-1}$ 为单位表示出水的流量。**
*   **解题思路**：直接应用连续性方程。
*   **解答**：
    (1) $v_2 = v_1 \frac{S_1}{S_2} = v_1 \left(\frac{d_1}{d_2}\right)^2 = 1.2 \times \left(\frac{15}{7.5}\right)^2 = 1.2 \times 4 = 4.8 \text{ cm/s}$。
    (2) $Q = S_1 v_1 = \pi \left(\frac{0.15}{2}\right)^2 \times 0.012 = \pi \times 0.005625 \times 0.012 \approx 2.12 \times 10^{-4} \text{ m}^3\text{/s}$。

**2.11 为了使救火水管中的水流可射达的竖直高度为 $20 \text{ m}$，问和水龙管相连接的总水管中需要的计示压强是多少？**
*   **解题思路**：运用伯努利方程，水管出口处压强为大气压（计示压强为0），速度由竖直上抛运动公式求出。
*   **解答**：
    设水管出口速度为 $v$，由运动学公式 $v^2 = 2gh$。
    对总水管和出口列伯努利方程（忽略高度差）：$p + 0 = p_0 + \frac{1}{2}\rho v^2$。
    计示压强 $p_{gauge} = p - p_0 = \frac{1}{2}\rho v^2 = \frac{1}{2}\rho (2gh) = \rho g h$。
    $p_{gauge} = 1000 \times 9.8 \times 20 = 1.96 \times 10^5 \text{ Pa}$。

**2.12 在水管的某一点 $A$，水的流速为 $1 \text{ m} \cdot \text{s}^{-1}$，计示压强为 $3 \times 10^5 \text{ Pa}$。如果沿水管到另一点 $B$，$B$ 点比 $A$ 点高度降低了 $20 \text{ cm}$，$B$ 点水管的横截面积为 $A$ 点的二分之一，求 $B$ 点的计示压强。**
*   **解题思路**：由连续性方程求 $B$ 点速度，再对 $A, B$ 列伯努利方程。
*   **解答**：
    $v_A = 1 \text{ m/s}$。由 $S_A v_A = S_B v_B$ 且 $S_B = 0.5 S_A$，得 $v_B = 2 v_A = 2 \text{ m/s}$。
    列伯努利方程：$p_A + \frac{1}{2}\rho v_A^2 + \rho g h_A = p_B + \frac{1}{2}\rho v_B^2 + \rho g h_B$。
    已知 $h_A - h_B = 0.2 \text{ m}$，$p_{A,gauge} = 3 \times 10^5 \text{ Pa}$。
    $p_{B,gauge} = p_A - p_0 + \frac{1}{2}\rho (v_A^2 - v_B^2) + \rho g (h_A - h_B)$
    $p_{B,gauge} = 3 \times 10^5 + \frac{1}{2} \times 1000 \times (1^2 - 2^2) + 1000 \times 9.8 \times 0.2$
    $p_{B,gauge} = 300000 - 1500 + 1960 = 300460 \text{ Pa} \approx 3.00 \times 10^5 \text{ Pa}$。

**2.13 在一横截面积为 $10 \text{ cm}^2$ 的水平管内有水流动，在管的另一段横截面积缩为 $5 \text{ cm}^2$。这两截面处的压强差为 $300 \text{ Pa}$。问一分钟内从管流出的水有多少立方米？**
*   **解题思路**：利用连续性方程和伯努利方程求出流速，再计算流量。
*   **解答**：
    $S_1 = 10 \text{ cm}^2, S_2 = 5 \text{ cm}^2 \Rightarrow v_2 = 2v_1$。
    $p_1 - p_2 = \frac{1}{2}\rho (v_2^2 - v_1^2) = \frac{1}{2} \times 1000 \times (4v_1^2 - v_1^2) = 1500 v_1^2 = 300$。
    $v_1^2 = 0.2 \Rightarrow v_1 = \sqrt{0.2} \approx 0.447 \text{ m/s}$。
    $Q = S_1 v_1 = 10 \times 10^{-4} \times 0.447 \approx 4.47 \times 10^{-4} \text{ m}^3\text{/s}$。
    一分钟流出的水量 $V = Q \times 60 = 4.47 \times 10^{-4} \times 60 \approx 0.0268 \text{ m}^3$。

**2.14 从一水平管中排水的流量是 $0.004 \text{ m}^3 \cdot \text{s}^{-1}$。管的横截面积为 $0.01 \text{ m}^2$ 处的绝对压强是 $12 \times 10^4 \text{ Pa}$。问管的横截面积压缩为多少时，才能使压强减少为 $1.0 \times 10^5 \text{ Pa}$？**
*   **解题思路**：由 $Q$ 和初始 $S$ 求 $v_1$，由伯努利方程求 $v_2$，再由 $S_2 = Q/v_2$ 求面积。
*   **解答**：
    $v_1 = Q / S_1 = 0.004 / 0.01 = 0.4 \text{ m/s}$。
    $p_1 - p_2 = \frac{1}{2}\rho (v_2^2 - v_1^2)$。
    $12 \times 10^4 - 1.0 \times 10^5 = 20000 = \frac{1}{2} \times 1000 \times (v_2^2 - 0.4^2)$。
    $40 = v_2^2 - 0.16 \Rightarrow v_2^2 = 40.16 \Rightarrow v_2 \approx 6.337 \text{ m/s}$。

**2.16 一个四壁竖直的大开口水槽，其中盛水，水深为 $H$。在槽的一侧水面下 $h$ 深处开一小孔。问：(1) 射出的水流到地面时，距槽底边的距离是多少？(2) 在槽壁上多高处再开一小孔，能使射出的水流有相同的射程？(3) 要想得到最大射程，小孔要开在水面以下多深处？最大射程为多少？**
*   **解题思路**：小孔射流为平抛运动。水平射程 $x = v t$，下落时间由高度决定。
*   **解答**：
    (1) 孔处速度 $v = \sqrt{2gh}$。孔距离地面高度 $H-h$。
    下落时间 $t = \sqrt{\frac{2(H-h)}{g}}$。
    射程 $x = v t = \sqrt{2gh} \cdot \sqrt{\frac{2(H-h)}{g}} = 2\sqrt{h(H-h)}$。
    (2) 设另一孔深度为 $h'$，则射程 $x' = 2\sqrt{h'(H-h')} = x = 2\sqrt{h(H-h)}$。
    解得 $h'(H-h') = h(H-h)$。整理得 $h'^2 - Hh' + h(H-h) = 0$。
    解得 $h' = h$（原位置）或 $h' = H - h$。
    所以在水面下 $H-h$ 深处开孔。
    (3) 射程 $x = 2\sqrt{h(H-h)}$，当 $h = H-h$ 即 $h = H/2$ 时，$x$ 最大。
    最大射程 $x_{max} = 2\sqrt{\frac{H}{2} \cdot \frac{H}{2}} = H$。

**2.17 一圆筒中的水高为 $H$，底面积为 $S_1$，桶的底部有一面积为 $S_2$ 的小孔。问桶中的水全部流尽需多长时间？**

*   **解答**：
    设 $t$ 时刻水深为 $h$。由连续性方程 $-S_1 \frac{dh}{dt} = S_2 \sqrt{2gh}$。

    积分：$t = \int_H^0 -\frac{S_1}{S_2 \sqrt{2g}} h^{-1/2} dh = \frac{S_1}{S_2 \sqrt{2g}} \int_0^H h^{-1/2} dh = \frac{S_1}{S_2 \sqrt{2g}} [2h^{1/2}]_0^H = \frac{S_1}{S_2} \sqrt{\frac{2H}{g}}$。

    $Q = v_{avg} \cdot \pi R^2 = 3.3 \times 10^{-4} \times \pi \times (2 \times 10^{-3})^2 = 4.15 \times 10^{-9} \text{ m}^3\text{/s}$。
    $\Delta p = \frac{8\eta l Q}{\pi R^4} = \frac{8 \times 4 \times 10^{-3} \times 10^{-3} \times 4.15 \times 10^{-9}}{\pi \times (2 \times 10^{-3})^4} = \frac{1.328 \times 10^{-13}}{\pi \times 1.6 \times 10^{-11}} \approx 2.64 \times 10^{-3} \text{ Pa}$。

**2.33 人的心脏不断向大动脉泵送血液，进入大动脉的最大流量约为 $500 \text{ cm}^3 \cdot \text{s}^{-1}$。假设大动脉的直径为 $2.5 \text{ cm}$，流动大致可认为是层流，血液是牛顿流体。求：(1) 大动脉内单位长度上的压强差；(2) 轴心处的流动速度。已知血液黏度 $\eta = 4 \times 10^{-3} \text{ Pa} \cdot \text{s}$。**
*   **解题思路**：(1) 泊肃叶方程变形求 $\Delta p / l$；(2) 平均流速 $v = Q/S$，轴心流速 $v_{max} = 2v$。
*   **解答**：
    (1) $Q = \frac{\pi R^4 \Delta p}{8\eta l} \Rightarrow \frac{\Delta p}{l} = \frac{8\eta Q}{\pi R^4}$。
    $Q = 500 \text{ cm}^3\text{/s} = 5 \times 10^{-4} \text{ m}^3\text{/s}$，$R = 1.25 \text{ cm} = 0.0125 \text{ m}$。
    $\frac{\Delta p}{l} = \frac{8 \times 4 \times 10^{-3} \times 5 \times 10^{-4}}{\pi \times (0.0125)^4} = \frac{1.6 \times 10^{-5}}{\pi \times 2.44 \times 10^{-8}} \approx 208.7 \text{ Pa/m}$。
    (2) 平均流速 $v_{avg} = \frac{Q}{\pi R^2} = \frac{5 \times 10^{-4}}{\pi \times (0.0125)^2} \approx 1.0186 \text{ m/s}$。
    轴心处速度 $v_{max} = 2 v_{avg} \approx 2.037 \text{ m/s}$。

**2.34 在其他参量不变的条件下，要使血管的流体力学阻尼（用 $\Delta p / Q$ 表示）增大 30%，半径需要改变多少？假设该情况为牛顿流体的层流行为。**
*   **解题思路**：由泊肃叶定律，流体力学阻尼 $R_f = \Delta p / Q = \frac{8\eta l}{\pi R^4}$，即 $R_f \propto R^{-4}$。
*   **解答**：
    设原阻尼为 $R_{f1}$，新阻尼为 $R_{f2} = 1.3 R_{f1}$，原半径为 $R_1$，新半径为 $R_2$。
    由 $R_f \propto R^{-4}$ 得：$\frac{R_{f2}}{R_{f1}} = \left(\frac{R_2}{R_1}\right)^{-4}$。
    $1.3 = \left(\frac{R_2}{R_1}\right)^{-4} \Rightarrow \frac{R_2}{R_1} = (1.3)^{-1/4} \approx 0.935$。
    所以半径需要改变为原来的约 $93.5\%$，即减小约 $6.5\%$。

# 第2章 流体力学基础 思考题+习题解答

## 思考题

### 2.1 伯努利方程成立的条件是什么？泊肃叶方程成立的条件又是什么？

**伯努利方程**
条件：**理想流体、定常流动、沿同一流线**。
（理想流体=不可压缩、无黏滞；定常流动：空间各点流速不随时间变化）

**泊肃叶方程** $Q=\frac{\pi R^4}{8\eta l}\Delta p$
条件：**黏性牛顿流体、圆管、层流、定常流动、不可压缩、水平细圆管**。

---

### 2.2 如何解释“巷小风大”？

思路：连续性原理+伯努利原理。
巷子狭窄，空气流通截面积$S$变小。由连续性方程 $S_1v_1=S_2v_2$，截面积减小，流速$v$增大；再由伯努利方程，流速大处压强低。巷子内空气流速明显大于巷外开阔区域，因此感觉风大。

---

### 2.3 如何解释“河道宽处水流缓，河道窄处水流急”？

思路：**连续性原理 $S_1v_1=S_2v_2$**
河水近似不可压缩，定常流动。河道宽处横截面积$S$大，流速$v$就小；河道窄处$S$小，流速$v$大。

---

### 2.4 定常流动是理想流体所特有的吗？黏性流体有无定常流动？

不是。**黏性流体也可以有定常流动**。
定常流动定义：流场中任意空间点流速不随时间改变，和流体有没有黏性无关。
例：圆管中层流（泊肃叶流动）就是黏性流体的定常流动；黏性只是会带来能量损耗，不破坏定常的定义。

---

### 2.5 平行行驶的两艘船为什么要保持一定距离？

思路：伯努利原理。
两船之间的水道狭窄，水流截面积小，水的流速增大；流速大处压强减小。船外侧水域宽阔、流速小、压强大。内外压强差会产生**相互吸引力**，两船会相互靠拢，容易发生碰撞事故，因此要保持间距。

---

### 2.6 液体在管中流动时，从连续性原理来看，管子越粗，流速越小，而从泊肃叶方程来看，管子越粗，流量越大，这两者是否矛盾，为什么？

**不矛盾**。

- 连续性原理 $S_1v_1=S_2v_2$：**同一根管道，流量$Q$固定不变**，$S\uparrow \Rightarrow v\downarrow$。
- 泊肃叶公式 $Q=\frac{\pi R^4}{8\eta l}\Delta p$：**压强差$\Delta p$、管长不变，换不同粗细管子**，管径$R$越大，能通过的流量$Q$越大。

两者前提条件不同：连续性是**同一管道，流量固定**；泊肃叶是**不同管道，两端压差固定**。

---

### 2.7 水平的玻璃管，由截面均匀而截面积不等的A、B、C三段串联而成。水从A段进入，从C段流出。在三段管壁上各打一小孔，A段上小孔有水射出，B段小孔有气泡出现，C段小孔不射水，也无气泡，设水为理想流体。问三段管子哪段内径最大，哪段内径最小，为什么？

思路：伯努利方程（水平管$h$相同，$p+\frac12\rho v^2=\text{常量}$）+连续性$S v=\text{常量}$

- A小孔向外喷水：管内压强 $p_A>p_0$（大气压）；
- B小孔进气泡：管内压强 $p_B<p_0$；
- C小孔无进出：$p_C=p_0$。

压强大小：$\boldsymbol{p_A>p_C>p_B}$
水平理想流体：压强越高，流速越小，截面积越大。
流速：$v_A<v_C<v_B$；截面积$S_A>S_C>S_B$。

✅ **A内径最大，B内径最小**

---

### 2.8 生物流体同一般流体有何区别？

1. 普通流体多为牛顿流体；**很多生物流体（血液）是非牛顿流体，黏度随流速/切变率改变**；
2. 生物流体常含细胞、大分子，是多相悬浮体系，不是单纯均质流体；
3. 生物管道（血管）不是刚性管，**血管可弹性形变，管径随压强变化**；
4. 流动常是非定常（心脏脉动，血流周期性变化）；
5. 有生物活性，会发生凝聚、沉降，受温度、pH影响。

## 2.16 大开口水槽水深$H$，水面下$h$处小孔。

托里拆利：小孔流速$v=\sqrt{2gh}$
水槽高$H$，小孔离地高度：$H-h$
(1) 平抛：竖直 $H-h=\frac12 g t^2 \Rightarrow t=\sqrt{\frac{2(H-h)}{g}}$
水平射程 $x=v t=\sqrt{2gh}\cdot\sqrt{\frac{2(H-h)}{g}}=\boldsymbol{2\sqrt{h(H-h)}}$

(2) 射程相同：$2\sqrt{h_1(H-h_1)}=2\sqrt{h_2(H-h_2)}$，解得 $h_2=H-h_1$，即在水面下$H-h$深处再开孔，射程一样。

(3) 求$x=2\sqrt{h(H-h)}$最大值。$f(h)=hH-h^2$，求导，极值$h=\frac H2$。
小孔在水面下$\boldsymbol{(\boldsymbol{H/2})}$，最大射程$x_\text{max}=2\sqrt{\frac H2\cdot\frac H2}=\boldsymbol{H}$

## 2.17

> 
> 圆筒水深$H$，底面积$S_1$，底部小孔$S_2$，求全部流空时间。
> 微元：$dt$内液面下降$dh$，流出体积$-S_1 dh=S_2 v dt=S_2\sqrt{2gh} dt$
> $$
> dt=-\frac{S_1}{S_2}\frac{dh}{\sqrt{2gh}}
> $$
> 积分：$h:H\rightarrow0$
> $$
> T=\int_0^T dt=\frac{S_1}{S_2\sqrt{2g}}\int_0^H h^{-1/2} dh
> =\frac{S_1}{S_2\sqrt{2g}}\cdot\left[2\sqrt{h}\right]_0^H
> $$
> $$
> \boldsymbol{T=\frac{S_1}{S_2}\sqrt{\frac{2H}{g}}}
> $$
