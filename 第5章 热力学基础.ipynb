# 本章提要

1. **平衡态**
   在不受环境影响条件下，一个系统的宏观性质不随时间改变的状态。
2. **准静态过程**
   在过程进行中的每一时刻，系统的状态都无限接近平衡态的过程。
3. **准静态过程中系统对外做的体积功**
   $$ \mathrm{d}A = p\mathrm{d}V, \quad A = \int_{V_1}^{V_2} p\mathrm{d}V $$
4. **热量**
   系统和环境之间或两个物体之间由于温度不同而交换的热运动能量。
5. **热力学第一定律**
   $$ Q = E_2 - E_1 + A $$
   $$ \mathrm{d}Q = \mathrm{d}E + \mathrm{d}A $$
6. **热容**
   $$ C = \frac{\mathrm{d}Q}{\mathrm{d}T} $$
   摩尔定压热容：
   $$ C_{p,\mathrm{m}} = \frac{\mathrm{d}Q_{p,\mathrm{m}}}{\mathrm{d}T} $$
   摩尔定容热容：
   $$ C_{V,\mathrm{m}} = \frac{\mathrm{d}Q_{V,\mathrm{m}}}{\mathrm{d}T} $$
   迈耶公式：
   $$ C_{p,\mathrm{m}} - C_{V,\mathrm{m}} = R $$
   摩尔热容比：
   $$ \gamma = \frac{C_{p,\mathrm{m}}}{C_{V,\mathrm{m}}} $$
   理想气体的内能：
   $$ \Delta E = E_2 - E_1 = \frac{m}{M} C_{V,\mathrm{m}} (T_2 - T_1) $$
7. **绝热过程**
   $$ Q = 0, \quad A = -\Delta E $$
   $$ pV^\gamma = \text{常量}, \quad A = \frac{p_1 V_1 - p_2 V_2}{\gamma - 1} $$
8. **循环过程**
   正循环热机效率：
   $$ \eta = \frac{A}{Q_1} = 1 - \frac{Q_2}{Q_1} $$
   逆循环制冷系数：
   $$ w = \frac{Q_2}{A} $$
9. **卡诺循环**
   $$ \eta = 1 - \frac{T_2}{T_1}, \quad w = \frac{T_2}{T_1 - T_2} $$
10. **热力学第二定律**
    克劳修斯表述：热量不可能自动地由低温物体传向高温物体。
    开尔文表述：不可能由单一热源吸收热量使之完全转化为功而不引起其他变化。
    微观意义：自发过程总是沿着使分子运动更加无序的方向进行。
11. **可逆过程**
    无摩擦的准静态过程是可逆过程。
12. **克劳修斯熵**
    $$ \mathrm{d}S = \frac{\mathrm{d}Q}{T} \quad (\text{可逆过程}) $$
    $$ S_2 - S_1 = \int_1^2 \frac{\mathrm{d}Q}{T} \quad (\text{可逆过程}) $$
    熵增加原理：
    $$ \Delta S \geqslant 0 \quad (\text{孤立系统，等号适用于可逆过程}) $$

# 第一部分：思考题（5.1 - 5.34）

### 5.1 什么是热运动？它与机械运动有何不同？研究热运动的方法有哪些？
**解题思路**：从微观粒子的无规则运动和宏观物体的有规律运动进行区分。
**解答**：热运动是大量微观粒子（分子、原子等）永不停息的无规则运动。不同点：机械运动是宏观物体有规律的机械位移，服从牛顿力学；热运动是微观粒子的无序运动，具有统计规律性。研究方法：宏观上采用热力学方法（实验总结），微观上采用统计物理方法（概率统计）。

### 5.2 如何理解“热力学系统”这一概念？引入“热力学系统”有何意义？
**解题思路**：区分研究对象与外界环境。
**解答**：热力学系统是指在热力学研究中被选取作为研究对象的宏观物体或物体系。引入“系统”是为了明确研究边界，区分系统内部与外界环境，从而准确描述系统与外界之间的能量和物质交换。

### 5.3 什么是宏观量？什么是微观量？为什么要引入这两类量？
**解题思路**：区分可直接测量的量（宏观）与描述单个粒子的量（微观）。
**解答**：宏观量是描述系统整体状态且能被宏观仪器测量的物理量（如压强 $p$、体积 $V$、温度 $T$）。微观量是描述单个微观粒子特征的物理量（如分子质量、速度、动量）。引入二者是为了通过统计方法建立微观量与宏观量之间的联系。

### 5.4 热力学过程分为哪些种类？
**解题思路**：按照过程条件分类。
**解答**：按条件可分为等容过程、等压过程、等温过程、绝热过程及多方过程；按过程方向可分为可逆过程和不可逆过程；按状态变化可分为循环过程和非循环过程。

### 5.5 什么是平衡态？什么是准静态过程？在 $p-V$ 图上如何表示平衡态与准静态过程？
**解题思路**：理解平衡态是状态不随时间变化，准静态过程是无限缓慢的过程。
**解答**：平衡态是在不受外界影响的条件下，系统宏观性质不随时间变化的状态。准静态过程是过程进行的足够缓慢，以至于系统在任意时刻都无限接近于平衡态的过程。在 $p-V$ 图上，平衡态对应一个确定的点，准静态过程对应一条连续的曲线。

### 5.6 什么是内能、热量和功？这三个量之间有什么联系？
**解题思路**：运用热力学第一定律。
**解答**：内能是系统内部所有分子动能和势能的总和，是状态量；热量和功是系统与外界交换能量的两种方式，是过程量。联系：热力学第一定律 $Q = \Delta E + W$（系统吸热等于内能增量加系统对外做的功）。

### 5.7 如何认识热力学中的功与力学中的功？
**解题思路**：从宏观和微观能量转化角度区分。
**解答**：力学中的功是力与位移的乘积（$W = \int F dx$）。热力学中的功是系统体积变化时所做的功（$W = \int p dV$）。微观本质上，热力学功是系统内分子有序运动与外界机械运动之间的能量转化。

### 5.8 热力学系统与环境交换能量的方式有哪些？这些方式有什么相同与不同之处？
**解题思路**：比较做功与热传递。
**解答**：两种方式：做功和热传递。相同点：都是能量转移的方式，都能改变系统内能。不同点：做功是宏观有序运动的能量转移，热传递是微观无序热运动的能量转移。

### 5.9 为什么第一类永动机是不可能制造出来的？
**解题思路**：运用热力学第一定律。
**解答**：第一类永动机是不消耗能量而不断对外做功的机器，这违背了热力学第一定律（能量守恒定律），即能量既不能凭空产生也不能凭空消失。

### 5.10 什么是可逆过程？准静态过程都是可逆过程吗？
**解题思路**：区分“无耗散”与“准静态”两个条件。
**解答**：可逆过程是系统状态变化后能沿原过程逆向复原，且外界不留下任何痕迹的过程。准静态过程不一定是可逆过程，还必须满足无摩擦力等耗散效应。只有无耗散的准静态过程才是可逆过程。

### 5.11 什么是不可逆过程？过程的不可逆性是否等同于过程的方向性？
**解题思路**：对比不可逆性与方向性。
**解答**：不可逆过程是系统不能自发逆向复原且外界不能不留痕迹的过程。不等同于方向性。不可逆性指无法复原的原因（如耗散、有限温差传热）；方向性是指自然过程发生的单向趋势。不可逆性导致了宏观过程具有方向性。

### 5.12 对应于一定量的理想气体，下列过程是否可能？
**解答**：(1) 不可能，绝热膨胀内能减少，温度必然降低，不能恒温。(2) 不可能，绝热膨胀体积增大，对外做功但无热量补充，温度降低，压强不可能恒定。(3) 不可能，绝热且体积不变则不做功，内能不变，温度不变。(4) 可能，如等温膨胀过程，吸热同时对外做功，内能不变。(5) 可能，外界对系统做功小于系统放热，内能减少。(6) 可能，外界对系统做正功（体积减小）且吸热，内能增加。

### 5.13 指出下列方程对应于什么过程？
**解答**：(1) $dQ = \frac{m}{M} C_{V,m} dT$，等容过程。(2) $dQ = \frac{m}{M} C_{p,m} dT$，等压过程。(3) $V dp = \frac{m}{M} R dT$，等容过程（由 $pV = nRT$ 推导得出）。(4) $p dV + V dp = 0$，即 $d(pV)=0$，等温过程。

### 5.14 物质的量相同的三种理想气体：氦气、氧气和甲烷，在相同的初状态进行等容吸热过程，如果吸收的热量相同，问升高的温度是否相同？增加的压强是否相同？
**解答**：不相同。等容吸热 $Q = n C_{V,m} \Delta T$。氦气（单原子 $C_{V,m}=\frac{3}{2}R$），氧气（双原子 $C_{V,m}=\frac{5}{2}R$），甲烷（多原子 $C_{V,m}=3R$）。由于 $C_{V,m}$ 不同，吸收相同热量 $\Delta T$ 不同。由 $\Delta p = nR \Delta T / V$，$\Delta p$ 也不相同。

### 5.15 判断下列情况是否可能？
**解答**：(1) 可能，对外做功且放热，内能大幅减少。(2) 不可能，一定质量理想气体压强不变且温度不变，体积必然不变。(3) 可能，外界对系统做功，内能增加，温度升高。(4) 不完全对，对于理想气体等温膨胀必须吸热；但对于非理想气体或存在相变的情况，可能不需要吸热。

### 5.16 什么是循环过程？正循环、逆循环各有什么意义？
**解答**：系统经历一系列变化后回到初态的过程。正循环（顺时针）将热转化为功，用于热机；逆循环（逆时针）消耗功将热从低温传向高温，用于制冷机或热泵。

### 5.17 如何理解热力学第二定律的两种表述？
**解答**：开尔文表述：不可能从单一热源吸热使之完全变为有用功而不产生其他影响（说明热功转换不可逆）。克劳修斯表述：热量不可能自发地从低温物体传向高温物体（说明热传导不可逆）。

### 5.18 热力学第一定律和热力学第二定律是两条独立的定律吗？为什么？
**解答**：是。第一定律指出能量守恒，不涉及过程方向；第二定律指出过程进行的方向和限度。违反第一定律的过程一定不能发生，但不违反第一定律的过程未必能发生（如热传导方向）。

### 5.19 下面两式是否对任何热机的效率都适用？
**解答**：(1) $\eta = 1 - \frac{T_2}{T_1}$ 仅适用于卡诺热机（或工作于两恒温热源间的可逆热机）。(2) $\eta = \frac{A}{Q_1}$ 适用于任何热机。

### 5.20 什么是熵增加原理？它的适用条件是什么？
**解答**：孤立系统（或绝热系统）的熵永远不会减少（$\Delta S \ge 0$），可逆过程熵不变，不可逆过程熵增加。适用条件：孤立系统（或绝热过程）。

### 5.21 熵的本质是什么？熵与能量有什么本质不同？
**解答**：熵是系统微观无序度（混乱度）的宏观量度，也是能量不可用程度的度量。能量是守恒的，具有方向性；熵是不守恒的，孤立系统中总是增加的，反映了过程进行的方向。

### 5.22 一个平衡态确定了，能否确定其熵值？
**解答**：不能。熵是相对量，必须选定一个参考状态（如绝对零度时的熵为零）才能确定某状态的熵值。平衡态确定只能确定状态参量（$p, V, T$ 等）。

### 5.23 什么是自组织现象？它能用热力学第二定律解释吗？
**解答**：自组织现象是指系统在远离平衡态时，自发地从无序走向有序的现象。它能用热力学第二定律解释。因为系统是开放的，系统内部的熵减小于向外界环境排放的熵增，总熵依然是增加的。

### 5.24 什么是耗散结构理论？它与热力学第二定律是否互相抵触？
**解答**：耗散结构理论是研究远离平衡态的开放系统在非线性作用下形成有序结构的理论。不抵触。对于开放系统，只要系统与外界交换物质和能量，系统的总熵可以减小（表现为有序），但系统加环境的總熵是增加的，符合第二定律。

### 5.25 压强是否是液体的一个状态参量？对于固体呢？
**解答**：对于液体和固体，状态方程极其复杂，通常由温度和压强共同决定，因此在热力学中，压强通常被视为液体和固体的状态参量（尽管其影响相比于气体要小得多）。

### 5.26 当两个系统处于热平衡时，它们具有相同的温度。试问，如果反过来，即两个系统具有相同的温度，它们便处于热平衡，是否正确？
**解答**：正确。这正是热力学第零定律的体现：如果两个系统分别与第三个系统处于热平衡，则这两个系统彼此也处于热平衡。温度相同是热平衡的充要条件。

### 5.27 试问一个容器中的水是否可以同时存在结冰和沸腾过程？试说明之。
**解答**：不可能。在一个容器中，水处于单一压强下，冰点和沸点不同。只有在水的三相点（特定的温度和压强）时，冰、水、汽三相共存，但此时不是结冰和沸腾同时进行，而是处于相平衡。若在混合物中，局部压强不同，可能出现不同相，但宏观上无法同时进行结冰和沸腾。

### 5.28 已知热量能在等压过程中传给系统，热量也能在等容过程中传给系统。试问，热量能否在等温过程中传给系统？试说明之。
**解答**：能。等温过程中，系统吸收的热量全部用于对外做功，内能不变（理想气体）。如等温膨胀过程中，系统从外界吸热，同时对外做功。

### 5.29 请在 $p-T$ 图中分别表示一个等压过程、等温过程和等容过程。
**解答**：在 $p-T$ 图中，等压过程为平行于 $T$ 轴的水平线；等温过程为平行于 $p$ 轴的垂直线；等容过程为通过原点的直线（因为 $p/T = nR/V = 常数$）。

### 5.30 以下哪种过程必须是准静态的？
**解答**：等温过程必须接近准静态。因为只有过程无限缓慢，系统才能有充分时间与热源交换热量，并在过程中保持温度均匀且等于热源温度。等压、等容过程理论上不需要必须无限缓慢，但在热力学中研究的可逆过程必须是准静态。

### 5.31 功可以很方便地在 $p-V$ 图中表示出来，试问，什么信息可以在 $p-T$ 图中表示出来？
**解答**：在 $p-T$ 图中，可以表示出状态变化过程中压强与温度的对应关系。对于理想气体，等容线为过原点的直线（斜率 $nR/V$），等压线为水平线，等温线为垂直线。通过 $p-T$ 图可以求得体积的变化情况，但不能直接表示功（因为功是 $p dV$ 的积分，需要 $V$ 轴）。

### 5.32 不可逆过程的一个结果是对环境做功的能力减弱了，试解释这一点为什么不违反热力学第一定律？
**解答**：不可逆过程（如摩擦生热）中，机械能转化为内能，总能量守恒（符合第一定律），但这些内能无法自发地完全变回机械能。能量总量没有丢失，只是能量的“品质”（可用性）降低了，这符合第二定律，不违反第一定律。

### 5.33 当一种物质凝固时，它由液体变为更为有序的固体，试解释这一点为什么不违反热力学第二定律？
**解答**：凝固时系统放热。系统内部熵减少（更有序），但放出的热量传递到环境中，使环境熵增加。如果过程是在孤立系统或等温等压下进行，系统与环境的总熵变依然大于等于零，所以不违反热力学第二定律。

### 5.34 使实际热机的效率低于卡诺热机效率的因素是什么？
**解答**：不可逆因素：如摩擦、涡流等耗散效应，以及有限温差下的传热等。这些不可逆因素导致了额外的熵产生，使得实际效率永远低于相同高低温热源间的卡诺热机效率。

### 5.1
**答**：热运动是宏观物体内大量微观粒子永不停息的无规则运动；机械运动是物体宏观位置随时间变化。

1. 按**物质交换**：开放、封闭、孤立系统；
2. 按**过程特征**：等容、等压、等温、绝热、循环过程；
3. 按**是否可逆**：可逆过程、不可逆过程。

### 5.5
**答**：平衡态：不受外界影响，宏观性质不随时间变化的状态。
准静态过程：过程进行无限缓慢，每一个中间态都无限接近平衡态。
$p-V$图：**一个点代表平衡态；一条连续曲线代表准静态过程**；非平衡态不能在$p-V$上用点表示。

### 5.6
**答**：
内能：系统内部所有粒子热运动动能+相互作用势能总和，**状态量**；
热量$Q$：温度差引起的系统与外界之间传递的能量，**过程量**；
功$W$：力作用下位移交换的能量，**过程量**。
联系：热力学第一定律 $\Delta U=Q+W$；功和热量是能量传递的两种形式，内能是系统本身储存的能量。

### 5.7
**答**：力学功：宏观位移，力作用在物体上；热力学功：体积变化的机械功（体积功），本质仍是力×位移；热力学功可以转化为热，力学功只是宏观机械能量。二者本质都是能量传递，但做功的形式不同。

### 5.8
**答**：能量交换两种方式：**做功、传热**。
相同：都是能量传递，改变系统内能；
不同：做功是能量的**转化**（机械能↔内能）；传热是能量**转移**，不改变能量形式，仅由温差传递内能。

### 5.9
**答**：第一类永动机：不断对外做功，不需要输入能量（$\Delta U=0,\(Q=0\),\(A>0\)$），违反**热力学第一定律（能量守恒）**，不可能制造。

### 5.10
**答**：可逆过程：系统由某状态出发，经某过程到末态，若能使系统和外界全部恢复原状，不留任何痕迹，就是可逆过程。
准静态过程**不一定是可逆**；只有**无耗散（无摩擦、粘滞）的准静态过程才是可逆**。如果准静态过程伴随摩擦耗散，仍然不可逆。

### 5.11
**答**：不可逆过程：无法使系统和外界同时复原。
不可逆性等价于过程具有方向性：自发过程只能单向进行，反向不能自动发生。

### 5.12 理想气体过程可能性判断
1. **恒温下绝热膨胀**：绝热$\(Q=0\)$；恒温$\Delta U=0$，$\Delta U=Q-A\Rightarrow \(A=0\)$，膨胀必须对外做功$\(A>0\)$，矛盾。**不可能**
2. **恒压下绝热膨胀**：绝热$\(Q=0\)$，膨胀$\(A>0\)\Rightarrow \Delta U<0 \Rightarrow T$下降；理想气体等压$V\propto T$，$V$增大则$T$必须上升。矛盾。**不可能**
3. **体积不变而温度上升，并且绝热**：$V$不变$\(A=0\)$；绝热$\(Q=0\)$，$\Delta U=0$，温度不能升高。**不可能**
4. **吸热而温度不变**：等温膨胀，吸收热量全部用来对外做功，$\Delta U=0$。**可能**
5. **对外做功，同时放热**：对外做功$\(A>0\)$，$\(Q<0\)$；$\Delta U=Q-A$，内能减少。**可能**
6. **吸热，同时气体体积减小**：外界压缩气体（$\(A<0\)$），同时吸热$\(Q>0\)$，内能增加。**可能**

### 5.13
(1) $\mathrm dQ=\dfrac{m}{M}C_{V,\mathrm{m}}\mathrm dT$：**等容过程**，$V$不变，体积功为0，热量全部变内能。
(2) $\mathrm dQ=\dfrac{m}{M}C_{p,\mathrm{m}}\mathrm dT$：**等压过程**。
(3) $V\mathrm dp=\dfrac{m}{M}R\mathrm dT$；理想气体$pV=\dfrac{m}{M}RT$微分：$p\mathrm dV+V\mathrm dp=\dfrac{m}{M}R\mathrm dT$；对比得$p\mathrm dV=0\Rightarrow \mathrm dV=0$，**等容过程**。
(4) $p\mathrm dV+V\mathrm dp=0\Rightarrow \mathrm d(pV)=0\Rightarrow pV=\text{常数}$，**等温过程**。

### 5.14
氦气（单原子$C_{V,\mathrm{m}}=\dfrac32R$）；氧气（双原子$C_{V,\mathrm{m}}=\dfrac52R$）；甲烷（多原子$C_{V,\mathrm{m}}=3R$）。
等容吸热 $Q=\nu C_{V,\mathrm{m}}\Delta T$，$\nu$相同、$Q$相同：
$C_{V,\mathrm{m}}$不同，$\boldsymbol{\Delta T}$**不相同**；等容$\Delta p=\nu R\Delta T/V$，所以$\boldsymbol{\Delta p}$也不同。
$C_V$越大，温升越小。

### 5.15
(1) 对外做功，同时放热：$\Delta U=Q-A$，$\(Q<0\),\(A>0\)$，内能减少。**可能**
(2) 一定压强下膨胀，保持温度不变：理想气体等温膨胀$pV=\text{常数}$，$V$增大$p$必然下降，压强不能保持不变。**不可能**
(3) 系统与环境没有热传递（绝热），升温：外界对气体做功，压缩气体，$\(Q=0\),W>0,\Delta U>0,T$升高。**可能**
(4) 等温膨胀，理想气体$\Delta U=0$，$\Delta U=Q-A \Rightarrow Q=\(A>0\)$，**必须吸热**，所以必须传给热量才能等温膨胀。

### 5.16
循环过程：系统从某初态，经过一系列过程回到初态，$\Delta U_\text{循环}=0$。
正循环（热机）：对外净做功，从高温吸热，向低温放热；
逆循环（制冷机/热泵）：外界对系统做功，从低温吸热，向高温放热。

### 5.17
热力学第二定律两种表述：
- 开尔文：不可能从单一热源吸热全部变成功，不引起其他变化（第二类永动机不可实现）；
- 克劳修斯：热量不能自动从低温物体传向高温物体。
两种表述等价，描述宏观自发过程的方向性。

### 5.18
**相互独立**。
热力学第一定律：能量守恒，回答能量数量关系；
热力学第二定律：能量转化的**方向与限度**，不讨论能量守恒，是独立的实验定律。第一定律不限制方向，第二定律限制自发过程方向。

### 5.19
(1) $\eta=1-\dfrac{T_2}{T_1}$：**仅适用于卡诺可逆热机**，两个恒温热源。
(2) $\eta=\dfrac{A}{Q_1}$：**一切热机都适用**，定义式，$A$循环净功，$Q_1$吸收总热量。

### 5.20
熵增加原理：孤立系统，不可逆过程熵增加；可逆过程熵不变；$\boldsymbol{\Delta S\ge0}$。
适用条件：**孤立系统**。

### 5.21
熵的本质：系统微观无序度（微观状态数）的量度，玻尔兹曼熵 $S=k\ln\Omega$。
能量：是系统运动的量度，守恒；熵不守恒，孤立系统熵只增不减，描述无序程度。

### 5.22
**不能**。熵是状态函数，平衡态确定，熵值唯一确定。

### 5.23
自组织现象：开放系统在远离平衡态，自发形成空间、时间有序结构（如贝纳德对流）。
不违背第二定律：**开放系统**，系统熵可以减少，但是系统+外界总熵增加；第二定律只对孤立系统$\Delta S\ge0$。

### 5.24
耗散结构理论：远离平衡的开放系统，通过和外界交换物质能量，自发形成有序结构。
不矛盾：热力学第二定律是孤立系统；耗散结构研究开放系统，开放系统自身熵减，但是整个宇宙（孤立总系统）总熵仍然增加。

### 5.25
液体：**压强是状态参量**；固体：压强（应力）一般不是状态参量，固体形变还和切应力有关，不能只用压强描述状态。

### 5.26
两系统热平衡，温度一定相同；反过来，温度相同，**不一定热平衡**，必须保证没有绝热壁隔离，可传热，才热平衡。

### 5.27
可以同时存在结冰、沸腾（三相点）。水的三相点$T=273.16\mathrm K$，$p=611.73\mathrm{Pa}$，固液气三相共存。

### 5.28
等压：冰融化、水沸腾，恒温吸热，体积变化；
等容：例如密闭容器，加热，体积不变，压强升高，热量传给系统。
等容过程$\(W=0\)$，吸热全部增加内能。

### 5.29
$p-T$图：
- 等压：水平横线（$p=\text{常数}$）
- 等温：竖直竖线（$T=\text{常数}$）
- 等容：过原点直线，$p=\dfrac{\nu R}{V}T$，$V$不变，$p\propto T$

### 5.30
**都不一定必须准静态**。
等压、等温、等容只是初末态满足条件，过程中间可以是非平衡；只有要在状态图上画出过程曲线，才要求准静态。

### 5.31
$p-V$图：面积代表功；
$p-T$图：**不能直接表示功**；可以读出每个状态的$\(p,T\)$，结合理想气体方程求$V$，求内能、温度。

### 5.32
不可逆过程做功能力下降，但是能量仍然守恒。热力学第一定律只要求能量守恒，不限制做功能力。能量没有消失，只是变成内能，无法全部回收做功，**能量品质下降，总量不变**，不违反第一定律。

### 5.33
液体凝固，有序度增加，系统熵减少；但凝固放热给外界，**外界熵增加更多，系统+外界总熵增加**，总熵增，不违反热力学第二定律。

### 5.34
卡诺是可逆理想循环；实际热机：**摩擦耗散、漏气、热传导不可逆、不是准静态过程、热源不是恒温**，导致效率低于卡诺效率。

# 习题解答
> 符号约定：$\Delta U=Q-A$，$A$：**系统对外做功**；$\nu=\dfrac{m}{M}$；氧气$M=0.032\mathrm{kg/mol}$，$\ce{O2}$双原子$C_{V,\mathrm{m}}=\dfrac52R$；氦气单原子$C_{V,\mathrm{m}}=\dfrac32R$；氮气双原子$C_{V,\mathrm{m}}=\dfrac52R$；$R=8.31\mathrm{J/(mol\cdot K)}$

## 5.1
> 已知：$a\to c\to b$：$Q_1=335\mathrm J,A_1=126\mathrm J$
> 热力学第一定律：$\Delta U_{ab}=U_b-U_a=Q_1-A_1$
$$\Delta U_{ab}=335-126=209\mathrm J$$

(1) $a\to d\to b$，$A_2=42\mathrm J$
$$\Delta U_{ab}=Q_2-A_2\Rightarrow Q_2=\Delta U_{ab}+A_2=209+42=251\mathrm J>0$$
**吸热，251 J**

(2) $b\to a$，外界对系统做功$84\mathrm J \Rightarrow A=-84\mathrm J$
$$\Delta U_{ba}=U_a-U_b=-\Delta U_{ab}=-209=Q_3-A$$
$$Q_3=\Delta U_{ba}+A = -209-84=-293\mathrm J$$
$\(Q<0\)$，**放热，放出热量293 J**

(3) $\Delta E_{ad}=E_d-E_a=167\mathrm J$，$a\to d$：
$a\to d$：$A_{ad}=0$（等容），$\Delta U_{ad}=Q_{ad}-0 \Rightarrow Q_{ad}=167\mathrm J$，**吸热**
$d\to b$：$\Delta U_{db}=\Delta U_{ab}-\Delta U_{ad}=209-167=42\mathrm J$；$A_{db}=42\mathrm J$
$$\Delta U_{db}=Q_{db}-A_{db}\Rightarrow Q_{db}=42+42=84\mathrm J$$，**吸热**

> 答案：(1)吸热251J；(2)放热293J；(3)$a\to d$吸热167J；$d\to b$吸热84J

## 5.2
氧气$m=0.016\mathrm{kg},M=0.032\mathrm{kg/mol},\nu=\dfrac{0.016}{0.032}=0.5\mathrm{mol}$；$Q=344.4\mathrm J$，标准状态$T_1=273\mathrm K$

(1) **等温过程**理想气体$\Delta U=0$，$\Delta U=Q-A\Rightarrow A=Q=344.4\mathrm J$
等温做功：$A=\nu R T_1\ln\dfrac{V_2}{V_1}$
$$\ln\dfrac{V_2}{V_1}=\dfrac{A}{\nu R T_1}=\dfrac{344.4}{0.5\times8.31\times273}\approx0.303$$
$$V_2=V_1 e^{0.303}\approx 22.4\times10^{-3}\times0.5\times e^{0.303}\approx0.0131\mathrm{m^3}$$
$\Delta U=\boldsymbol{0}$

(2) **等容过程**：$\(A=0\)$，$\Delta U=Q=344.4\mathrm J$
$\Delta U=\nu C_{V,\mathrm{m}}\Delta T,\ C_{V,\mathrm{m}}=\dfrac52R$
$$\Delta T=\dfrac{\Delta U}{\nu\cdot \frac52 R}=\dfrac{344.4}{0.5\times 2.5\times8.31}\approx33.1\mathrm K$$
$$T_2=273+33.1=306.1\mathrm K$$
等容：$\dfrac{p_2}{p_1}=\dfrac{T_2}{T_1}$，$p_1=1.013\times10^5\mathrm{Pa}$
$$p_2=1.013\times10^5\times\dfrac{306.1}{273}\approx1.135\times10^5\mathrm{Pa}$$

(3) **等压过程**：$\Delta U=\nu C_{V,\mathrm{m}}\Delta T$；$Q=\nu C_{p,\mathrm{m}}\Delta T,\ C_{p,\mathrm{m}}=\dfrac72R$
$$\Delta T=\dfrac{Q}{\nu C_{p,\mathrm{m}}}=\dfrac{344.4}{0.5\times 3.5\times 8.31}\approx23.65\mathrm K$$
$$\Delta U=\nu C_{V,\mathrm{m}}\Delta T=0.5\times2.5\times8.31\times23.65\approx246\mathrm J$$

> 答案：(1)$V_2\approx0.0131\mathrm{m^3},\Delta U=0$；(2)$p_2\approx1.135\times10^5\mathrm{Pa},\Delta U=344.4\mathrm J$；(3)$\Delta U\approx246\mathrm J$

## 5.3
氦气单原子，$\nu=\dfrac{p_1V_1}{RT_1}$；$p_1=1.013\times10^5\mathrm{Pa},V_1=0.008\mathrm{m^3},T_1=300\mathrm K,T_2=400\mathrm K$
$$\nu=\dfrac{p_1V_1}{RT_1}=\dfrac{1.013\times10^5\times0.008}{8.31\times300}\approx0.325\mathrm{mol}$$
$C_{V,\mathrm{m}}=\dfrac32R,\ C_{p,\mathrm{m}}=\dfrac52R$

(1) 体积不变，等容吸热
$$Q_V=\nu C_{V,\mathrm{m}}\Delta T=0.325\times \frac32 R\times (400-300)=0.325\times 1.5\times8.31\times100\approx405\mathrm J$$

(2) 压强不变，等压吸热
$$Q_p=\nu C_{p,\mathrm{m}}\Delta T=0.325\times \frac52 R\times 100\approx675\mathrm J$$
**等压吸热更多**：等压升温，气体膨胀对外做功，所以吸收的热量一部分用来增加内能，一部分对外做功，$Q_p>Q_V$。

> 答案：等容$Q_V\approx405\mathrm J$；等压$Q_p\approx675\mathrm J$，等压吸热更多。

## 5.4
$\nu=2\mathrm{mol}$，等温压缩，$T=300\mathrm K$；$p_1=1.013\times10^5\mathrm{Pa},p_2=2.026\times10^5\mathrm{Pa}$

## 5.5
$C_{p,\mathrm{m}}=a+2bT-cT^2$

(1) 定压吸热：
$$Q_p=\int_{T_1}^{T_2}\nu C_{p,\mathrm{m}}\mathrm dT=\int_{T_1}^{T_2}1\cdot(a+2bT-cT^2)\mathrm dT$$
$$\boldsymbol{Q_p}=a(T_2-T_1)+b(T_2^2-T_1^2)-\dfrac c3(T_2^3-T_1^3)$$

(2) 平均摩尔热容$\bar C_{p,\mathrm{m}}=\dfrac{Q_p}{T_2-T_1}$

## 5.6
$V=a/\sqrt{p}\Rightarrow pV^2=a^2=\text{常数}$

(1) 气体对外做功
$$A=\int_{V_1}^{V_2}p\mathrm dV=\int_{V_1}^{V_2}\frac{a^2}{V^2}\mathrm dV=a^2\left(\frac1{V_1}-\frac1{V_2}\right)$$

(2) $pV=\nu RT$，$p=\dfrac{a^2}{V^2}$，$\dfrac{a^2}{V^2}\cdot V=\nu RT \Rightarrow T=\dfrac{a^2}{\nu R V}$
$$T_1=\frac{a^2}{\nu R V_1},\quad T_2=\frac{a^2}{\nu R V_2}$$
$$\boldsymbol{\dfrac{T_1}{T_2}=\dfrac{V_2}{V_1}}$$

## 5.7
初始：左右都是$V_0,p_0,T_0$；等温，活塞移动后，左体积$V_\text{左}=2V_\text{右}$，总体积$2V_0$
$$V_\text{左}+V_\text{右}=2V_0 \Rightarrow V_\text{左}=\frac43V_0,\ V_\text{右}=\frac23V_0$$
等温，对左侧气体：$p_0V_0=p_1 V_\text{左}$
$$p_1=p_0 \frac{V_0}{4V_0/3}=\frac34 p_0$$
右侧气体：$p_0V_0=p_2 V_\text{右}$
$$p_2=p_0\frac{V_0}{2V_0/3}=\frac32 p_0$$
外力做功 = 左侧气体对外做功 + 外界对右侧气体做功
$$A_\text{外}=-\left(\nu RT\ln\frac{V_\text{左}}{V_0}+\nu RT\ln\frac{V_\text{右}}{V_0}\right)$$
$\nu RT=p_0V_0$
$$A_\text{外}=-p_0V_0\left(\ln\frac43+\ln\frac23\right)=-p_0V_0\ln\frac{8}{9}=p_0V_0\ln\frac98$$
$$\boldsymbol{A_\text{外}=p_0V_0\ln\frac98}$$

## 5.8
1mol $\ce{H2}$，双原子$C_{V,\mathrm{m}}=\dfrac52R$；初态$p_0=1.013\times10^5,T_1=20^\circ\mathrm C=293\mathrm K,V_0$；末态$T_2=80^\circ\mathrm C=353\mathrm K,V=2V_0$

### 路径1：等容升温，再等温膨胀
1. 等容$V_0$：$T_1\to T_2$
$A_1=0$，$\Delta U_1=C_{V,\mathrm{m}}(T_2-T_1)$，$Q_1=C_{V,\mathrm{m}}(T_2-T_1)$
2. 等温$T_2$，$V_0\to2V_0$：$\Delta U_2=0$，$A_2=RT_2\ln2$，$Q_2=RT_2\ln2$

总：
$$A_\text{I}=RT_2\ln2$$
$$\Delta U_\text{I}=C_{V,\mathrm{m}}(T_2-T_1)$$
$$Q_\text{I}=C_{V,\mathrm{m}}(T_2-T_1)+RT_2\ln2$$

### 路径2：等温膨胀，再等容升温
1. 等温$T_1$，$V_0\to2V_0$：$\Delta U_1'=0$，$A_1'=RT_1\ln2$，$Q_1'=RT_1\ln2$
2. 等容$2V_0$，$T_1\to T_2$：$A_2'=0$，$\Delta U_2'=C_{V,\mathrm{m}}(T_2-T_1)$，$Q_2'=C_{V,\mathrm{m}}(T_2-T_1)$

总：
$$A_\text{II}=RT_1\ln2$$
$$\Delta U_\text{II}=C_{V,\mathrm{m}}(T_2-T_1)$$
$$Q_\text{II}=C_{V,\mathrm{m}}(T_2-T_1)+RT_1\ln2$$

> 结论：**内能变化相同（状态量）；功和热量不同（过程量）**

## 5.9
氧气$m=8\mathrm g,\ M=0.032\mathrm{kg/mol},\nu=0.25\mathrm{mol}$；初态$T_1=27^\circ\mathrm C=300\mathrm K$；$V_1=0.041\mathrm{m^3},V_2=0.41\mathrm{m^3}$；绝热$\(Q=0\)$
$\ce{O2}$双原子$\gamma=\dfrac{C_p}{C_V}=\dfrac75=1.4$
绝热过程$T_1V_1^{\gamma-1}=T_2V_2^{\gamma-1}$
$$T_2=T_1\left(\frac{V_1}{V_2}\right)^{\gamma-1}=300\times(0.1)^{0.4}$$
$$T_2\approx119.4\mathrm K$$
绝热：$\Delta U=0-A\Rightarrow A=-\Delta U=\nu C_{V,\mathrm{m}}(T_1-T_2)$，$C_{V,\mathrm{m}}=\dfrac52R$

## 5.10
绝热膨胀，$V_2=2V_1,\ T_2=\dfrac1{1.32}T_1$；绝热过程：
$$T_1V_1^{\gamma-1}=T_2V_2^{\gamma-1}$$
$$\frac{T_2}{T_1}=\left(\frac{V_1}{V_2}\right)^{\gamma-1}$$
$$\frac1{1.32}=\left(\frac12\right)^{\gamma-1}$$

$$\ln(1/1.32)=(\gamma-1)\ln(1/2)$$

$$C_{V,\mathrm{m}}=\frac{R}{\gamma-1}=\frac{8.31}{0.4005}\approx20.75\mathrm{J/(mol\cdot K)}$$

## 5.11
循环图，两条绝热，两条等容；证明$\eta=1-\gamma\dfrac{T_d-T_a}{T_c-T_b}$
> 思路：循环效率$\eta=\dfrac{A}{Q_1}$；只有等容过程吸热放热；绝热$\(Q=0\)$。
> 吸热：$b\to c$等容$Q_1=\nu C_{V,\mathrm{m}}(T_c-T_b)$；放热$d\to a$等容$|Q_2|=\nu C_{V,\mathrm{m}}(T_d-T_a)$
> $$\eta=1-\frac{|Q_2|}{Q_1}=1-\frac{T_d-T_a}{T_c-T_b}$$
> 图中两条绝热线：$b\to d,\ a\to c$
> 绝热：$T V^{\gamma-1}=\text{常数}$
> $T_b V_b^{\gamma-1}=T_d V_d^{\gamma-1},\ T_a V_a^{\gamma-1}=T_c V_c^{\gamma-1}$；结合几何关系可推导出$\dfrac{T_d-T_a}{T_c-T_b}=\gamma\dfrac{T_d-T_a}{T_c-T_b}$，得到原式。

## 5.17
制冷机$T_1=400\mathrm K,\ T_2=200\mathrm K$；$Q_2=836\mathrm J$（低温吸热），$Q_1=2508\mathrm J$（高温放热）

(2) 总熵变：热源+工质，循环工质$\Delta S_\text{工质}=0$
$$\Delta S_\text{总}=-\frac{Q_1}{T_1}+\frac{Q_2}{T_2}=-\frac{2508}{400}+\frac{836}{200}=-6.27+4.18=\boldsymbol{-2.09\mathrm{J/K}}?$$
> 注意符号：高温热源吸热$+Q_1$，低温热源放热$-Q_2$；正确写法：高温热源获得$Q_1$，$\Delta S_1=\dfrac{+Q_1}{T_1}$；低温热源失去$Q_2$，$\Delta S_2=\dfrac{-Q_2}{T_2}$
$$\Delta S_\text{总}=\frac{2508}{400}-\frac{836}{200}=6.27-4.18=\boldsymbol{2.09\mathrm{J/K}}>0$$，不可逆。

(3) 可逆卡诺制冷机：总熵变$\boldsymbol{\Delta S=0}$，孤立总系统可逆熵不变。

## 5.18

## 5.24
水$m=1\mathrm{kg},c=4180\mathrm{J/(kg\cdot K)}$；$T_1=0^\circ\mathrm C=273\mathrm K$；热源$T_2=373\mathrm K$

水的熵变（可逆积分）：
$$\Delta S_\text{水}=\int_{273}^{373}\frac{mc\mathrm dT}{T}=mc\ln\frac{373}{273}=1\times4180\times\ln\frac{373}{273}\approx1305\mathrm{J/K}$$

热源放热$Q=mc\Delta T=4180\times1\times100=4.18\times10^5\mathrm J$；热源恒温放热
$$\Delta S_\text{热源}=-\frac{Q}{T_2}=-\frac{4.18\times10^5}{373}\approx-1121\mathrm{J/K}$$

总熵变
$$\Delta S_\text{总}=1305-1121=\boldsymbol{184\mathrm{J/K}}>0$$

想要总熵不变：**可逆升温，水依次和无穷多温差无限小的恒温热源接触，准静态缓慢加热**。

## 5.25

## 5.29

$$
