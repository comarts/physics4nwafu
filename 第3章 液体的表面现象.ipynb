### 3.2
昆虫体重很小，脚压在水面，表面张力的向上分力可以平衡昆虫重力；
人的体重巨大，重力远大于水表面张力能提供的最大支撑力，所以人不能在水上行走。

### 3.3
肥皂泡有内外两层液膜，附加压强$\Delta p=\dfrac{4\alpha}{R}$，泡内部压强 = 外界大气压$+\dfrac{4\alpha}{R}$。
停止吹气时，泡内压强大于管外大气压，气体就从泡内沿管子流出，肥皂泡体积缩小。

### 3.4
总表面能**增大**。
一滴大水珠分成很多小水滴，**总表面积变大**；表面能$E=\alpha S$，$\alpha$不变，表面积$S$增大，总表面能增加。（分裂过程需要外界做功）

### 3.5
小肥皂泡不断缩小，大肥皂泡不断变大，直到小泡收缩到只剩一层薄膜贴在管口。
原因：肥皂泡附加压强$\Delta p=\dfrac{4\alpha}{R}$，半径越小，泡内压强越大；气体从小泡（高压）流向大泡（低压）。

### 3.6
扩大液体表面做功，转化为**液体的表面能**；
拉开橡皮膜做功转化为**橡皮膜的弹性势能**。二者本质不同：液体表面能来源于分子间作用力，橡皮膜是固体弹性形变。

### 3.7
液体表面层分子受力不对称，存在表面张力；当固体和液体分子间附着力与液体内聚力不相等时，表面张力使液面发生弯曲。

### 3.8
雾是小液滴。根据弯曲液面附加压强，**小液滴饱和蒸气压比平面液体高**；秋冬气温低，饱和蒸气压下降，空气中水汽容易以凝结核为基础凝结成小液滴，形成雾。

### 3.9
- 接触角$\boldsymbol{\theta}$：固‑液接触面的切线与液体表面切线，在液体内部测量的夹角。
- 润湿：$\theta<90^\circ$；不润湿：$\theta>90^\circ$。
微观：附着力（固‑液分子引力）>内聚力（液‑液分子引力）→润湿；附着力<内聚力→不润湿。

### 3.10
气体栓塞：液体细管中，气泡阻碍液体流动的现象。
产生：润湿液体在毛细管，气泡两端液面曲率不同，产生方向相反的附加压强；多个气泡附加压强叠加，阻碍液体流动。输液、输油管要避免气泡。

### 3.11
凸液面分子更容易逃出。
凸液面分子周围分子少，受到的向内吸引力小，束缚弱，更容易脱离液面蒸发；凹液面分子受到向内的引力大，难逃出。

## 习题解答
### 3.1
液膜有前后**两个表面**。
$$F_1=2\alpha_1 L,\;F_2=2\alpha_2 L$$
合力：
$$
\begin{align*}
F &= 2L(\alpha_2-\alpha_1) \\
&=2\times 0.1\times(70\times10^{-3}-40\times10^{-3}) \\
&=\boldsymbol{6.0\times10^{-3}\ \mathrm{N}}
\end{align*}
$$
方向：指向表面张力系数大的一侧。

### 3.2
不能产生人造喷泉。
毛细管总上升高度$h=\dfrac{2\alpha\cos\theta_0}{\rho g r}$，$\theta_0=0$。
露出水面$\dfrac h2$，液面只能上升到管口$\dfrac h2$处，液面曲率改变，接触角变为$\theta$满足：
$$\frac h2=\frac{2\alpha\cos\theta}{\rho g r} \implies \cos\theta=\frac12 \implies \boldsymbol{\theta=60^\circ}$$
上升高度为$\boldsymbol{\dfrac h2}$，水不会喷出。

### 3.3
设大气压$p_0$，毛细管上升高度$h$。
- $A$：管内凹液面上方，$\boldsymbol{p_A=p_0}$
- $B$：凹液面下方，凹液面附加压强向下，$\boldsymbol{p_B=p_0-\dfrac{2\alpha}{r}}$
- $C$：管内与外面液面同一水平面，$\boldsymbol{p_C=p_0}$
- $D$：容器液面下同一水平面，$\boldsymbol{p_D=p_0}$

### 3.4
$d=5.0\times10^{-2}\,\mathrm{m},\;R=2.5\times10^{-2}\,\mathrm{m}$，肥皂泡双表面。
表面积 $S=2\times4\pi R^2$
做功：
$$
W=\alpha \Delta S=\alpha\cdot 8\pi R^2 =2.5\times10^{-2}\times8\pi\times(2.5\times10^{-2})^2 \approx \boldsymbol{3.93\times10^{-4}\ \mathrm{J}}
$$
内外压强差
$$
\Delta p=\frac{4\alpha}{R}=\frac{4\times2.5\times10^{-2}}{2.5\times10^{-2}}=\boldsymbol{4.0\ \mathrm{Pa}}
$$

### 3.5
等温，理想气体 $pV=\text{常数}$，外泡内压强$p_0+\dfrac{4\alpha}{r}$，内层小泡压强$p_0+\dfrac{4\alpha}{r_1}+\dfrac{4\alpha}{r_0}$。
(1) 由气体总量守恒：
$$\Big(p_0+\frac{4\alpha}{r_0}\Big)\frac43\pi r_0^3 +\Big(p_0+\frac{4\alpha}{r_1}\Big)\frac43\pi r_1^3
=\Big(p_0+\frac{4\alpha}{r}\Big)\frac43\pi r^3$$
$$\boldsymbol{\left(p_0+\frac{4\alpha}{r_0}\right)r_0^3+\left(p_0+\frac{4\alpha}{r_1}\right)r_1^3=\left(p_0+\frac{4\alpha}{r}\right)r^3}$$

(2) 吹第二个泡做功等于表面能增量
$$W=\Delta E=\alpha \big(8\pi r_1^2\big)=\boldsymbol{8\pi\alpha r_1^2}$$

### 3.6
平行板间毛细上升公式 $h=\dfrac{2\alpha\cos\theta}{\rho g d}$，$\theta=0,\cos\theta=1$，$d=0.5\times10^{-3}\,\mathrm{m}$
$$
h=\frac{2\times7.3\times10^{-2}}{1000\times9.8\times 0.5\times10^{-3}} \approx \boldsymbol{0.0298\ \mathrm{m}\approx 3.0\ \mathrm{cm}}
$$

### 3.7
$d=1.0\ \mathrm{mm},\;r=0.5\times10^{-3}\,\mathrm{m}$，$\theta=140^\circ,\cos140^\circ\approx-0.7660$，$\alpha=47\times10^{-2}\,\mathrm{N/m}$

### 3.9
降雨量$H=50\ \mathrm{mm}=0.05\,\mathrm{m}$；总体积$V=S\cdot H=50\times10^6 \times0.05=2.5\times10^6\ \mathrm{m^3}$。
每个雨滴体积$v_0=\dfrac43\pi r^3$，$r=1.0\times10^{-3}\,\mathrm{m}$。
雨滴总个数 $N=\dfrac{V}{v_0}$。
总表面积 $S_\text{总}=N\cdot4\pi r^2$；原来湖面面积$S_0=50\ \mathrm{km^2}$。
