## 流体力学简介



## 理想流体的定常流动

根据质量守恒定律，任取定常流动的一细流管，经过时间$\Delta t$，流入的流体质量 $\Delta m_1 = \rho_1S_1v_1\Delta t$ 与流出的质量 $\Delta m_2 = \rho_2S_2v_2\Delta t$ 相同

$$ \rho_1S_1v_1\Delta t = \rho_2S_2v_2\Delta t\tag{2.1a}$$
或
$$ \rho Sv=常量 \tag{2.1b}$$

该细流管任意垂直截面均满足式$(2.1\text{b})$

如果假设流体式不可压缩的，则它的密度不变，式$(2.1\text{b})$可以简化为
$$ Q=Sv=常量 \tag{2.2}$$

式中 $Q$ 为流管的流量。

上式称为流体的连续性原理。

## 伯努利方程及其应用

### 伯努利方程的推导

在做定常流动的理想流体中任取一根细流管，

$$ p_1 + \frac{1}{2}\rho v_1^2 + \rho g h_1 = p_2 + \frac{1}{2}\rho v_2^2 + \rho g h_2 \tag{2.3a}$$
或
$$ \rho + \frac{1}{2} \rho v^2 + \rho g h=常量 \tag{2.3b}$$
式$(2.3)a$或$(2.3b)$即伯努利方程


伯努利方程表明，在同一细流管的不同截面处，单位体积流体的动能 $\frac{1}{2}pv^2$、单位体积流体的势能$\rho gh$、单位体积流体的压强能$p$三项之和为常量。

在工程上，伯努利方程常写成：
$$ \frac{p}{\rho g} + \frac{v^2}{2g} + h=常量 \tag{2.4} $$
上式左边三项依次称为压强头、速度头和高度头。

## 黏性流体的定常流动
### 牛顿黏性定律

实验证明，实际流体在固体管中流动时，如果流速不是很大，流体将做分层流动，此时不同流层得流速不同，但流层之间只做相对滑动彼此不相混合，这种流动状态称为层流（laminar flow）。

若取层间距 $\Delta y \to 0$，$\Delta v$ 与 $\Delta y$ 之比得极限值就反映了流速得空间变化率，称为**速度梯度**，其定义为
$$ \lim_{\Delta y\to 0}\frac{\Delta v}{\Delta y}=\frac{\operatorname{d}v}{\operatorname{d}y}  $$

实验证明，相邻流层之间的摩擦系数 $F_f$ 的大小，与两流层的接触面积 $S$ 成正比，与速度梯度 $\frac{\operatorname{d}v}{\operatorname{d}y}$ 成正比，即有
$$ F_f=\eta \frac{\operatorname{d}v}{\operatorname{d}y}S \tag{2.10}$$

上式中的比例系数 $\eta$ 称为流体的黏度（viscosity），它的单位是 $\mathrm{Pa\cdot s}$，该公式称为**牛顿黏性定律**。

### 黏性流体的伯努利方程

若单位体积的黏性流体从 1 到 2 时摩擦损耗的机械能用 $\Delta w_{21}$ 表示，应用功能原理，就可以得到黏性流体的伯努利方程为
$$ p_1 + \frac{1}{2}\rho v_1^2 + \rho g h_1 = p_2 + \frac{1}{2}\rho v_2^2 + \rho g h_2 + \Delta w_{21}\tag{2.11}$$



### 湍流 雷诺数

实验表明，当流体的流速增大到一定数值后，层流将被破坏，相邻层流间不但有滑动，还有混合。这是流体做不规则运动，有垂直于流管轴线方向的分速度产生，这种运动称为湍流（turbulent flow）。流体做湍流时，阻力大流量小，能量损耗增加。实验证明，能量损耗 $\Delta E$ 与速度的关系为
$$ \Delta E=kv^2 \tag{2.12} $$
式中 $k$ 是比例常数，它与管道的形状、大小以及管道的材料有关，$v$ 是平均流速。

流体的流动状态由雷诺数 $\rho vl/\eta$ 决定，雷诺数小时做层流，雷诺数大时做湍流。流体由层流到湍流过渡的雷诺数，叫临界雷诺数（critical Reynolds number），记作 $Re$。对于圆形管道，引入
$$ Re=\frac{\rho v d}{\eta} \tag{2.13}$$

## 泊肃叶定律 斯托克斯定律

### 泊肃叶定律

黏性流体在粗细均匀的管道中做稳定流动时，流量与压强、黏度、管的粗细和长短等因素的关系。

圆管中实际流体的流速随半径的分布规律:

$$v=\frac{p_1-p_2}{4\eta l}\big(R^2-r^2\big) \tag{2.14}$$

泊肃叶方程：

$$ Q=\frac{\pi R^4}{8\eta l}(p_1-p_2)=\frac{\pi R^4}{8\eta l}\Delta p \tag{2.15}$$

达西定律：
若将 $R'=\frac{8\eta l}{\pi R^4}$ 代入式$(2.15)$，得
$$Q=\frac{p_1-p_2}{R'} \tag{2.16}$$

此式与电学中的欧姆定律相似，因此把 $R'$ 称为流体阻力系数(简称流阻)。式$(2.16)$是生命科学中常用的公式，称为达西定律(Darcy theorem)。



### 奥氏黏度计

$$\frac{\eta_2}{\eta_1}=\frac{\rho_2\, t_2}{\rho_1\, t_1}$$



## 思考题
### 思考题2.1
伯努利方程成立的条件是什么？泊肃叶方程成立的条件又是什么？

🏍️解答：
伯努利方程成立条件：理想流体、定常流动、同一流管。

### 思考题2.2
如何解释“巷小风大”？

🏍️解答：

巷道狭窄，空气流通截面积小，根据连续性方程 $S_1v_1=S_2v_2$，截面积变小流速增大；对应伯努利原理，流速大处压强低。

### 思考题2.3

如何解释“河道宽处水流缓，河道窄处水流急”？

🏍️解答：

河水近似不可压缩流体，连续性方程 $S v=\text{常量}$。河道宽处横截面积$S$大，则流速$v$小；河道窄处横截面积$S$小，则流速$v$大。

### 思考题2.4

定常流动是理想流体所特有的吗？黏性流体有无定常流动？

🏍️解答：

不是理想流体特有。黏性流体也可以发生定常流动：空间各点流速不随时间变化，只是层内存在内摩擦力，泊肃叶公式描述的圆管层流就是黏性流体的定常流动。

### 思考题2.5

平行行驶的两艘船为什么要保持一定距离？

🏍️解答：

两船内侧水流通道变窄，水流流速大，压强小；船外侧水流流速小、压强大。内外压强差会把两船相互推向一起，容易发生碰撞，因此必须保持间距。

### 思考题2.6

液体在管中流动时，从连续性原理来看，管子越粗，流速越小，而从泊肃叶方程来看，管子越粗，流量越大，这两者是否矛盾，为什么？

🏍️解答：

不矛盾。

连续性方程 $\(Sv=Q\)$：流量$Q$固定不变前提下，截面积$S$越大流速$v$越小。

泊肃叶公式 $Q=\dfrac{\pi R^4\Delta p}{8\eta l}$：管道两端压强差$\Delta p$固定不变前提下，半径越大流量$Q$越大。

两个公式的前提条件不一样。

水平的玻璃管，由截面均匀而截面积不等的A、B、C三段串联而成。水从A段进入，从C段流出。在三段管壁上各打一小孔，A段上小孔有水射出，B段小孔有气泡出现，C段小孔不射水，也无气泡，设水为理想流体。问三段管子哪段内径最大，哪段内径最小，为什么？

- C段压强等于大气压，流速介于两者之间。


### 思考题2.8

生物流体同一般流体有何区别？

🏍️解答：

1. 很多生物流体（血液）属于非牛顿流体，黏度会随流速/切变率改变，不服从牛顿黏性定律；

### 习题2.12

在水管的某一点$A$，水的流速为$1\ \mathrm{m\cdot s^{-1}}$，计示压强为$3\times10^5\ \mathrm{Pa}$。如果沿水管到另一点$B$，$B$点比$A$点高度降低了20 cm，$B$点水管的横截面积为$A$点的二分之一，求$B$点的计示压强。

🏍️解答：

已知条件

$v_A=1\ \mathrm{m/s}$，A点计压强 $p_{A示}=3\times 10^5\ \mathrm{Pa}$

B比A降低 $h_A-h_B=0.20\ \mathrm{m}$

$S_B=\frac12 S_A$，水密度 $\rho=1000\ \mathrm{kg/m^3}$

步骤1：连续性方程求$v_B$

$$S_A v_A = S_B v_B$$

$$v_B = v_A\cdot \frac{S_A}{S_B}=1\cdot \frac{S_A}{\frac12 S_A}= 2\ \mathrm{m/s}$$


步骤2：伯努利方程（计示压强，大气压$p_0$抵消）

$$p_{A示} \frac12\rho v_A^2+\rho g h_A = p_{B示} + \frac12\rho v_B^2+\rho g h_B$$
移项：
$$
p_{B示} = p_{A示} + \frac12\rho(v_A^2-v_B^2)+\rho g(h_A-h_B)
$$

代入数值：
$$
\begin{aligned}
\frac12\rho(v_A^2-v_B^2)&=0.5\times 1000\times(1^2-2^2)=500\times(-3)=-1500\ \mathrm{Pa}\\
\rho g(h_A-h_B)&=1000\times 9.8\times 0.20 = 1960\ \mathrm{Pa}
\end{aligned}
$$

$$
\begin{aligned}
p_{B示}&=3\times10^5 -1500 +1960 \\
&=300000 + 460 \\
&=3.0046\times 10^{5}\ \mathrm{Pa}
\end{aligned}
$$

### 习题2.33

人的心脏不断向大动脉泵送血液，进入大动脉的最大流量约为$500\ \mathrm{cm^3\cdot s^{-1}}$。假设大动脉的直径为2.5 cm，流动大致可认为是层流，血液是牛顿流体。求：(1)大动脉内单位长度上的压强差；(2)轴心处的流动速度。已知血液黏度$\eta=4\times10^{-3}\ \mathrm{Pa\cdot s}$。

🏍️解答：

$Q=500\mathrm{cm^3/s}=5\times10^{-4}\mathrm{m^3/s}$；$D=2.5\mathrm{cm}$，$R=0.0125\mathrm{m}$；$\eta=4\times10^{-3}\mathrm{Pa\cdot s}$。
泊肃叶 $Q=\dfrac{\pi R^4}{8\eta}\cdot \dfrac{\Delta p}{L}$
1. 单位长度压强差 $\dfrac{\Delta p}{L}$
$$\frac{\Delta p}{L}=\frac{8\eta Q}{\pi R^4}=\frac{8\times4\times10^{-3}\times5\times10^{-4}}{\pi\times(0.0125)^4}\approx \boldsymbol{167\ \mathrm{Pa/m}}$$

2. 层流平均流速 $\bar v=\dfrac{Q}{\pi R^2}$；轴心最大流速 $v_\text{max}=2\bar v$
$$\bar v=\frac{5\times10^{-4}}{\pi\times0.0125^2}\approx 1.019\mathrm{m/s}$$
$$v_\text{max}=2\bar v=\boldsymbol{2.04\ \mathrm{m/s}}$$



### 习题2.34

在其他参量不变的条件下，要使血管的流体力学阻尼(用$\Delta p/Q$表示)增大30%，半径需要改变多少？假设该情况为牛顿流体的层流行为。

🏍️解答：

流体阻尼 $Z=\dfrac{\Delta p}{Q}$；泊肃叶 $Q=\dfrac{\pi R^4 \Delta p}{8\eta L}$，$Z=\dfrac{8\eta L}{\pi R^4}$。
$Z \propto R^{-4}$；$Z_2=1.3 Z_1$。
$$\frac{Z_2}{Z_1}=\left(\frac{R_1}{R_2}\right)^4=1.3$$
$$\frac{R_2}{R_1}=1.3^{-1/4}\approx 0.933$$
半径变为原来的0.933倍；**半径减小约6.7%**。
