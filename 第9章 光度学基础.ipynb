### 思考题解答

**9.1 试述辐射度量与光度量联系和区别。**
*   **解题思路：** 从物理客观量和生理主观量两个角度区分。
*   **答案：** 
    *   **联系：** 两者都是描述光辐射能量的物理量，光度量是由辐射度量经过人眼视觉特性（光谱光视效率函数）加权换算得来的。
    *   **区别：** 辐射度量是纯物理量，客观描述辐射能量的多少，单位如瓦特（W）；光度量是生理心理物理量，考虑了人眼对不同波长光的敏感程度，单位如流明（lm）、坎德拉（cd）等。

**9.2 简述发光强度、亮度、光出射度、照度等定义及其单位。**
*   **解题思路：** 回忆各光度学基本概念的定义及国际单位制。
*   **答案：**
    *   **发光强度（\(I\)）：** 光源在给定方向上单位立体角内发出的光通量。单位：坎德拉（cd）。
    *   **亮度（\(L\)）：** 光源单位面积在给定方向上的发光强度。单位：坎德拉每平方米（\(\text{cd/m}^2\)）或尼特（nt）。
    *   **光出射度（\(M\)）：** 离开光源单位面积的光通量。单位：流明每平方米（\(\text{lm/m}^2\)）。
    *   **照度（\(E\)）：** 照射到单位面积上的光通量。单位：勒克斯（lx）。

**9.3 光谱光视效率函数的含义是什么？人眼有几种光谱光视效率函数？**
*   **解题思路：** 理解人眼对不同波长光的感知能力。
*   **答案：**
    *   **含义：** 描述人眼对不同波长光的视觉敏感程度的函数，通常用 \(V(\lambda)\) 表示，峰值在 \(555 \text{ nm}\) 处（明视觉）。
    *   **种类：** 有两种。明视觉光谱光视效率函数 \(V(\lambda)\)（锥体细胞起作用）和暗视觉光谱光视效率函数 \(V'(\lambda)\)（杆体细胞起作用，峰值在 \(507 \text{ nm}\)）。

**9.4 简述颜色混合的种类，光色相加混合符合哪些定性的基本规律？**
*   **解题思路：** 结合图片上半部分的正文内容作答。
*   **答案：**
    *   **种类：** 相加混合（如不同色光投射叠加）和相减混合（如颜料混合）。
    *   **基本规律：**
        1.  **代替律：** 相似色混合，混合色仍相似（如文中 \(A=B, C=D \Rightarrow A+C=B+D\)）。
        2.  **亮度相加律：** 混合色光的亮度等于各色光亮度之和，即 \(L=L_1+L_2+\cdots+L_n\)。

**9.5 人眼的视细胞有哪几类？各有什么特点？**
*   **解题思路：** 区分视锥细胞和视杆细胞的生理结构与功能。
*   **答案：**
    *   **视锥细胞：** 集中在视网膜中央凹，感光灵敏度低，只在明亮环境下起作用（明视觉），能分辨颜色，有高空间分辨率。
    *   **视杆细胞：** 分布在视网膜边缘，感光灵敏度高，在暗环境下起作用（暗视觉），不能分辨颜色，空间分辨率低。

**9.6 人眼视觉分为哪三种响应？明暗和色彩适应各指什么？**
*   **解题思路：** 理解人眼视觉的生理反应机制。
*   **答案：**
    *   **三种响应：** 明视觉、暗视觉和中间视觉。
    *   **明暗适应：** 人眼对亮度变化产生的适应性反应，从亮处到暗处称暗适应，反之称明适应。
    *   **色彩适应：** 人眼在特定色光环境下长时间停留后，对该颜色的敏感度下降，导致对该颜色的感知发生变化的现象。

**9.7 什么是颜色的宽容量？根据颜色宽容量如何构造均匀颜色空间？**
*   **解题思路：** 理解人眼对颜色差异的分辨极限。
*   **答案：**
    *   **颜色宽容量：** 人眼感觉不到颜色差异的最大范围（即恰可察觉差异 JND 的范围）。
    *   **构造均匀颜色空间：** 由于人眼对不同波长、不同亮度下的颜色宽容量不同（如人眼对蓝紫色区域分辨力弱，对绿色区域分辨力强），为了使颜色空间中的几何距离与人眼感觉的色差成正比，需要根据颜色宽容量数据，通过非线性变换来构造如 \(L^*u^*v^*\) 或 \(L^*a^*b^*\) 这样的均匀颜色空间。

**9.8 简述辐射体的分布温度、色温（相关色温）、亮温和辐射温度的定义，它们的适应性以及与真实温度的关系如何？**
*   **解题思路：** 比较不同“温度”概念的定义及适用对象（黑体与非黑体）。
*   **答案：**
    *   **分布温度：** 当辐射体的相对光谱功率分布与黑体在某一温度下的相对光谱功率分布一致时，该黑体温度即为分布温度。适用于相对光谱分布相似的辐射体。
    *   **色温（相关色温）：** 辐射体的色度与黑体在某一温度下的色度相同时，该黑体温度即为色温。若在均匀色度图上最接近黑体轨迹的色温点则称相关色温。适用于一般光源。
    *   **亮温：** 在某一波长下，辐射体与黑体具有相同的光谱辐射亮度时黑体的温度。适用于高温测量。
    *   **辐射温度：** 辐射体的总辐射出射度等于黑体的总辐射出射度时黑体的温度。
    *   **与真实温度关系：** 对于实际物体（非黑体，发射率 \(<1\)），通常有：真实温度 \(T >\) 亮温 \(T_b\)，真实温度 \(T >\) 辐射温度 \(T_r\)。

**9.9 什么是三刺激值？什么是光谱三刺激值？光谱三刺激值有什么意义？**
*   **解题思路：** 结合文中“5. 光谱的三刺激值”公式作答。
*   **答案：**
    *   **三刺激值：** 在RGB颜色系统中，为了匹配某一特定颜色所需的三原色（红、绿、蓝）的数量，记为 \(R, G, B\)。
    *   **光谱三刺激值：** 匹配等能单色光所需的三原色数量，是波长 \(\lambda\) 的函数，分别记为 \(\bar{r}(\lambda), \bar{g}(\lambda), \bar{b}(\lambda)\)。
    *   **意义：** 它们是颜色计算的基础。只要知道光源的相对光谱功率分布 \(\varphi(\lambda)\)，就可以根据文中的积分公式计算出任何颜色的三刺激值 \(R, G, B\)。

**9.10 物体色的三刺激值与哪些因素有关系？**
*   **解题思路：** 考虑物体色是如何形成的（光源照射+物体反射/透射）。
*   **答案：** 与以下因素有关：
    1.  光源的相对光谱功率分布 \(\varphi(\lambda)\)。
    2.  物体的光谱反射率 \(\rho(\lambda)\) 或光谱透射率 \(\tau(\lambda)\)。
    3.  标准观察者的光谱三刺激值（即人眼的视觉特性）。

**9.11 当波长 \(\lambda=700 \text{ nm}\) 的光照射到一个黄色物体上时，该物体呈现什么颜色？**
*   **解题思路：** 根据物体的光谱反射特性和单色光波长来分析。
*   **答案：** 呈现**黑色（或暗色）**。
    *   **理由：** 黄色物体之所以呈现黄色，是因为它主要反射黄光（约 \(570\sim 590 \text{ nm}\)）而吸收其他波长的光。\(700 \text{ nm}\) 属于红光，该物体不反射红光（对该波长反射率极低），因此该物体在此光照下几乎无反射光进入人眼，所以呈现暗色或黑色。

**9.12 光度导轨的主要功能是什么？**
*   **解题思路：** 结合实验仪器的作用作答。
*   **答案：** 主要用于测量光源的发光强度、光通量，以及验证距离平方反比定律（\(E = I / r^2\)）等光度学基本定律。

**9.13 两辐射强度相等的单色光源，其峰值波长分别为 \(500 \text{ nm}\) 和 \(550 \text{ nm}\)，如果用人眼去观测，哪个比较亮？为什么？**
*   **解题思路：** 对比明视觉光谱光视效率函数 \(V(\lambda)\) 的峰值。
*   **答案：** **波长为 \(550 \text{ nm}\) 的光源比较亮。**
    *   **原因：** 人眼明视觉的光谱光视效率函数 \(V(\lambda)\) 的峰值在 \(555 \text{ nm}\) 附近，\(550 \text{ nm}\) 离峰值更近，其 \(V(\lambda)\) 值大于 \(500 \text{ nm}\) 处的 \(V(\lambda)\) 值（\(500 \text{ nm}\) 处于人眼敏感度下降区域）。虽然两者辐射强度（物理功率）相等，但 \(550 \text{ nm}\) 的光被人眼感知的光通量更大。

### 习题解答

**9.1 假定一个功率（辐射通量）为 \(60 \text{ W}\) 的钨丝充气灯泡在各方向均匀发光，求其发光强度。**
*   **解题思路：** 本题未给出光视效能，需假设该灯泡全部辐射能量转化为 \(555 \text{ nm}\) 的可见光（或已知光视效能）。在此条件下，假设为理想发光体，光通量 \(\Phi_v = K_m \Phi_e\)，其中光视效能 \(K_m \approx 683 \text{ lm/W}\)。
    *   注：若不加特殊说明，有时大学物理题会默认为 \(1\text{W} \rightarrow 683 \text{lm}\)。
*   **计算过程：**
    1.  计算总光通量：\(\Phi_v = 60 \text{ W} \times 683 \text{ lm/W} = 40980 \text{ lm}\)。
    2.  各方向均匀发光，立体角 \(\Omega = 4\pi \text{ sr}\)。
    3.  发光强度：\(I = \frac{\Phi_v}{\Omega} = \frac{40980}{4\pi} \approx 3261 \text{ cd}\)。
    *   *(注：若您的教材中给定了钨丝灯的光视效能常数，如 \(10\sim 20 \text{ lm/W}\)，请代入对应值计算)*。

**9.2 一束光通量为 \(620 \text{ lm}\)、波长为 \(460 \text{ nm}\) 的蓝光射在一个白色屏幕上，问屏幕在 \(1 \text{ min}\) 内接收多少能量？**
*   **解题思路：** 通过波长找到该波长下的光谱光视效率 \(V(\lambda)\)，利用 \(K_m\) 换算出辐射通量，再乘以时间。
*   **计算过程：**
    1.  查表得，\(\lambda=460 \text{ nm}\) 时，明视觉光谱光视效率 \(V(460) \approx 0.06\)。
    2.  复现光视效能 \(K = K_m \cdot V(\lambda) = 683 \times 0.06 = 40.98 \text{ lm/W}\)。
    3.  计算辐射通量（功率）：\(\Phi_e = \frac{\Phi_v}{K} = \frac{620}{40.98} \approx 15.13 \text{ W}\)。
    4.  计算 \(1 \text{ min}\) (\(60 \text{ s}\)) 内接收的能量：\(Q = \Phi_e \cdot t = 15.13 \times 60 \approx 907.8 \text{ J}\)。

**9.3 一发光强度为 \(60 \text{ cd}\) 的灯挂在长、宽、高分别为 \(5 \text{ m}, 3 \text{ m}, 3 \text{ m}\) 的房间天花板正中心，离地面 \(2.5 \text{ m}\)，试求：(1)灯正下方地板的光照度；(2)在房间角落处地板的光照度。**
*   **解题思路：** 利用点光源照度公式 \(E = \frac{I \cos\theta}{r^2}\)。
*   **计算过程：**
    *   已知 \(I = 60 \text{ cd}\)，灯离地面垂直距离 \(h = 2.5 \text{ m}\)。
    *   **(1) 正下方地板：**
        *   法线方向与光线同向，\(\theta = 0^\circ\)，距离 \(r = 2.5 \text{ m}\)。
        *   \(E_1 = \frac{I}{r^2} = \frac{60}{2.5^2} = \frac{60}{6.25} = 9.6 \text{ lx}\)。
    *   **(2) 房间角落处：**
        *   天花板中心在水平面上的投影为房间中心，即坐标 \((2.5, 1.5)\)（以长方向为x，宽方向为y）。
        *   角落处坐标 \((0,0)\)，水平距离 \(d = \sqrt{2.5^2 + 1.5^2} = \sqrt{6.25+2.25} = \sqrt{8.5} \text{ m}\)。
        *   光线距离 \(r = \sqrt{d^2 + h^2} = \sqrt{8.5 + 2.5^2} = \sqrt{8.5 + 6.25} = \sqrt{14.75} \text{ m}\)。
        *   入射角余弦 \(\cos\theta = \frac{h}{r} = \frac{2.5}{\sqrt{14.75}}\)。
        *   照度 \(E_2 = \frac{I \cos\theta}{r^2} = \frac{I \cdot h}{r^3} = \frac{60 \times 2.5}{(\sqrt{14.75})^3} = \frac{150}{14.75 \times \sqrt{14.75}} \approx \frac{150}{14.75 \times 3.84} \approx \frac{150}{56.64} \approx 2.65 \text{ lx}\)。

**9.4 已知太阳辐射亮度为 \(L_0 = 2\times10^7 \text{ W} \cdot \text{s}^{-2} \cdot \text{sr}^{-1}\)（即 \(\text{W} \cdot \text{m}^{-2} \cdot \text{sr}^{-1}\)），太阳的半径为 \(r_0 = 6.957\times10^8 \text{ m}\)，地球的半径为 \(r_e = 6.374\times10^6 \text{ m}\)，太阳到地球的年平均距离为 \(l = 1.496\times10^{11} \text{ m}\)，求太阳的辐射出射度 \(M_0\)、辐射强度 \(I_0\)、辐射能通量 \(\Phi_0\)、地球接收的辐射能通量 \(\Phi_e\) 以及地球大气层边沿的辐照度 \(E_e\)。**
*   **解题思路：** 运用光辐射度学基本公式：\(M = \pi L\)（对朗伯体），\(I = M \cdot S\)（总辐射强度，为 \(L \cdot S \cdot \pi\) 假设各向同性），\(\Phi = I \cdot \Omega\) 或 \(I \cdot 4\pi\)，以及距离平方反比定律。
*   **计算过程：**
    *   假设太阳为朗伯辐射体（各方向亮度一致）。
    1.  **辐射出射度 \(M_0\)**：
        \(M_0 = \pi L_0 = \pi \times 2\times10^7 \approx 6.283 \times 10^7 \text{ W/m}^2\)。
    2.  **辐射强度 \(I_0\)**：
        太阳总面积 \(S_0 = 4\pi r_0^2\)。由于各方向均匀，\(I_0 = L_0 \cdot S_0 = L_0 \cdot 4\pi r_0^2\)。
        \(I_0 = 2\times10^7 \times 4\pi \times (6.957\times10^8)^2 \approx 1.216 \times 10^{26} \text{ W/sr}\)。
        *(或者 \(I_0 = M_0 \cdot S_0 / \pi \cdot \pi\) 的推导是一样的)*。
    3.  **辐射能通量 \(\Phi_0\)**：
        \(\Phi_0 = 4\pi I_0 = 4\pi \times 1.216 \times 10^{26} \approx 1.528 \times 10^{27} \text{ W}\)。
    4.  **地球接收的辐射能通量 \(\Phi_e\)**：
        地球接收到的能量等于太阳辐射到达地球距离处的能量。地球迎光截面积为 \(\pi r_e^2\)。辐照度 \(E_e\) 乘以此截面积即为接收的总通量。
        先求地球大气层边沿辐照度 \(E_e\)：利用距离平方反比定律，\(E_e = \frac{I_0}{l^2}\) （注意：这里的 \(I_0\) 为太阳辐射强度，计算距离为 l 处的照度）。
        \(E_e = \frac{1.216 \times 10^{26}}{(1.496 \times 10^{11})^2} \approx \frac{1.216 \times 10^{26}}{2.238 \times 10^{22}} \approx 5435 \text{ W/m}^2\)。（该值即为太阳常数）。
        再计算地球接收总通量：\(\Phi_e = E_e \cdot \pi r_e^2\)。
        \(\Phi_e = 5435 \times \pi \times (6.374 \times 10^6)^2 \approx 5435 \times 1.276 \times 10^{14} \approx 6.94 \times 10^{17} \text{ W}\)。
    *   整理以上答案：
        *   \(M_0 \approx 6.28 \times 10^7 \text{ W/m}^2\)
        *   \(I_0 \approx 1.22 \times 10^{26} \text{ W/sr}\)
        *   \(\Phi_0 \approx 1.53 \times 10^{27} \text{ W}\)
        *   \(\Phi_e \approx 6.94 \times 10^{17} \text{ W}\)
        *   \(E_e \approx 5.44 \times 10^3 \text{ W/m}^2\)

## 思考题
### 9.1 试述辐射度量与光度量的联系和区别
**思路**：区分物理客观量（辐射度量）与人眼主观感知量（光度量），通过视见函数建立联系。
- **辐射度量**：纯物理量，描述电磁辐射能量，和人眼无关；单位：W、$\mathrm{W/m^2}$ 等。包含辐射通量、辐射强度、辐射亮度等。
- **光度量**：人眼视觉响应的量，只针对可见光；由辐射度量乘以**光谱光视效率函数 $V(\lambda)$** 积分得到。单位：lm、\(\mathrm{cd}\)、\(\mathrm{lx}\)等。
- **联系**：光通量
$$\Phi_v = K_m \int_\lambda \Phi_e(\lambda) V(\lambda) d\lambda$$
$K_m=683\ \mathrm{lm/W}$，最大光谱光视效能。

### 9.2 简述发光强度、亮度、光出射度、照度等定义及其单位
1. **发光强度 $I$**：点光源在给定方向上单位立体角内的光通量。
$$I=\frac{d\Phi}{d\Omega}$$
单位：坎德拉 $\mathrm{\(\mathrm{cd}\)}$。
2. **光亮度 $L$**：面光源，某方向单位投影面积、单位立体角的光通量。
$$L=\frac{d\Phi}{d\Omega \\(\mathrm{cd}\)ot dA\cos\theta}$$
单位：$\mathrm{\(\mathrm{cd}\)/m^2}$。
3. **光出射度 $M$**：发光表面单位面积发出的总光通量。
$$M=\frac{d\Phi}{dA}$$
单位：$\mathrm{lm/m^2}$。
4. **照度 $E$**：被照表面单位面积接收的光通量。
$$E=\frac{d\Phi}{dA}$$
单位：勒克斯 $\mathrm{\(\mathrm{lx}\)}$。

### 9.3 光谱光视效率函数的含义是什么？人眼有几种光谱光视效率函数？
**思路**：人眼对不同波长可见光敏感度不同。
- **光谱光视效率 $V(\lambda)$**：在同等辐射功率下，人眼对波长 $\lambda$ 的光的主观亮度感受，相对555nm黄绿光（最大值=1）的比值。
- 两种：
  1. **明视觉 $V(\lambda)$**：亮度高，锥体细胞起作用，峰值555nm；
  2. **暗视觉 $V'(\lambda)$**：弱光，杆体细胞起作用，峰值507nm。

### 9.4 简述颜色混合的种类，光色相混合符合哪些定性的基本规律？
两类颜色混合：**加色混合（光混合）**、**减色混合（颜料/染料混合）**。
光相加混合（RGB）4条定性规律：
1. **补色律**：互补色混合得到白光；
2. **中间色律**：任意两非补色混合，产生中间色；
3. **代替律**：视觉相同的色，混合效果等效；
4. **亮度相加律**：混合光总亮度等于各组分亮度之和。

### 9.5 人眼的视细胞有哪几类？各有什么特点？
两类：**锥体细胞（视锥）**、**杆体细胞（视杆）**
1. **视锥细胞**：集中黄斑区；强光（明视觉）；分辨色彩；分辨细节；感光灵敏度低。
2. **视杆细胞**：分布视网膜外围；弱光（暗视觉）；不能分辨颜色；对光敏感；分辨细节差。

### 9.6 人眼视觉分为哪三种响应？明暗和色彩适应各指什么？
三种视觉响应：**明视觉、暗视觉、中间视觉**。
- **明暗适应**：环境亮度改变，人眼感光灵敏度随时间调整。暗适应：亮→暗处，灵敏度升高；明适应：暗→亮处，灵敏度下降。
- **色彩适应**：长时间看某颜色，视锥细胞疲劳；移开后看到补色，人眼自动校正环境色偏。

### 9.7 什么是颜色的宽容量？根据颜色宽容量如何构造均匀颜色空间？
**思路**：麦克亚当椭圆。
- **颜色宽容量**：人眼刚好分辨不出差异的两个颜色点，在色品图上的距离，即人眼辨色阈值。色品图上相同几何距离，人眼色差感受不等，各点宽容量不同（麦克亚当椭圆）。
- **均匀颜色空间**：坐标变换，拉伸/压缩色品图，使得空间内任意两点欧氏距离，与人眼感受到的色差成正比，如CIE Lab。

### 9.8 简述辐射体的分布温度、色温（相关色温）、亮温和辐射温度的定义，它们的适应性以及与真实温度的关系如何？
1. **分布温度**：光谱分布和某黑体光谱形状一致，该黑体温度；仅光谱形状匹配，非所有辐射体都有。
2. **色温（黑体）/相关色温**：光源色品坐标最接近某黑体色品，该黑体温度；只有黑体才有色温；非黑体用**相关色温CCT**。只匹配颜色，不等于真实温度。
3. **亮度温度**：同一波长下，光源光谱亮度等于黑体光谱亮度，黑体温度；一般小于真实温度。
4. **辐射温度**：光源总辐射出射度等于黑体，黑体温度；一般小于真实温度。
> 小结：这几个温度都是**等效黑体温度**，只有黑体时等效温度=真实热力学温度；非黑体均不等于真实温度。

### 9.9 什么是三刺激值？什么是光谱三刺激值？光谱三刺激值有什么意义？
- **光谱三刺激值 $\bar r(\lambda),\bar g(\lambda),\bar b(\lambda)$（CIE1931）**：匹配波长 $\lambda$、单位辐射功率单色光，所需要的R、G、B三原色的量。
- **三刺激值 $\(R,G,B\)$**：光源/物体颜色，把光谱乘以光谱三刺激值积分得到。
$$R=k\int_\lambda \varphi(\lambda)\bar r(\lambda)d\lambda$$
- **意义**：任何颜色都可用一组三刺激值定量描述；是CIE色品图、颜色测量的基础。

### 9.10 物体色的三刺激值与哪些因素有关系？
$$R=k\int_\lambda S(\lambda)\rho(\lambda)\bar r(\lambda)d\lambda$$
1. 照明光源光谱 $S(\lambda)$；
2. 物体光谱反射率 $\rho(\lambda)$（透射物体用透射率 $\tau(\lambda)$）；
3. CIE光谱三刺激值 $\bar r,\bar g,\bar b(\lambda)$；
4. 归一化系数 $k$。

### 9.11 当波长 $\lambda=700\ \mathrm{nm}$ 的光照射到一个黄色物体上时，该物体呈现什么颜色？
**思路**：黄色物体光谱特性：反射红+绿光，吸收蓝光；700nm属于红光波段。
黄色物体可以反射700nm红光，没有其他波长，因此物体呈现**红色**。

### 9.12 光度导轨的主要功能是什么？
光度导轨（光具座）：测量光源光度量。
功能：承载光源、光阑、照度计；改变光源与探测器距离；验证距离平方反比律；测量发光强度、照度，做颜色与光度标定实验。

### 9.13 两辐射强度相等的单色光源，其峰值波长分别为500 nm和550 nm，如果用人眼去观测，哪个比较亮？为什么？
**思路**：光通量和主观亮度由辐射强度 × $V(\lambda)$。
辐射强度相等：比较光谱光视效率 $V(\lambda)$。
明视觉：$V(550\mathrm{nm})\approx1$；$V(500\mathrm{nm})\approx0.323$。
👉 **550nm光源看起来更亮**。
原因：同等辐射功率，人眼对555nm附近黄绿光敏感度最高。

## 习题
### 9.1 假定一个功率(辐射通量)为60 W的钨丝充气灯泡在各方向均匀发光，求其发光强度。
**思路**：均匀向全空间发光，立体角 $\Omega=4\pi\ \mathrm{sr}$；先求光通量，再求发光强度。
> 钨灯：这里题目默认**全部辐射通量转化为光通量**（简化题）
光通量：$\Phi = 60\ \mathrm{lm}$
$$I=\frac{\Phi}{\Omega}=\frac{\Phi}{4\pi}$$
$$I=\frac{60}{4\pi}\approx 4.77\ \mathrm{\(\mathrm{cd}\)}$$

> 补充说明：真实钨灯不能全部转为可见光，本题是习题简化模型。

### 9.2 一束光通量为620 lm、波长为460 nm的蓝光射在一个白色屏幕上，问屏幕在1 min内接收多少能量？
**思路**：光通量 $\Phi_v$，光通量与辐射通量关系 $\Phi_v=K_m V(\lambda)\Phi_e$；能量 $E=\Phi_e\\(\mathrm{cd}\)ot t$
已知：
$\Phi_v=620\ \mathrm{lm}$；$t=60\mathrm{s}$；$K_m=683\ \mathrm{lm/W}$；查明视觉 $V(460\mathrm{nm})=0.060$

$$\Phi_e=\frac{\Phi_v}{K_m V(\lambda)}$$
$$\Phi_e=\frac{620}{683\times 0.060}\approx 15.11\ \mathrm{W}$$
接收能量：
$$W=\Phi_e\\(\mathrm{cd}\)ot t=15.11\times60\approx906.6\ \mathrm{J}$$

### 9.3 一发光强度为60 \(\mathrm{cd}\)的灯挂在长、宽、高分别为5 m、3 m、3 m的房间天花板正中心，离地面2.5 m，试求:(1)灯正下方地板的光照度;(2) 在房间角落处地板的光照度。
**思路**：点光源照度公式，距离平方反比：
$$E=\frac{I\cos\theta}{r^2}$$
$I=60\mathrm{\(\mathrm{cd}\)}$；灯在天花板中心，离地高度 $h=2.5\mathrm{m}$

(1) **灯正下方**：光线垂直地板，入射角 $\theta=0,\cos\theta=1$，距离 $r=h=2.5\mathrm{m}$
$$E_1=\frac{I}{h^2}=\frac{60}{2.5^2}=\frac{60}{6.25}=9.6\ \mathrm{\(\mathrm{lx}\)}$$

(2) **房间地板角落**
地板中心到角落水平距离：$d=\sqrt{(5/2)^2+(3/2)^2}=\sqrt{2.5^2+1.5^2}=\sqrt{8.5}\approx2.915\mathrm{m}$
光源到角落空间距离：
$$r=\sqrt{h^2+d^2}=\sqrt{2.5^2+8.5}=\sqrt{14.75}\approx3.841\mathrm{m}$$
$\cos\theta=\dfrac{h}{r}$
$$E_2=\frac{I\cos\theta}{r^2}=\frac{I h}{r^3}$$
$$E_2=\frac{60\times2.5}{3.841^3}\approx\frac{150}{56.72}\approx 2.64\ \mathrm{\(\mathrm{lx}\)}$$

### 9.4 已知太阳辐射亮度为 $L_0=2\times10^7\ \mathrm{W\\(\mathrm{cd}\)ot s^{-2}\\(\mathrm{cd}\)ot sr^{-1}}$，太阳的半径为 $r_0=6.957\times10^8\mathrm{m}$，地球的半径为 $r_e=6.374\times10^6\mathrm{m}$，太阳到地球的年平均距离为 $l=1.496\times10^{11}\mathrm{m}$，求太阳的辐射出射度 $M_0$、辐射强度 $I_0$、辐射能通量 $\Phi_0$、地球接收的辐射能通量 $\Phi_e$ 以及地球大气层边沿的辐照度 $E_e$。
> 注意：朗伯源，辐射出射度 $M=\pi L$

#### 1. 太阳辐射出射度 $M_0$
朗伯面：
$$M_0=\pi L_0$$
$$M_0=\pi\times 2\times10^7 \approx 6.283\times10^7\ \mathrm{W/m^2}$$

#### 2. 太阳辐射通量 $\Phi_0$
太阳表面积 $A_0=4\pi r_0^2$
$$\Phi_0=M_0 A_0=\pi L_0 \\(\mathrm{cd}\)ot 4\pi r_0^2=4\pi^2 L_0 r_0^2$$
$$\Phi_0=4\pi^2\times 2\times10^7\times (6.957\times10^8)^2$$
$$\Phi_0\approx 3.82\times 10^{26}\ \mathrm{W}$$

#### 3. 太阳辐射强度 $I_0$
太阳远处视为点源，总辐射通量均匀分布 $4\pi$ 立体角
$$I_0=\frac{\Phi_0}{4\pi}$$
$$I_0=\frac{3.82\times10^{26}}{4\pi}\approx 3.04\times10^{25}\ \mathrm{W/sr}$$

#### 4. 地球大气层边沿辐照度（太阳常数）$E_e$
距离平方反比：
$$E_e=\frac{I_0}{l^2}$$
$$E_e=\frac{3.04\times10^{25}}{(1.496\times10^{11})^2}\approx1358\ \mathrm{W/m^2}$$

#### 5. 地球接收的辐射通量 $\Phi_e$
地球接收太阳辐射等效为**横截面积** $A_\perp=\pi r_e^2$
$$\Phi_e=E_e\\(\mathrm{cd}\)ot \pi r_e^2$$
$$\Phi_e=1358\times\pi\times(6.374\times10^6)^2\approx1.73\times10^{17}\ \mathrm{W}$$
