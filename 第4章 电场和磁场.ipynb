# 本章提要

**1. 电场强度**

(1) 定义

$$E = \frac{F}{q_0}$$

(2) 叠加原理

$$E = \int dE = \int \frac{1}{4\pi\varepsilon_0} \frac{dq}{r^2} e_r$$

**2. 磁感应强度 毕奥-萨伐尔定律**

(1) 定义

$$B = \frac{F_{\max}}{qv}$$
$B$ 的方向由 $F \times qv$ 决定。

(2) 毕奥-萨伐尔定律

$$mathrm{d}B = \frac{\mu_0}{4\pi} \frac{Idl \times e_r}{r^2}$$

(3) 叠加原理

$$B = \int_l dB = \int_l \frac{\mu_0}{4\pi} \frac{Idl \times e_r}{r^2}$$

**3. 高斯定理**

(1) 电场强度通量

$$\Phi_e = \int_S E \cdot dS$$
(2) 磁感应强度通量
$$\Phi_m = \int_S B \cdot dS$$
(3) 高斯定理

电场：
$$\oint_S E \cdot dS = \frac{1}{\varepsilon_0} \sum_{i=1}^{n} q_i$$
磁场：
$$\oint_S B \cdot dS = 0$$

**4. 环路定理**

(1) 静电场
$$\oint_L E \cdot dl = 0$$

(2) 磁场
$$\oint_L B \cdot dl = \mu_0 \sum_{i=1}^{n} I_i$$

**5. 电势**

(1) 定义
$$U_P = \int_P^{\infty} E \cdot dl$$
(2) 叠加原理
$$U = \sum_{i=1}^{n} \frac{1}{4\pi\varepsilon_0} \frac{q_i}{r_i}$$
(3) 电势的计算
$$U = \int \frac{1}{4\pi\varepsilon_0} \frac{dq}{r}$$

**6. 电介质 磁介质**

(1) 电介质及电介质极化

电阻率很大、导电能力很差的物质称为电介质。电介质表面出现极化电荷的现象称为电介质的极化。

(2) 电介质内部的电场
$$E = E_0 + E' = \frac{E_0}{\varepsilon_r}$$
电位移矢量：
$$D = \varepsilon E$$
(3) 电介质中的高斯定理
$$\oint_S D \cdot dS = \sum_{i=1}^{n} q_{0i}$$
(4) 磁介质及磁介质磁化

能够影响磁场分布的物质统称为磁介质。外加磁场对磁介质中的分子电流产生作用，从而产生附加磁场的现象称为磁化。磁介质分类：顺磁质、抗磁质、铁磁质。

(5) 均匀磁介质中的磁场
$$B = \mu_r B_0$$
$$H = \frac{B}{\mu}$$
(6) 磁介质中的环路定理
$$\oint_L H \cdot dl = \sum I_i$$

**7. 带电粒子在磁场中的运动**

(1) 洛伦兹力
$$F_L = qv \times B$$
(2) 回转半径和回转周期
$$R = \frac{mv_0}{qB}$ , $T = \frac{2\pi m}{qB}$$
(3) 霍尔效应
$$U_H = K \frac{IB}{d}$$

### 思考题解答

**4.1 什么是电？怎样理解电与电荷、电荷与物质、电荷与电场的关系？**
*   **解答**：电是物质的一种基本属性，表现为电荷及电荷间的相互作用。电荷是电的载体，物质由带电粒子构成。电荷激发电场，电场是电荷间相互作用的媒介。

**4.2 根据点电荷的场强公式 \(E=\frac{q}{4\pi\varepsilon r^2}\)，当 \(r\to 0\) 时，场强 \(E\to\infty\)，这是没有物理意义的，对这个问题应怎样理解？**
*   **解答**：点电荷是理想化模型，其体积为零。当 \(r\to 0\) 时，该模型不再适用，实际的带电体总有一定大小，此时需考虑带电体内部的电荷分布，不能简单套用点电荷公式。

**4.3 如果一个电子在通过某一区域时不偏转，能否肯定这个区域中没有磁场？如果它发生偏转，能否肯定该区域中存在磁场？**
*   **解答**：不能肯定无磁场，因为电子运动方向与磁场平行时不受洛伦兹力，不偏转；也不能肯定有磁场，因为电场也会使电子偏转。

**4.4 如果把静止点电荷放在电场中，此电荷是否一定沿着电场线运动？**
*   **解答**：不一定。只有在电场线为直线且电荷初速度为零或初速度方向与电场线共线时，才沿电场线运动。

**4.5 由高斯定理能否得到库仑定律？如果库仑定律中 \(r\) 的指数不恰好为 2，能否由高斯定理得到库仑定律？如果库仑定律中 \(r\) 的指数不恰好为 2，高斯定理能否成立？**
*   **解答**：由高斯定理和球对称性可推导出库仑定律。如果指数不是2，则高斯定理的积分形式将改变，无法得出标准的库仑定律。高斯定理是电磁学的基本定理，其成立不依赖于库仑定律的具体形式（但若是非平方反比，电场线通量概念会改变）。

**4.6 一个静止的点电荷能在它的周围任一点产生电场，一个线电流元是否也能够在它的周围空间任一点产生磁场？**
*   **解答**：不能。根据毕奥-萨伐尔定律，线电流元在自身延长线上的各点产生的磁场为零。

**4.7 如果高斯面上场强 \(E\) 处处为零，能否认为高斯面内一定无电荷？如果高斯面内无电荷，能否肯定高斯面上场强 \(E\) 处处为零？**
*   **解答**：高斯面上 \(E\) 处处为零，只能说明面内净电荷为零，不能说明无电荷（可能有等量异号电荷）。高斯面内无电荷，不能肯定面上 \(E\) 为零，因为面外电荷也会在面上产生电场。

**4.8 有两根载流直导线，均通有电流 \(I\)，方向如图所示。环绕两导线有四种闭合积分路径，问每种情况的 \(\oint \mathbf{B} \cdot d\mathbf{l}\) 为多少？在各条闭合积分路径上 \(\mathbf{B}\) 的大小是否相等？**
*   **解答**：
    *   根据安培环路定理 \(\oint \mathbf{B} \cdot d\mathbf{l} = \mu_0 \sum I_{\text{内}}\)。
    *   图 \(L_1\)：包围右侧导线，\(\oint \mathbf{B} \cdot d\mathbf{l} = \mu_0 I\)。
    *   图 \(L_2\)：包围左侧导线，\(\oint \mathbf{B} \cdot d\mathbf{l} = \mu_0 I\)（方向需注意，与电流方向成右手螺旋关系）。
    *   图 \(L_3\)：同时包围两根导线，且电流方向相反（由图示符号看，一个叉一个点），\(\sum I = 0\)，故 \(\oint \mathbf{B} \cdot d\mathbf{l} = 0\)。
    *   图 \(L_4\)：未包围任何导线，\(\oint \mathbf{B} \cdot d\mathbf{l} = 0\)。
    *   各条路径上 \(\mathbf{B}\) 的大小不相等，因为磁场分布不均匀，靠近导线处 \(\mathbf{B}\) 大，远离处 \(\mathbf{B}\) 小。

**4.9 磁场和电场有何异同？**
*   **解答**：相同点：都是物质存在的形式，都具有能量和动量，都对电荷/电流有力的作用。不同点：电场线起于正电荷止于负电荷（不闭合），磁场线是闭合曲线；静电场是保守场（做功与路径无关），磁场是非保守场（涡旋场）。

**4.10 假如电场的功与路径有关，定义电势差的公式 \(U_{AB}=U_A-U_B=\int_A^B \mathbf{E} \cdot d\mathbf{l}\) 还有没有意义？从原则上说，这时还能不能引入电势差、电势的概念？**
*   **解答**：如果做功与路径有关，该积分值将不是唯一的，上述定义失去意义。电势差和电势的概念是基于保守场（做功与路径无关）建立的，因此在这种情况下不能引入电势的概念。

**4.11 为什么要引入电位移矢量，它和电场强度矢量有什么关系？为什么要引入磁场强度？它和磁感应强度有什么联系和区别？**
*   **解答**：引入电位移矢量 \(\mathbf{D}\) 是为了方便处理介质中的高斯定理，避开复杂的极化电荷。\(\mathbf{D} = \varepsilon_0 \mathbf{E} + \mathbf{P}\)，在线性介质中 \(\mathbf{D} = \varepsilon \mathbf{E}\)。引入磁场强度 \(\mathbf{H}\) 是为了简化介质中的安培环路定理。\(\mathbf{H} = \frac{\mathbf{B}}{\mu_0} - \mathbf{M}\)，线性介质中 \(\mathbf{H} = \frac{\mathbf{B}}{\mu}\)。\(\mathbf{D}\) 和 \(\mathbf{E}\) 描述电场，\(\mathbf{H}\) 和 \(\mathbf{B}\) 描述磁场，\(\mathbf{E}\) 和 \(\mathbf{B}\) 是基本物理量，\(\mathbf{D}\) 和 \(\mathbf{H}\) 是辅助量。

**4.12 电介质的极化现象和导体的静电现象有些什么区别？怎样从物理概念上来说明自由电荷与极化电荷的差别？**
*   **解答**：导体静电感应是自由电荷在电场力作用下宏观移动，达到静电平衡时内部场强为零；电介质极化是束缚电荷微观位移（取向或位移极化），内部场强不为零。自由电荷不受原子核束缚，可宏观移动；极化电荷被束缚在原子或分子内，不能宏观移动。

**4.13 如图所示，\(a,b,c,d,e\) 是从 \(O\) 点发出的一些带电粒子的轨迹，问：(1) 哪些轨迹是属于带正电的粒子的？(2) 哪些轨迹是属于带负电的粒子的？(3) \(a,b,c\) 三条表示同种粒子的轨迹中，哪条轨迹表明带电粒子速度最大？哪条最小？**
*   **解答**：(1) 磁场向里，根据左手定则，带正电粒子受力方向与运动方向及磁场垂直。轨迹 \(a, b\) 向下偏转（若从O点向右上运动），向右偏转的应判断洛伦兹力方向。观察图，假设粒子从O向右运动，带正电粒子受力向下（轨迹 \(a, b\)），带负电粒子受力向上（轨迹 \(d, e\)）。结合答案，(1) \(a, b\)；(2) \(d, e\)；(3) 由半径公式 \(R = \frac{mv}{qB}\) 可知，速度 \(v\) 越大，半径 \(R\) 越大。由图，轨迹 \(a\) 半径最大，故速度最大；轨迹 \(c\) 半径最小，速度最小。

**4.14 在均匀磁场中有一电子枪，它可以发射出速率分别为 \(v\) 和 \(2v\) 的两个电子，这两个电子的速度方向相同，且均与 \(\mathbf{B}\) 垂直。试问这两个电子各绕行一周所需的时间是否相同？**
*   **解答**：相同。回旋周期 \(T = \frac{2\pi m}{qB}\)，与速度 \(v\) 无关。

### 习题解答

**4.1 两个点电荷所带电荷量之和为 \(q\)，问它们各自电荷量为多少时，相互之间的作用力最大？**
*   **思路**：设电荷量为 \(q_1, q_2\)，且 \(q_1 + q_2 = q\)。库仑力 \(F = k\frac{q_1 q_2}{r^2}\)。
*   **解答**：由均值不等式，当 \(q_1 = q_2 = q/2\) 时，乘积 \(q_1 q_2\) 最大，此时相互作用力最大。

**4.2 两个点电荷，电荷量分别为 \(+q\) 和 \(-3q\)，相距 \(d\)，试问：在它们的连线上电场强度 \(E=0\) 的点与电荷量为 \(+q\) 的点电荷相距多远？**
*   **思路**：\(E=0\) 的点必须在 \(+q\) 和 \(-3q\) 连线的外侧（靠近 \(+q\) 的那一侧，因为 \(-3q\) 场强大，要抵消必须离 \(-3q\) 远）。
*   **解答**：设距离 \(+q\) 为 \(x\)，则距离 \(-3q\) 为 \(x+d\)。
    \(\frac{kq}{x^2} = \frac{k(3q)}{(x+d)^2} \Rightarrow \frac{1}{x^2} = \frac{3}{(x+d)^2} \Rightarrow \frac{x+d}{x} = \sqrt{3} \Rightarrow x = \frac{d}{\sqrt{3}-1} = \frac{\sqrt{3}+1}{2}d\)。

**4.3 若电荷量 \(q\) 均匀分布在长为 \(L\) 的细棒上，求在棒的延长线上，离棒中心 \(a\) 处的电场强度。**
*   **思路**：建立坐标，取微元 \(dq = \lambda dx\)，积分求 \(E\)。
*   **解答**：\(\lambda = q/L\)。设中心为原点，积分区间 \([-L/2, L/2]\)，点位于 \(x = a + L/2\)（或 \(a\)）。
    \(E = \int_{-L/2}^{L/2} \frac{k (q/L) dx}{(a - x)^2}\)（假设观测点在 \(a\) 处，棒沿x轴）。
    积分得：\(E = \frac{kq}{a^2 - (L/2)^2}\)。

**4.4 一细玻璃棒被弯成半径为 \(R\) 的半圆形。沿其上半部分均匀分布有电荷量 \(+Q\)，沿其下半部分均匀分布有电荷量 \(-Q\)，求半圆中心 \(P\) 处的电场强度 \(E\)。**
*   **思路**：利用对称性。上半部分带正电，下半部分带负电，它们在中心 \(P\) 处产生的电场在垂直于对称轴（x轴）方向的分量相互抵消，水平方向（沿对称轴）分量叠加。
*   **解答**：设对称轴为x轴。取 \(dq = \frac{Q}{\pi R/2} R d\theta = \frac{2Q}{\pi} d\theta\)。
    \(dE_x = \frac{k dq}{R^2} \cos\theta\)。
    积分范围：上半圆 \(\theta\) 从 \(0\) 到 \(\pi/2\)，下半圆 \(\theta\) 从 \(-\pi/2\) 到 \(0\)。
    总 \(E = 2 \int_0^{\pi/2} \frac{k (2Q/\pi) d\theta}{R^2} \cos\theta = \frac{4kQ}{\pi R^2} [\sin\theta]_0^{\pi/2} = \frac{4kQ}{\pi R^2} = \frac{Q}{\pi^2 \varepsilon_0 R^2}\)。方向沿对称轴指向带负电一侧。

**4.5 半径 \(R\) 为 50 cm 的圆形细塑料棒，两端空隙 \(d\) 为 2 cm，总电荷量为 \(3.12\times10^{-9}\) C 的正电荷均匀分布在棒上。求圆心 \(O\) 处电场强度的大小和方向。**
*   **思路**：利用补偿法。完整的带电圆环在圆心处场强为零。该题相当于完整圆环减去一小段 \(d\) 的电荷。
*   **解答**：空缺部分电荷量 \(q' = \frac{d}{2\pi R} Q\)。
    圆心处场强由空缺部分对应的异号电荷产生：\(E = \frac{k q'}{R^2} = \frac{k Q d}{2\pi R^3}\)。
    代入数据：\(E = \frac{9\times10^9 \times 3.12\times10^{-9} \times 0.02}{2\pi \times 0.5^3} \approx 0.072 \, \text{N/C}\)。方向由圆心指向空缺处。

**4.6 两无限长直导线中通有电流 \(I\)，方向如图所示，已知一点 \(P\) 在两直导线所在平面内，且到两直导线距离相等，均为 \(a\)，求 \(P\) 点的磁感应强度大小和方向。**
*   **思路**：根据安培环路定理，单根无限长直导线产生磁场 \(B = \frac{\mu_0 I}{2\pi r}\)。两根导线产生的磁场进行矢量叠加。
*   **解答**：两导线在 \(P\) 点产生的磁场方向相反（假设电流反向）或相同（假设电流同向）。根据图（略），若电流同向，则大小相加：\(B = 2 \times \frac{\mu_0 I}{2\pi a} = \frac{\mu_0 I}{\pi a}\)；若电流反向，则大小相减：\(B = 0\)。

**4.7 如图所示，无限长载流长直导线中的电流为 \(I\)，求穿过矩形回路的磁通量。**
*   **思路**：\(\Phi = \int \mathbf{B} \cdot d\mathbf{S}\)。
*   **解答**：\(B = \frac{\mu_0 I}{2\pi x}\)，\(dS = l dx\)（\(l\)为矩形高）。
    \(\Phi = \int_a^b \frac{\mu_0 I}{2\pi x} l dx = \frac{\mu_0 I l}{2\pi} \ln\frac{b}{a}\)。

**4.8 将一根导线折成正 \(n\) 边形，其外接圆半径为 \(a\)，设导线电流为 \(I\)。(1) 求外接圆中心处磁感应强度；(2) 当 \(n\to\infty\) 时，上述结果又如何？**
*   **思路**：正 \(n\) 边形每一边在中心产生的磁场叠加。
*   **解答**：(1) 每一边对应圆心角 \(2\pi/n\)，单边在中心产生的磁感应强度 \(B_1 = \frac{\mu_0 I}{4\pi (a \cos(\pi/n))} [\sin(\pi/n) - (-\sin(\pi/n))] = \frac{\mu_0 I}{2\pi a} \tan(\pi/n)\)。
    总 \(B = n B_1 = \frac{\mu_0 n I}{2\pi a} \tan(\pi/n)\)。
    (2) 当 \(n\to\infty\) 时，\(\tan(\pi/n) \approx \pi/n\)，则 \(B \to \frac{\mu_0 I}{2a}\)（即圆形电流中心的磁场）。

**4.9 一宽为 \(b\) 的薄长金属板，其电流为 \(I\)，试求在薄板所在平面上，距板的一边为 \(d\) 的 \(P\) 点的磁感应强度。**
*   **思路**：将板看作无数无限长直导线组成。
*   **解答**：面电流密度 \(j = I/b\)。取宽 \(dx\) 的窄条，电流 \(di = j dx\)。
    \(dB = \frac{\mu_0 di}{2\pi x} = \frac{\mu_0 I}{2\pi b x} dx\)。
    \(B = \int_d^{d+b} \frac{\mu_0 I}{2\pi b x} dx = \frac{\mu_0 I}{2\pi b} \ln\left(\frac{d+b}{d}\right)\)。

**4.10 厚度为 \(b\) 的“无限大”均匀带电平板，其电荷体密度为 \(\rho\)。求板外任一点的电场强度。**
*   **思路**：利用高斯定理。作一个穿过平板的对称圆柱形高斯面。
*   **解答**：由对称性，电场垂直于平板向外。作底面积为 \(S\) 的高斯面，穿过板。
    \(\oint \mathbf{E} \cdot d\mathbf{S} = 2 E S = \frac{\rho S b}{\varepsilon_0}\)。
    \(E = \frac{\rho b}{2\varepsilon_0}\)。方向垂直于板面向外。

**4.11 三块平行金属板 A、B、C 面积均为 \(200 \, \text{cm}^2\)，A、B 相距 4 mm，A、C 相距 2 mm，B 和 C 两板都接地。如果使 A 板带正电荷 \(3.0\times10^{-7}\) C，求：(1) B、C 板上的感应电荷；(2) A 板的电势。**
*   **思路**：B、C接地，电势为0，A板电势为 \(U_A\)。电荷守恒和电势差关系。
*   **解答**：(1) 设B板感应电荷为 \(-q_B\)，C板为 \(-q_C\)，则 \(q_B + q_C = Q_A = 3.0\times10^{-7}\) C。
    由于B、C接地，A板与两板间的电场强度不同，但有 \(U_{AB} = U_{AC}\)。
    \(E_{AB} d_{AB} = E_{AC} d_{AC} \Rightarrow \frac{\sigma_B}{\varepsilon_0} d_{AB} = \frac{\sigma_C}{\varepsilon_0} d_{AC}\)。
    即 \(\frac{q_B}{S} d_{AB} = \frac{q_C}{S} d_{AC} \Rightarrow q_B d_{AB} = q_C d_{AC}\)。
    结合 \(q_B + q_C = Q_A\)，且 \(d_{AB}=4\text{mm}, d_{AC}=2\text{mm}\)，可得 \(2q_B = q_C\)（假设A板电量全部分布在两表面，且与B、C感应电荷对应）。
    错！这里推导有误。A板上下表面电荷分别与B、C板感应电荷对应。设A板上下表面电荷分别为 \(q_1, q_2\)（\(q_1+q_2=Q_A\)），则B板感应 \(-q_1\)，C板感应 \(-q_2\)。
    \(E_{AB} = q_1 / (\varepsilon_0 S)\)，\(E_{AC} = q_2 / (\varepsilon_0 S)\)。
    \(U_{AB} = E_{AB} d_{AB} = U_{AC} = E_{AC} d_{AC} \Rightarrow q_1 d_{AB} = q_2 d_{AC}\)。
    \(q_1 \times 4 = q_2 \times 2 \Rightarrow q_2 = 2 q_1\)。
    又 \(q_1 + q_2 = Q_A \Rightarrow 3q_1 = 3\times10^{-7} \Rightarrow q_1 = 10^{-7}\) C, \(q_2 = 2\times10^{-7}\) C。
    所以B板感应电荷 \(q_B = -10^{-7}\) C，C板 \(q_C = -2\times10^{-7}\) C。
    (2) \(U_A = E_{AC} d_{AC} = \frac{q_2}{\varepsilon_0 S} d_{AC} = \frac{2\times10^{-7}}{8.85\times10^{-12} \times 0.02} \times 0.002 \approx 2.26\times10^4\) V。

**4.12 两根长直导线沿半径方向引到铁环上的 A、B 两点，并与很远的电源相连，求环中心 O 处的磁感应强度。**
*   **思路**：直导线在圆心产生的磁场为零（延长线过圆心）。铁环分为两段圆弧，并联分流。
*   **解答**：圆弧在中心产生的磁场 \(B = \frac{\mu_0 I \theta}{4\pi R}\)。
    由于并联，电压相等，电流与电阻成反比，即 \(I_1 R_1 = I_2 R_2\)。而电阻 \(R \propto \text{弧长} \propto \theta\)。
    所以 \(I_1 \theta_1 = I_2 \theta_2\)，两段圆弧在中心产生的磁场大小相等、方向相反，相互抵消。总 \(B=0\)。

**4.13 设真空中有一无限长载流圆柱，圆柱半径为 \(R\)，圆柱横截面上均匀地通有电流 \(I\)，电流沿轴线流动。求磁场分布。**
*   **思路**：安培环路定理，分 \(r<R\) 和 \(r>R\) 讨论。
*   **解答**：\(r>R\) 时，\(\oint \mathbf{B} \cdot d\mathbf{l} = B \cdot 2\pi r = \mu_0 I \Rightarrow B = \frac{\mu_0 I}{2\pi r}\)。
    \(r<R\) 时，\(\oint \mathbf{B} \cdot d\mathbf{l} = B \cdot 2\pi r = \mu_0 I \frac{\pi r^2}{\pi R^2} \Rightarrow B = \frac{\mu_0 I r}{2\pi R^2}\)。

**4.14 带有电荷量 \(q\) 的点电荷处在导体球壳的中心，球壳的内、外半径分别为 \(R_1\) 和 \(R_2\)，求球壳内、外及球壳上任意一点的电场强度和电势。**
*   **思路**：静电感应，导体内部场强为零。高斯定理求 \(E\)，积分求 \(U\)。
*   **解答**：球壳内表面感应 \(-q\)，外表面感应 \(+q\)。
    \(r < R_1\)：\(E = \frac{kq}{r^2}\)，方向径向向外。
    \(R_1 < r < R_2\)：\(E = 0\)（导体内部）。
    \(r > R_2\)：\(E = \frac{kq}{r^2}\)。
    电势（取无穷远为零）：
    \(r \ge R_2\)：\(U = \frac{kq}{r}\)。
    \(R_1 \le r \le R_2\)：\(U = \frac{kq}{R_2}\)（等势体）。
    \(r < R_1\)：\(U = \frac{kq}{r} - \frac{kq}{R_1} + \frac{kq}{R_2}\)。

**4.15 半径为 \(R_1\) 的导体球带有电荷量 \(q\)，球外套以内、外半径分别为 \(R_2\) 和 \(R_3\) 的同心导体球壳，球壳上带有电荷量 \(Q\)。求：(1) 球和球壳的电势；(2) 球与球壳间的电势差。**
*   **思路**：电荷分布：球 \(q\)，球壳内表面 \(-q\)，外表面 \(Q+q\)。
*   **解答**：(1) 球壳电势 \(U_{\text{壳}} = \frac{k(Q+q)}{R_3}\)。
    导体球电势 \(U_{\text{球}} = \frac{kq}{R_1} - \frac{kq}{R_2} + \frac{k(Q+q)}{R_3}\)。
    (2) 电势差 \(\Delta U = U_{\text{球}} - U_{\text{壳}} = kq(\frac{1}{R_1} - \frac{1}{R_2})\)。

**4.16 在磁感应强度为 \(\mathbf{B}\) 的均匀磁场中，有一半径为 \(R\) 的半球面，\(\mathbf{B}\) 与半球面的轴线夹角为 \(\alpha\)。求通过该半球面的磁通量。**
*   **思路**：高斯定理在磁场中的应用，通过闭合曲面的总磁通量为零。
*   **解答**：通过半球面的磁通量等于通过其底圆平面的磁通量。
    \(\Phi = B \cdot S \cdot \cos\alpha = B \cdot \pi R^2 \cdot \cos\alpha\)。

**4.17 电缆由半径为 \(r_1\) 的导体圆柱和同轴的内、外半径分别为 \(r_2\) 和 \(r_3\) 的导体圆筒构成。电流 \(I_0\) 从导体圆柱流入，从导体圆筒流出，设电流都均匀地分布在导体的横截面上，以 \(r\) 表示到轴线的垂直距离。试求 \(r\) 从 0 到 \(\infty\) 的范围内各处的磁感应强度。**
*   **思路**：安培环路定理，分四个区域。
*   **解答**：
    \(0 < r < r_1\)：\(B = \frac{\mu_0 I_0 r}{2\pi r_1^2}\)。
    \(r_1 < r < r_2\)：\(B = \frac{\mu_0 I_0}{2\pi r}\)。
    \(r_2 < r < r_3\)：\(B = \frac{\mu_0 I_0}{2\pi r} \cdot \frac{r_3^2 - r^2}{r_3^2 - r_2^2}\)（注意电流方向相反）。
    \(r > r_3\)：\(B = 0\)（总电流为零）。

**4.18 有一根半径为 \(R\) 的无限长载流直导体，其中电流 \(I\) 沿轴向流过，并均匀分布在横截面上。现在导体上有一半径为 \(R'\) 的圆柱形空腔，其轴与直导体的轴平行，两轴相距 \(d\)，试用安培环路定理求空腔中心处的磁感应强度，并证明空腔中的磁场是均匀磁场。**
*   **思路**：利用叠加原理。将空腔看作充满电流密度为 \(+j\) 和 \(-j\) 的叠加。
*   **解答**：空腔中心处的磁感应强度 \(B = \frac{\mu_0 j d}{2} = \frac{\mu_0 I d}{2\pi R^2}\)。
    证明：在空腔内任一点，由叠加原理可求得磁场为 \(\mathbf{B} = \frac{\mu_0 \mathbf{j} \times \mathbf{d}}{2}\)，与位置无关，故为均匀磁场。

**4.19 测定离子质量的质谱仪如图所示，离子源 S 产生质量为 \(m\)、电荷量为 \(q\) 的离子，离子的初速度很小，可认为其静止。经电势差 \(U\) 加速后离子进入磁感应强度为 \(B\) 的均匀磁场，并沿一半圆形轨道到达与入口处距离为 \(x\) 的感光板上。试证明：该离子的质量为 \(m=\frac{B^2 q x^2}{8U}\)。**
*   **思路**：动能定理求速度，洛伦兹力提供向心力求半径。
*   **解答**：\(qU = \frac{1}{2}mv^2 \Rightarrow v = \sqrt{\frac{2qU}{m}}\)。
    在磁场中 \(qvB = m\frac{v^2}{r} \Rightarrow r = \frac{mv}{qB}\)。
    由图 \(x = 2r \Rightarrow r = x/2\)。
    代入得 \(\frac{x}{2} = \frac{m}{qB} \sqrt{\frac{2qU}{m}} \Rightarrow \frac{x^2}{4} = \frac{m^2}{q^2B^2} \cdot \frac{2qU}{m} \Rightarrow m = \frac{B^2 q x^2}{8U}\)。

**4.20 半径为 \(R=0.10 \, \text{m}\) 的导体球带有电荷量 \(Q=1.0\times10^{-8} \, \text{C}\)，导体外有两层均匀介质，一层介质的相对电容率 \(\varepsilon_r=5.0\)，厚度 \(d=0.10 \, \text{m}\)。另一层介质为空气，充满其余空间。求：(1) 离球心 \(r=5 \, \text{cm}\) 处的 \(D\) 和 \(E\)；(2) 离球心 \(r=15 \, \text{cm}\) 处的电势 \(U\)；(3) 极化电荷面密度 \(\sigma'\)。**
*   **思路**：高斯定理求 \(D\)，\(E = D / (\varepsilon_0 \varepsilon_r)\)。
*   **解答**：(1) \(r=5\text{cm} < R\)，在导体内部，\(E=0, D=0\)。
    (2) \(r=15\text{cm}\)，位于介质外（空气），\(r > 0.1+0.1=0.2\text{m}\)。
    电势 \(U = \int_{0.15}^{\infty} E_{\text{air}} dr = \int_{0.15}^{\infty} \frac{Q}{4\pi\varepsilon_0 r^2} dr = \frac{Q}{4\pi\varepsilon_0 (0.15)} \approx 600 \, \text{V}\)。
    (3) 在介质内表面 \(r=R\) 处，\(D = \frac{Q}{4\pi R^2}\)，\(E = \frac{D}{\varepsilon_0 \varepsilon_r}\)。
    \(\sigma' = P_n = (\varepsilon_r - 1)\varepsilon_0 E = \frac{(\varepsilon_r - 1)Q}{4\pi \varepsilon_r R^2} \approx 1.27\times10^{-7} \, \text{C/m}^2\)。

**4.21 北京正负电子对撞机中电子在周长为 240 m 的储存环中做轨道运动。已知电子的动量是 \(1.49\times10^{-18} \, \text{kg}\cdot\text{m/s}\)，求偏转磁场的磁感应强度。**
*   **思路**：\(p = mv\)，\(R = \frac{p}{qB}\)。
*   **解答**：周长 \(2\pi R = 240 \Rightarrow R \approx 38.2 \, \text{m}\)。
    \(B = \frac{p}{eR} = \frac{1.49\times10^{-18}}{1.6\times10^{-19} \times 38.2} \approx 0.244 \, \text{T}\)。

**4.22 从太阳射来的速度为 \(0.8\times10^8 \, \text{m/s}\) 的电子进入地球赤道上空高层范艾伦带中，该处磁场的磁感应强度为 \(4\times10^{-7} \, \text{T}\)。此电子做圆周运动的轨道半径为多大？此电子同时沿绕地磁场磁感线的螺线缓慢地向地磁北极移动。当它到达地磁北极附近磁场强度为 \(2\times10^{-5} \, \text{T}\) 的区域时，其轨道半径又为多大？**
*   **思路**：洛伦兹力提供向心力，动量守恒（或速度不变）。
*   **解答**：\(R = \frac{mv}{eB}\)。
    初始：\(R_1 = \frac{9.1\times10^{-31} \times 0.8\times10^8}{1.6\times10^{-19} \times 4\times10^{-7}} \approx 1.14 \, \text{m}\)。
    到达北极附近：\(R_2 = \frac{mv}{eB_2} = R_1 \frac{B_1}{B_2} = 1.14 \times \frac{4\times10^{-7}}{2\times10^{-5}} \approx 0.0228 \, \text{m}\)。

**4.23 被电势差 \(U\) 加速的电子从电子枪口 T 发射出来，其初速度指向 x 方向。为使电子束能击中目标 M 点（直线 TM 与 x 轴夹角为 \(\theta\)），在电子枪外空间加一均匀磁场 \(\mathbf{B}\)，其方向与 TM 平行。已知从 T 到 M 的距离为 \(d\)，电子质量为 \(m\)，电荷量为 \(e\)，为使电子恰能击中 M 点，求磁感应强度 \(B\)。**
*   **思路**：电子做螺旋运动。沿磁场方向（TM方向）速度不变，垂直磁场方向做圆周运动。要击中M，圆周运动周期必须满足恰好完成整数圈。
*   **解答**：\(v = \sqrt{\frac{2eU}{m}}\)。
    沿TM方向速度 \(v_{\parallel} = v \cos\theta\)，垂直方向 \(v_{\perp} = v \sin\theta\)。
    到达M的时间 \(t = \frac{d}{v_{\parallel}} = \frac{d}{v \cos\theta}\)。
    圆周运动周期 \(T = \frac{2\pi m}{eB}\)。需满足 \(t = nT\) (\(n=1,2,\dots\))。
    \(\frac{d}{v \cos\theta} = n \frac{2\pi m}{eB} \Rightarrow B = \frac{2\pi n m v \cos\theta}{e d} = \frac{2\pi n \cos\theta}{d} \sqrt{\frac{2mU}{e}}\)。

**4.24 空间内有互相垂直的均匀电场 \(\mathbf{E}\) 和均匀磁场 \(\mathbf{B}\)，\(\mathbf{B}\) 沿 \(x\) 方向，\(\mathbf{E}\) 沿 \(z\) 方向，一电子开始时以速度 \(\mathbf{v}\) 向 \(y\) 方向前进。问电子运动的轨迹如何？**
*   **思路**：在正交电磁场中运动，可分解为两个方向。
*   **解答**：在 \(x\) 方向，不受力（因为速度在 \(y\) 方向，\(B\) 在 \(x\) 方向，\(v \times B\) 在 \(z\) 方向）。但在 \(z\) 方向受电场力 \(eE\) 和洛伦兹力 \(evB\)。
    由于电场和磁场共同作用，电子在垂直于 \(B\) 的平面（y-z平面）内做摆线运动（或称为漂移运动叠加回旋运动）。沿 \(x\) 方向不受力，速度保持不变？错，题目说一开始向y前进，所以初始 \(v_y = v\)。
    y方向：受洛伦兹力 \(F_y = -e(v_z B)\)，\(z\)方向：受电场力 \(F_z = -eE\) 和洛伦兹力 \(F_z = -e(v_y B)\)。
    这是一个二维运动，轨迹为摆线。

## 思考题解答

### 4.1 什么是电？怎样理解电与电荷、电荷与物质、电荷与电场的关系？

**思路**：从电荷是物质的基本属性、电场是电荷的场、电现象本质是电荷间相互作用入手。

1. **电**：是一种物质间的相互作用现象，根源是**电荷**。
2. 电荷是物质固有的基本属性，不是独立于物质的东西，有正负两种，遵从电荷守恒。
3. 电荷周围激发**电场**，电场是物质，电荷之间的静电力，本质是一个电荷的电场对另一个电荷的作用。> 
> 一句话总结：电荷是电的根源；电荷离不开物质；电场是电荷之间相互作用的媒介。

### 4.2 根据点电荷场强公式 $(\boldsymbol{(\boldsymbol E)})=\f(\boldsymbol{r})ac{q}{4\pi\va(\boldsymbol{r})epsilon_0 (\boldsymbol{r})^2}$，当 $(\boldsymbol{r})\to0$ 时，$(\boldsymbol{(\boldsymbol E)})\to\infty$，为什么无物理意义？

**思路**：点电荷是**理想化模型**。
当距离 $(\boldsymbol{r})$ 远小于带电体本身线度时，带电体不能再看成几何点，点电荷公式失效。实际带电电荷分布在一定空间体积，$(\boldsymbol{r})\to0$ 进入电荷内部，不能继续用这个公式。

### 4.3 电子通过区域不偏转，能否肯定无磁场？偏转能否肯定有磁场？

**思路**：洛伦兹力 $\boldsymbol{F}_L = q\boldsymbol{v}\times\boldsymbol{(\boldsymbol{(\boldsymbol B)})}$。

1. 不偏转：**不能肯定无磁场**。
若 $\boldsymbol{v}\pa(\boldsymbol{r})allel\boldsymbol{(\boldsymbol{(\boldsymbol B)})}$，$\boldsymbol{v}\times\boldsymbol{(\boldsymbol{(\boldsymbol B)})}=0$，洛伦兹力为0，粒子直线穿过；也可能电场力和洛伦兹力抵消（速度选择器）。
2. 发生偏转：**不能肯定只有磁场**。电场力 $q\boldsymbol{(\boldsymbol{(\boldsymbol E)})}$ 同样能让带电粒子偏转。

### 4.4 静止点电荷放在电场中，电荷是否一定沿电场线运动？

**思路**：电场线切线是加速度方向，不是速度方向。
不一定。只有电场线为**直线**，且初速度为0/初速度沿电场线，才沿电场线运动；电场线弯曲时，电荷加速度沿切线，速度不会跟着曲线电场线走。

### 4.5 高斯定理能否导出库仑定律？(\boldsymbol{r})的指数不是2时，高斯定理成立吗？

**思路**：高斯定理和库仑定律的依存关系

1. 高斯定理**在球面对称条件下可以推导出库仑定律**；
2. 如果库仑定律中距离的指数不是恰好等于2，则**高斯定理不再成立**。> 
> 本质：高斯定理是平方反比律的直接结果。

### 4.6 静止点电荷产生电场，线电流元能否在周围任一点产生磁场？

**思路**：毕奥–萨伐尔定律 $d\boldsymbol{(\boldsymbol{(\boldsymbol B)})}=\f(\boldsymbol{r})ac{\mu_0}{4\pi}\f(\boldsymbol{r})ac{Id\boldsymbol{l}\times\boldsymbol{\hat (\boldsymbol{r})}}{(\boldsymbol{r})^2}$
不能。当位置矢量 $\boldsymbol{(\boldsymbol{r})}$ 和电流元 $Id\boldsymbol{l}$ 平行时，叉乘 $d\boldsymbol{l}\times\boldsymbol{\hat (\boldsymbol{r})}=0$，该点磁感应强度为0。

> 
> 电流元在自身延长线上各点磁场为0。

### 4.7 高斯面上 $(\boldsymbol{(\boldsymbol E)})$ 处处为零，面内一定无电荷？面内无电荷，面上 $(\boldsymbol{(\boldsymbol E)})$ 处处为零？

1. 高斯面 $(\boldsymbol{(\boldsymbol E)})\equiv0$：**面内总电荷代数和为0，不是一定没有电荷**。可以存在等量正负电荷，内部电荷代数和为0。
2. 高斯面内无电荷：**不能断定面上场强处处为0**。高斯定理只说通量 $\oint \boldsymbol{(\boldsymbol{(\boldsymbol E)})}\cdot d\boldsymbol{S}=0$，通量为零，场强可以部分区域向外、部分向内。

### 4.8 两根载流直导线，电流方向如图，四个闭合回路的 $\oint_L \boldsymbol{(\boldsymbol{(\boldsymbol B)})}\cdot d\boldsymbol{l}$

**思路**：安培环路定理 $\oint_L \boldsymbol{(\boldsymbol{(\boldsymbol B)})}\cdot d\boldsymbol{l}=\sum I_{\text{穿过}}$，电流符号右手螺旋判定。

- $L_1$：包围右侧电流，$\boldsymbol{\oint (\boldsymbol{(\boldsymbol B)})\cdot dl}=\mu_0 I$
- $L_2$：包围左侧电流，$\boldsymbol{\oint (\boldsymbol{(\boldsymbol B)})\cdot dl}=-\mu_0 I$
- $L_3$：包围两个反向电流，代数和为0，$\oint \boldsymbol{(\boldsymbol{(\boldsymbol B)})}\cdot d\boldsymbol{l}=0$
- $L_4$：不包围任何电流，$\oint \boldsymbol{(\boldsymbol{(\boldsymbol B)})}\cdot d\boldsymbol{l}=0$

> 
> 环路积分只和环路包围的电流代数和有关；**同一环路上各点(\boldsymbol{(\boldsymbol B)})大小不一定相等**，(\boldsymbol{(\boldsymbol B)})是两根导线在空间叠加的总磁场。

### 4.9 磁场和电场有何异同？

**思路**：对比静电场、稳恒磁场的场源、高斯定理、环路定理、力。
✅ 相同点：都是物质场，具有能量动量，对电荷/电流施加作用力。
✅ 不同点：

1. 静电场源是静止电荷；磁场源是运动电荷（电流）；
2. 静电场高斯定理 $\oint\boldsymbol{(\boldsymbol{(\boldsymbol E)})}\cdot d\boldsymbol{S}=\sum q/\va(\boldsymbol{r})epsilon_0$，有源；磁场高斯定理 $\oint\boldsymbol{(\boldsymbol{(\boldsymbol B)})}\cdot d\boldsymbol{S}=0$，无源，没有磁单极；
3. 静电场环路 $\oint\boldsymbol{(\boldsymbol{(\boldsymbol E)})}\cdot d\boldsymbol{l}=0$，保守场，可定义电势；稳恒磁场环路不为0，非保守场，**不能定义标量磁势**；
4. 电场力 $q\boldsymbol{(\boldsymbol{(\boldsymbol E)})}$，对静止电荷也有力；洛伦兹力 $q\boldsymbol{v}\times\boldsymbol{(\boldsymbol{(\boldsymbol B)})}$，只对运动电荷有力，永不做功。

### 4.10 若电场做功与路径有关，电势差定义 $U_{A(\boldsymbol{(\boldsymbol B)})}=\int_A^(\boldsymbol{(\boldsymbol B)}) \boldsymbol{(\boldsymbol{(\boldsymbol E)})}\cdot d\boldsymbol{l}$ 是否有意义？

**思路**：电势差定义依赖**保守场（做功只和初末位置有关）**。
没有意义。积分结果会随路径不同而不同，A、(\boldsymbol{(\boldsymbol B)})两点之间没有唯一确定的值，无法定义唯一电势差、电势。只有保守场，才能引入电势。

### 4.11 为什么引入电位移矢量 $\boldsymbol{(\boldsymbol{D})}$？磁场强度 $\boldsymbol{(\boldsymbol{H})}$？

**思路**：介质极化/磁化会出现附加电荷、附加电流，直接计算总场麻烦。

1. $\boldsymbol{(\boldsymbol{D})}=\va(\boldsymbol{r})epsilon_0\boldsymbol{(\boldsymbol{(\boldsymbol E)})}+\boldsymbol{P}$，$\boldsymbol{(\boldsymbol{D})}=\va(\boldsymbol{r})epsilon\boldsymbol{(\boldsymbol{(\boldsymbol E)})}$：**消去极化电荷**，电介质高斯定理写成 $\oint\boldsymbol{(\boldsymbol{D})}\cdot d\boldsymbol{S}=\sum q_{\text{自由}}$，只需要自由电荷，简化介质电场计算。
关系：$\boldsymbol{(\boldsymbol{D})}$ 是辅助量，$\boldsymbol{(\boldsymbol{(\boldsymbol E)})}$ 是真实宏观电场。
2. $\boldsymbol{(\boldsymbol{H})}=\boldsymbol{(\boldsymbol{(\boldsymbol B)})}/\mu_0-\boldsymbol{M}$，$\boldsymbol{(\boldsymbol{(\boldsymbol B)})}=\mu\boldsymbol{(\boldsymbol{H})}$：**消去磁化电流**，环路定理写成 $\oint\boldsymbol{(\boldsymbol{H})}\cdot d\boldsymbol{l}=\sum I_{\text{自由}}$，只需要自由传导电流。
区别：$\boldsymbol{(\boldsymbol{(\boldsymbol B)})}$（磁感应强度，描述磁场本身，可测）；$\boldsymbol{(\boldsymbol{H})}$ 是引入的辅助物理量。

### 4.12 电介质极化现象、导体静电现象区别；自由电荷、极化电荷区别

**思路**：
✅ 现象区别：

✅ 电荷区别：

- **自由电荷**：可以宏观移动，导体中自由电子，能转移到物体外部；
- **极化电荷**：束缚在分子上，不能脱离介质，只能在分子尺度偏移，只出现在介质表面。

## 4.2 $(+q),-3q$ 相距 $d$，连线上电场 $(\boldsymbol{(\boldsymbol E)})=0$ 的位置

**思路**：场强矢量叠加。正负电荷之间场强同向，零点只能在 $(+q)$ 外侧远离负电荷一侧。

$$

$$

## 4.3 长 $L$ 细棒，总电荷 $q$，棒延长线上离中心 $a$ 处场强

$$
d(\boldsymbol{(\boldsymbol E)})=\f(\boldsymbol{r})ac{\lambda dx'}{4\pi\va(\boldsymbol{r})epsilon_0(a-x')^2}

积分限 $x'=-L/2 \to +L/2$
$$
(\boldsymbol{(\boldsymbol E)})=\int_{-L/2}^{L/2}\f(\boldsymbol{r})ac{\lambda dx'}{4\pi\va(\boldsymbol{r})epsilon_0(a-x')^2}=\f(\boldsymbol{r})ac{\lambda}{4\pi\va(\boldsymbol{r})epsilon_0}\left(\f(\boldsymbol{r})ac{1}{a-L/2}-\f(\boldsymbol{r})ac{1}{a+L/2}\(\boldsymbol{r})ight)

$$
(\boldsymbol{(\boldsymbol E)})=\f(\boldsymbol{r})ac{1}{4\pi\va(\boldsymbol{r})epsilon_0}\f(\boldsymbol{r})ac{q}{a^2-(L/2)^2}
$$
