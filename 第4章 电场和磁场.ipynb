# 本章提要

**1. 电场强度**

(1) 定义

$$\boldsymbol{E} = \dfrac{\boldsymbol{F}}{q_0}$$

(2) 叠加原理

$$\boldsymbol{E} = \int \mathrm{d}\boldsymbol{E} = \int \dfrac{1}{4\pi\varepsilon_0} \dfrac{\mathrm{d}q}{r^2} \boldsymbol{e}_r$$

**2. 磁感应强度 毕奥-萨伐尔定律**

(1) 定义

$$B = \frac{F_{\max}}{qv}$$
$\boldsymbol{B}$ 的方向由 $\boldsymbol{F} \times q\boldsymbol{v}$ 决定。

(2) 毕奥-萨伐尔定律

$$\mathrm{d}B = \frac{\mu_0}{4\pi} \frac{Id\boldsymbol{l} \times \boldsymbol{e}_r}{r^2}$$

(3) 叠加原理

$$B = \int_l dB = \int_l \frac{\mu_0}{4\pi} \frac{Idl \times e_r}{r^2}$$

**3. 高斯定理**

(1) 电场强度通量

$$\varPhi_e = \int_S \boldsymbol{E} \cdot d\boldsymbol{S}$$
(2) 磁感应强度通量
$$\varPhi_m = \int_S \boldsymbol{B} \cdot d\boldsymbol{S}$$
(3) 高斯定理

电场：
$$\oint_S \boldsymbol{E} \cdot d\boldsymbol{S} = \frac{1}{\varepsilon_0} \sum_{i=1}^{n} q_i$$
磁场：
$$\oint_S \boldsymbol{B} \cdot d\boldsymbol{S} = 0$$

**4. 环路定理**

(1) 静电场
$$\oint_L \boldsymbol{E} \cdot d\boldsymbol{l} = 0$$

(2) 磁场
$$\oint_L \boldsymbol{B} \cdot d\boldsymbol{l} = \mu_0 \sum_{i=1}^{n} I_i$$

**5. 电势**

(1) 定义
$$U_P = \int_P^{\infty} \boldsymbol{E} \cdot d\boldsymbol{l}$$
(2) 叠加原理
$$U = \sum_{i=1}^{n} \frac{1}{4\pi\varepsilon_0} \frac{q_i}{r_i}$$
(3) 电势的计算
$$U = \int \frac{1}{4\pi\varepsilon_0} \frac{dq}{r}$$

**6. 电介质 磁介质**

(1) 电介质及电介质极化

电阻率很大、导电能力很差的物质称为电介质。电介质表面出现极化电荷的现象称为电介质的极化。

(2) 电介质内部的电场
$$E = E_0 + E' = \frac{E_0}{\varepsilon_r}$$
电位移矢量：
$$D = \varepsilon E$$
(3) 电介质中的高斯定理
$$\oint_S \boldsymbol{D} \cdot d\boldsymbol{S} = \sum_{i=1}^{n} q_{0i}$$
(4) 磁介质及磁介质磁化

能够影响磁场分布的物质统称为磁介质。外加磁场对磁介质中的分子电流产生作用，从而产生附加磁场的现象称为磁化。磁介质分类：顺磁质、抗磁质、铁磁质。

(5) 均匀磁介质中的磁场
$$\boldsymbol{B} = \mu_r \boldsymbol{B}_0$$
$$\boldsymbol{H} = \frac{\boldsymbol{B}}{\mu}$$
(6) 磁介质中的环路定理
$$\oint_L \boldsymbol{H} \cdot d\boldsymbol{l} = \sum I_i$$

**7. 带电粒子在磁场中的运动**

(1) 洛伦兹力
$$\boldsymbol{F}_L = q\boldsymbol{v} \times \boldsymbol{B}$$
(2) 回转半径和回转周期
$$R = \frac{mv_0}{qB}$ , $T = \frac{2\pi m}{qB}$$
(3) 霍尔效应
$$U_H = K \frac{IB}{d}$$

# 思考题

## 思考题4.1

什么是电？怎样理解电与电荷、电荷与物质、电荷与电场的关系？

解题思路：从电荷是物质的基本属性、电场是电荷的场、电现象本质是电荷间相互作用入手。

解答：
(1) 电是一种物质间的相互作用现象，根源是电荷。

(2) 电荷是物质固有的基本属性，不是独立于物质的东西，有正负两种，遵从电荷守恒。

(3) 电荷周围激发电场，电场是物质，电荷之间的静电力，本质是一个电荷的电场对另一个电荷的作用。>

一句话总结：电荷是电的根源；电荷离不开物质；电场是电荷之间相互作用的媒介。

## 思考题4.2

根据点电荷的场强公式 $E=\dfrac{q}{4\pi\varepsilon r^2}$，当 $r\to 0$ 时，场强 $E\to\infty$，这是没有物理意义的，对这个问题应怎样理解？

解答：点电荷是理想化模型，其体积为零。当 $r\to 0$ 时，该模型不再适用，实际的带电体总有一定大小，此时需考虑带电体内部的电荷分布，不能简单套用点电荷公式。

## 思考题4.3

如果一个电子在通过某一区域时不偏转，能否肯定这个区域中没有磁场？如果它发生偏转，能否肯定该区域中存在磁场？**

解题思路：洛伦兹力 $\boldsymbol{F}_L = q\boldsymbol{v}\times\boldsymbol{B}$。

解答：不能肯定无磁场，因为电子运动方向与磁场平行时不受洛伦兹力，不偏转；也不能肯定有磁场，因为电场也会使电子偏转。


(1) 不偏转：不能肯定无磁场。

若 $\boldsymbol{v}\parallel\boldsymbol{B}$，$\boldsymbol{v}\times\boldsymbol{B}=0$，洛伦兹力为 $0$，粒子直线穿过；也可能电场力和洛伦兹力抵消（速度选择器）。

(2) 发生偏转：不能肯定只有磁场。

电场力 $q\boldsymbol{E}$ 同样能让带电粒子偏转。

## 思考题4.4

如果把静止点电荷放在电场中，此电荷是否一定沿着电场线运动？

解题思路：电场线切线是加速度方向，不是速度方向。

解答：

不一定。只有电场线为直线，且初速度为 $0$ /初速度沿电场线，才沿电场线运动；电场线弯曲时，电荷加速度沿切线，速度不会跟着曲线电场线走。

## 思考题4.6

一个静止的点电荷能在它的周围任一点产生电场，一个线电流元是否也能够在它的周围空间任一点产生磁场？**

解题思路：毕奥–萨伐尔定律 $d\boldsymbol{B}=\dfrac{\mu_0}{4\pi}\dfrac{Id\boldsymbol{l}\times\boldsymbol{e}_r}{r^2}$

解答：

不能。当场点位于电流元 $Id\boldsymbol{l}$ 自身延长线上时，叉乘 $d\boldsymbol{l}\times\boldsymbol{r}=0$，磁感应强度为 $0$。

## 思考题4.7

如果高斯面上场强 $\boldsymbol{E}$ 处处为零，能否认为高斯面内一定无电荷？如果高斯面内无电荷，能否肯定高斯面上场强 $E$ 处处为零？



解答：

高斯面上场强 $\boldsymbol{E}$ 处处为零，只能说明面内净电荷为零，不能说明无电荷。例如将一个电偶极子（一个带电量为 $+q$ 的点电荷和一个带电量为 $-q$ 的点电荷）放置在球心，外面取一个同心的高斯球面，在球面任一点，电场强度处处为零。

高斯面内无电荷，不能肯定面上场强 $E$ 为零，因为面外电荷也会在面上产生电场。

## 思考题4.9 

磁场和电场有何异同？


解题思路：对比静电场、稳恒磁场的场源、高斯定理、环路定理、力。

相同点：都是物质存在的形式，都具有能量和动量，都对电荷/电流有力的作用。

不同点：

(1) 静电场源是静止电荷；磁场源是运动电荷（电流）；

(2) 电场线起于正电荷止于负电荷（不闭合），磁场线是闭合曲线；

(3) 静电场高斯定理 $\oint\boldsymbol{E}\cdot d\boldsymbol{S}=\sum q/\varepsilon_0$，有源；磁场高斯定理 $\oint\boldsymbol{B}\cdot d\boldsymbol{S}=0$，无源，没有磁单极；

(4) 静电场环路 $\oint\boldsymbol{E}\cdot d\boldsymbol{l}=0$，保守场，可定义电势；稳恒磁场环路 $\oint_L \boldsymbol{B} \cdot d\boldsymbol{l} = \mu_0 \sum\limits_{i=1}^n I_i$，不为 $0$，非保守场，**不能定义标量磁势**；

(5) 电场力 $q\boldsymbol{E}$，对静止电荷也有力；洛伦兹力 $q\boldsymbol{v}\times\boldsymbol{B}$，只对运动电荷有力，永不做功。

## 思考题4.10

假如电场的功与路径有关，定义电势差的公式 $U_{AB}=U_A-U_B=\int_A^B \boldsymbol{E} \cdot d\boldsymbol{l}$ 还有没有意义？从原则上说，这时还能不能引入电势差、电势的概念？

解题思路：电势差定义依赖保守场（做功只和初末位置有关）。

解答：

如果做功与路径有关，该积分值将不是唯一的，上述定义失去意义。电势差和电势的概念是基于保守场（做功与路径无关）建立的，因此在这种情况下不能引入电势的概念。

## 思考题4.11 

为什么要引入电位移矢量，它和电场强度矢量有什么关系？为什么要引入磁场强度？它和磁感应强度有什么联系和区别？

解题思路：介质极化/磁化会出现附加电荷、附加电流，直接计算总场麻烦。

解答：

引入电位移矢量 $\boldsymbol{D}=\varepsilon_r\varepsilon_0\boldsymbol{E}$ 是为了方便处理介质中的高斯定理，避开复杂的极化电荷。通过任意一个闭合曲面的电位移通量等于该曲面所包围的自由电荷的代数和 $\oint_S \boldsymbol{D} \cdot d\boldsymbol{S} = Q_0$。

引入磁场强度 $\boldsymbol{H}$ 是为了简化介质中的安培环路定理，消去磁化电流。有磁介质时的安培环路定理 $\oint_L \boldsymbol{H} \cdot d\boldsymbol{l} = \sum\limits_{i=1}^n I_i$，其中 $\boldsymbol{B} = \mu_0 \mu_r \boldsymbol{H}$。

## 思考题4.12

电介质的极化现象和导体的静电现象有些什么区别？怎样从物理概念上来说明自由电荷与极化电荷的差别？

解题思路：从现象和电荷作出区别。

解答：

导体静电感应是自由电荷在电场力作用下宏观移动，达到静电平衡时内部场强为零，电荷只分布表面；电介质极化是束缚电荷微观位移（取向或位移极化），没有宏观可移动自由电荷，内部场强不为零。

自由电荷不受原子核束缚，可宏观移动；极化电荷被束缚在原子或分子内，不能宏观移动。

## 习题4.1

两个点电荷所带电荷量之和为 $q$，问它们各自电荷量为多少时，相互之间的作用力最大？

解题思路：库仑定律，数学求极值。

解答：

设 $q_1=x,\ q_2=q-x$，距离 $r$
$$F=\dfrac{1}{4\pi\varepsilon_0}\dfrac{x(q-x)}{r^2}$$

求函数 $f(x)=x(q-x)$ 的最大值，二次函数顶点在 $x=\dfrac q2$

$q_1=q_2=\dfrac q2$，两电荷均分总电荷，作用力最大。


## 习题4.2

两个点电荷，电荷量分别为 $+q 和 $-3q$，相距 $d$，试问：在它们的连线上电场强度 $E=0$ 的点与电荷量为 $+q$ 的点电荷相距多远？

解题思路：场强矢量叠加，正负电荷之间场强同向。

解答：

设距离 $+q$ 为 $x$，则距离 $-3q$ 为 $x+d$

$\dfrac{kq}{x^2} = \dfrac{k(3q)}{(x+d)^2} \Rightarrow \dfrac{1}{x^2} = \dfrac{3}{(x+d)^2} \Rightarrow \dfrac{x+d}{x} = \sqrt{3} \Rightarrow x = \dfrac{d}{\sqrt{3}-1} = \dfrac{\sqrt{3}+1}{2}d$    

## 习题4.3

若电荷量 $q$ 均匀分布在长为 $L$ 的细棒上，求在棒的延长线上，离棒中心 $a$ 处的电场强度。

解题思路：建立坐标，取微元 $dq = \lambda dx$，积分求 $E$。

解答：

线电荷密度 $\lambda=\dfrac qL$
坐标：棒中心原点，棒范围 $[-L/2,+L/2]$，场点在 $x=a$

取微元 $dx$，电荷 $dq=\lambda dx$，微元到场点距离 $\boldsymbol{r}=a-x$
$$
d\boldsymbol{E}=\frac{\lambda dx}{4\pi\varepsilon_0(a-x)^2}
$$
积分限 $x=-L/2 \to +L/2$

$$\boldsymbol{E}=\int_{-L/2}^{L/2}\frac{\lambda dx}{4\pi\varepsilon_0(a-x)^2}$$ 

代入 $\lambda=q/L$
$$\boldsymbol{E}=\frac{1}{4\pi\varepsilon_0}\frac{q}{a^2-(L/2)^2}$$
方向沿棒向外（$q>0$）。

## 习题4.4

一细玻璃棒被弯成半径为 $R$ 的半圆形。沿其上半部分均匀分布有电荷量 $+Q$，沿其下半部分均匀分布有电荷量 $-Q$，求半圆中心 $P$ 处的电场强度 $E$。

![](./images/习题4.4图.png)

解题思路：利用对称性。上半部分带正电，下半部分带负电，它们在中心 $P$ 处产生的电场在垂直于对称轴（x轴）方向的分量相互抵消，水平方向（沿对称轴）分量叠加。

解答：

设对称轴为x轴，取 $dq = \dfrac{Q}{\pi R/2} R d\theta = \dfrac{2Q}{\pi} d\theta$

$$dE_x = \dfrac{k dq}{R^2} \cos\theta$$

积分范围：上半圆 $\theta$ 从 $0$ 到 $\pi/2$，下半圆 $\theta$ 从 $-\pi/2$ 到 $0$

总的电场强度 $$E = 2 \int_0^{\pi/2} \frac{k (2Q/\pi) d\theta}{R^2} \cos\theta = \frac{4kQ}{\pi R^2} [\sin\theta]_0^{\pi/2} = \frac{4kQ}{\pi R^2} = \frac{Q}{\pi^2 \varepsilon_0 R^2}$$

方向竖直向下。

## 习题4.5

半径 $R$ 为 $50 \text{cm}$ 的圆形细塑料棒，两端空隙 $d$ 为 $2 \text{cm}$，总电荷量为 $3.12\times10^{-9} \text{C}$ 的正电荷均匀分布在棒上。求圆心 $O$ 处电场强度的大小和方向。

解题思路：利用补偿法。完整的带电圆环在圆心处场强为零。该题相当于完整圆环减去一小段 $d$ 的电荷。

解答：

空缺部分电荷量 $q' = \dfrac{d}{2\pi R} Q$

圆心处场强由空缺部分对应的异号电荷产生：$E = \dfrac{k q'}{R^2} = \dfrac{k Q d}{2\pi R^3}$

代入数据：$E = \dfrac{9\times10^9 \times 3.12\times10^{-9} \times 0.02}{2\pi \times 0.5^3} \approx 0.072 \, \text{N/C}$

方向由圆心指向空缺处。

## 习题4.12

两根长直导线沿半径方向引到铁环上的 A、B 两点，并与很远的电源相连，求环中心 O 处的磁感应强度。

<img src="./images/习题4.12图.png" width=200/>

解题思路：直导线在圆心产生的磁场为零（延长线过圆心）。铁环分为两段圆弧，并联分流。

解答：

圆弧在中心产生的磁场 $B = \dfrac{\mu_0 I \theta}{4\pi R}$

由于并联，电压相等，电流与电阻成反比，即 $I_1 R_1 = I_2 R_2$

而电阻 $R \propto \text{弧长} \propto \theta$

所以 $I_1 \theta_1 = I_2 \theta_2$，两段圆弧在中心产生的磁场大小相等、方向相反，相互抵消。

总 $B=0$。



(1) 两根长直导线在 $O$ 点产生的磁场

这两根长直导线沿半径方向引到 A、B 两点，因此它们的延长线都经过圆心 $O$。

根据毕奥-萨伐尔定律：
$$d\vec{B} = \frac{\mu_0 I}{4\pi} \frac{d\vec{l} \times \vec{r}}{r^3}$$
因为导线电流方向 $d\vec{l}$ 与位置矢量 $\vec{r}$ 平行（夹角为 $0$ 或 $\pi$），所以 $d\vec{l} \times \vec{r} = 0$。

因此，两根长直导线在圆心 $O$ 处产生的磁感应强度为：
$$B_{\text{直}} = 0$$

(2) 铁环两段圆弧在 $O$ 点产生的磁场

铁环被 A、B 两点分成两部分，设其弧长分别为 $l_1$ 和 $l_2$，对应的电阻分别为 $R_1$ 和 $R_2$，通过的电流分别为 $I_1$ 和 $I_2$。铁环半径为 $r$。

根据电阻定律 $R = \rho \dfrac{l}{S}$，可知两段圆弧的电阻与其弧长成正比，即：
$$\dfrac{R_1}{R_2} = \dfrac{l_1}{l_2}$$
由于这两段圆弧并联在 A、B 两点间，它们两端的电压相等，即 $U_{AB} = I_1 R_1 = I_2 R_2$。

结合上式可得：
$$I_1 l_1 = I_2 l_2$$
一段圆弧在圆心产生的磁感应强度公式为 $B = \dfrac{\mu_0 I l}{4\pi r^2}$（其中 $l$ 为弧长）。

则两段圆弧在 $O$ 点产生的磁感应强度大小分别为：
$$B_1 = \frac{\mu_0 I_1 l_1}{4\pi r^2}$$
$$B_2 = \frac{\mu_0 I_2 l_2}{4\pi r^2}$$
代入 $I_1 l_1 = I_2 l_2$，可得 $B_1 = B_2$。

根据右手螺旋定则，两段圆弧中的电流在圆心 $O$ 处产生的磁场方向相反（例如，若一段产生垂直纸面向外的磁场，另一段则产生垂直纸面向里的磁场）。

(3) 总磁感应强度

圆心 $O$ 处的总磁感应强度为直导线与两段圆弧产生的磁场的矢量和：
$$B_O = B_{\text{直}} + B_1 - B_2 = 0 + 0 = 0$$

## 习题4.14

带有电荷量 $q$ 的点电荷处在导体球壳的中心，球壳的内、外半径分别为 $R_1$ 和 $R_2$，求球壳内、外及球壳上任意一点的电场强度和电势。

<img src="images/习题4.14图.png" width=200/>

解题思路：静电感应，导体内部场强为零。高斯定理求 $E$，积分求 $U$。

电场强度（分三个区域）

$\boldsymbol{r<R_1}$

$$
\oint \boldsymbol{E}\cdot dS=\frac{q}{\varepsilon_0} \implies E\cdot 4\pi r^2=\frac{q}{\varepsilon_0}
$$
$$
E=\frac{q}{4\pi\varepsilon_0 r^2}\vec{e}_r
$$

$\boldsymbol{R_1<r<R_2}$

静电平衡，导体内部场强：
$$E=0$$

$\boldsymbol{r>R_2}$

高斯面内总电荷 $q+(-q)(+q)=q$
$$E\cdot 4\pi r^2=\frac{q}{\varepsilon_0}$$
$$E=\frac{q}{4\pi\varepsilon_0 r^2}\vec{e}_r$$

$\boldsymbol{r<R_1}$

&=\frac{q}{4\pi\varepsilon_0}\left(\frac1r-\frac{1}{R_1}+\frac{1}{R_2}\right)
\end{align*}
$$

$\boldsymbol{R_1\le r\le R_2}$

导体是等势体：
$$
\begin{align*}
V&=\int_{R_1}^{R_2}0\cdot dr'+\int_{R_2}^{\infty}\frac{q}{4\pi\varepsilon_0 r'^2}dr'\\
&=\frac{q}{4\pi\varepsilon_0 R_2}
\end{align*}
$$

$\boldsymbol{r>R_2}$

$$
\begin{align*}
V(r)&=\int_{r}^{\infty}\frac{q}{4\pi\varepsilon_0 r'^2}dr'\\
&=\frac{q}{4\pi\varepsilon_0 r}
\end{align*}
$$

$z$ 方向：受电场力 $F_z = -eE$ 和洛伦兹力 $F_z = -e(v_y B)$。

这是一个二维运动，轨迹为摆线。
