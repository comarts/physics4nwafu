# 本章提要

**1. 理想气体压强的微观公式**
\[
p = \frac{1}{3} n m_0 \overline{v^2} = \frac{2}{3} n \overline{\varepsilon}
\]

**2. 温度的微观统计定义**
\[
\overline{\varepsilon} = \frac{3}{2} kT
\]

**3. 自由度**
确定一个物体空间位置的独立坐标数目。对刚性气体分子而言，单原子分子有3个自由度，双原子分子有5个自由度，多原子分子有6个自由度。

**4. 能量均分定理**
在平衡态下，分子热运动的每一个自由度的平均动能都相等，且等于 \(\frac{1}{2}kT\)。以 \(i\) 表示分子热运动的总自由度，则一个分子的总平均动能为
\[
\overline{\varepsilon} = \frac{i}{2} kT
\]
物质的量为 \(\nu\) 的理想气体的内能为
\[
E = \frac{i}{2} \nu RT
\]

**5. 速率分布函数**
\[
f(v) = \frac{dN_v}{N dv}
\]

**6. 麦克斯韦速率分布函数**
\[
f(v) = 4\pi \left( \frac{m_0}{2\pi kT} \right)^{3/2} v^2 e^{-m_0 v^2 / 2kT}
\]

**7. 三种速率**
最概然速率：
\[
v_p = \sqrt{\frac{2kT}{m_0}} = \sqrt{\frac{2RT}{M}} \approx 1.41 \sqrt{\frac{RT}{M}}
\]
平均速率：
\[
\overline{v} = \sqrt{\frac{8kT}{\pi m_0}} = \sqrt{\frac{8RT}{\pi M}} \approx 1.60 \sqrt{\frac{RT}{M}}
\]
方均根速率：
\[
\sqrt{\overline{v^2}} = \sqrt{\frac{3kT}{m_0}} = \sqrt{\frac{3RT}{M}} \approx 1.73 \sqrt{\frac{RT}{M}}
\]

**8. 玻耳兹曼分布律**
平衡态下，某状态区间（粒子势能为 \(\varepsilon_p\)）的粒子数正比于 \(e^{-\varepsilon_p / kT}\)。
重力场粒子数密度按高度的分布（温度均匀）：
\[
n = n_0 e^{-m_0 gh / kT}
\]

**9. 气体分子的平均自由程**
\[
\overline{\lambda} = \frac{1}{\sqrt{2} \pi d^2 n} = \frac{kT}{\sqrt{2} \pi d^2 p}
\]

# 思考题解答

**6.1** 试分析气体为什么容易被压缩，却又不能被无限压缩？
**解题思路**：从分子间距离和分子间作用力的角度分析。
**解答**：气体分子间距很大，分子间作用力十分微弱，因此容易被压缩。当气体被压缩到一定程度时，分子间距显著减小，分子间的斥力急剧增大，抵抗进一步的压缩，因此不能被无限压缩。

**6.2** 在推导理想气体压强公式的过程中，什么地方用到了理想气体的假设？什么地方用到了平衡态的条件？什么地方用到了统计平均的概念？在推导压强公式时，为什么不考虑分子间的相互碰撞？
**解题思路**：回顾理想气体微观模型和压强公式推导过程。
**解答**：
*   **理想气体假设**：分子本身不占体积，除碰撞外分子间无相互作用力。
*   **平衡态条件**：分子速度分布各向同性，即 $\overline{v_x^2} = \overline{v_y^2} = \overline{v_z^2}$。
*   **统计平均概念**：对大量分子求平均值，如 $\overline{v^2}$。
*   **不考虑分子间碰撞的原因**：分子间的弹性碰撞不改变系统的总动量，也不改变分子速度分布的统计规律，因此对压强没有贡献。

**6.3** 气体在平衡态时有何特征？这时气体中有分子热运动吗？热力学中的平衡与力学中的平衡有何不同？
**解题思路**：区分宏观状态与微观运动，以及不同平衡概念的范畴。
**解答**：
*   **平衡态特征**：系统宏观性质（如压强、温度、体积）不随时间变化。
*   **有无热运动**：有。此时气体分子仍在做永不停息的无规则热运动，只是统计平均值不变。
*   **平衡概念的差异**：热力学平衡包含力学平衡、热平衡、相平衡和化学平衡，是动态的、统计意义上的平衡；而力学平衡仅指物体受力平衡，是静态的。

**6.4** 一定质量的理想气体，当温度不变时，其压强随体积的减小而增大；当体积不变时，其压强随温度的升高而增大。从微观上说，这两种压强的增大有何异同？
**解题思路**：从理想气体压强微观公式 $p = \frac{2}{3} n \overline{\varepsilon}$ 出发分析。
**解答**：
*   **相同点**：都导致气体压强增大。
*   **不同点**：温度不变体积减小，是分子数密度 $n$ 增大，导致单位时间与器壁碰撞的分子数增加；体积不变温度升高，是分子平均平动动能 $\overline{\varepsilon}$ 增大，导致分子平均速率增大，碰撞频率和每次碰撞的平均冲量都增加。

**6.5** 试述下列各式的物理意义：
**解题思路**：利用能量均分定理。
**解答**：
(1) $\frac{1}{2}kT$：分子在每一个自由度上的平均动能。
(2) $\frac{3}{2}kT$：分子的平均平动动能。
(3) $\frac{i}{2}RT$：1 mol 理想气体的内能。
(4) $\frac{m}{M}\frac{i}{2}RT$：质量为 $m$ 的理想气体的内能。

**6.6** 试述速率分布函数 $f(v)$ 的物理意义，并说明下列各式的物理意义：
**解题思路**：理解概率密度和统计平均的定义。
**解答**：$f(v)$ 表示速率在 $v$ 附近单位速率区间的分子数占总分子数的比例。
(1) $f(v)dv$：速率在 $v \sim v+dv$ 区间的分子数占总分子数的比例（概率）。
(2) $Nf(v)dv$：速率在 $v \sim v+dv$ 区间的分子数。
(3) $\int_0^v f(v)dv$：速率小于 $v$ 的分子数占总分子数的比例。
(4) $\frac{\int_0^\infty vf(v)dv}{\int_0^\infty f(v)dv}$：全体分子的平均速率 $\overline{v}$。
(5) $\int_0^\infty \frac{1}{2}m_0v^2f(v)dv$：全体分子的平均平动动能。

**6.7** 在相同的温度及压强下，氦气中的声速大于空气中的声速，如何从分子角度来理解这一事实？
**解题思路**：声速公式 $u = \sqrt{\gamma RT/M}$，比较绝热指数和摩尔质量。
**解答**：氦气为单原子分子，$\gamma=5/3$，摩尔质量 $M=4\text{ g/mol}$；空气主要为双原子分子，$\gamma=7/5$，$M \approx 29\text{ g/mol}$。氦气 $M$ 小且 $\gamma$ 大，故声速大。

**6.8** 什么是自由度？它与温度有关吗？
**解题思路**：经典物理中自由度的定义。
**解答**：确定一个物体空间位置所需的独立坐标数。在经典理论中，自由度与温度无关。

**6.9** 理想气体的内能与什么有关？
**解题思路**：理想气体内能公式。
**解答**：理想气体内能只与物质的量 $\nu$、分子自由度 $i$ 和温度 $T$ 有关，与体积和压强无关。公式为 $E = \frac{i}{2}\nu RT$。

**6.10** 什么是分子的平均碰撞频率？什么是平均自由程？
**解题思路**：分子动理论基本概念。
**解答**：
*   **平均碰撞频率** $\overline{Z}$：一个分子在单位时间内与其他分子碰撞的平均次数。
*   **平均自由程** $\overline{\lambda}$：分子在连续两次碰撞之间自由飞行的平均距离。

**6.11** 若盛有气体的容器相对于某坐标系从静止开始运动，容器内的分子速度相对于这个坐标系也将增大，则气体的温度会不会因此升高？
**解题思路**：区分热运动动能和整体定向运动动能。
**解答**：不会。温度是分子热运动平均平动动能的量度，是相对于容器质心系的。整体的定向运动动能不属于热运动，不改变温度。

**6.13** 回答下列问题：
**解题思路**：速率分布函数的概率意义。
**解答**：
(1) 概率为 $f(v)dv$。
(2) 概率为 0（连续分布中取某一确定值的概率为零）。
(3) 较长时间内的平均速率等于该时刻的统计平均速率 $\overline{v}$。

**6.14** 试问：气体分子与地球之间的引力势能，比起分子动能来说是否可以忽略？分子之间的引力相互作用势能又如何呢？
**解题思路**：比较势能与动能的量级。
**解答**：气体分子与地球的引力势能相比分子动能通常可以忽略（除非研究大气标高）。分子间的引力势能不能忽略，否则不能视为理想气体。

**6.15** 气体分子的最概然速率、平均速率和方均根速率是怎样定义的？它们的大小由哪些因素决定，各有什么用处？
**解题思路**：三种速率的定义、决定因素和用途。
**解答**：
*   **定义**：最概然速率 $v_p = \sqrt{2kT/m} = \sqrt{2RT/M}$；平均速率 $\overline{v} = \sqrt{8kT/\pi m} = \sqrt{8RT/\pi M}$；方均根速率 $\sqrt{\overline{v^2}} = \sqrt{3kT/m} = \sqrt{3RT/M}$。
*   **决定因素**：由温度 $T$ 和摩尔质量 $M$ 决定。
*   **用处**：$v_p$ 用于分析分布曲线；$\overline{v}$ 用于碰撞、扩散等；$\sqrt{\overline{v^2}}$ 用于计算平均动能和压强。

**6.16** 已知分子与其他分子碰撞时其速率可以改变，那么分布函数 $f(v)$ 是否与时间有关，为什么？
**解题思路**：平衡态下统计规律的稳定性。
**解答**：无关。平衡态下，虽然分子碰撞不断改变速度，但统计规律保持不变，分布函数不随时间改变。

**6.17** 气体在绝热过程中与环境之间没有热交换，那么为什么温度会有变化呢？
**解题思路**：绝热过程中做功与内能变化的关系。
**解答**：绝热压缩时，外界对气体做功，分子与运动活塞碰撞时获得额外动能，导致分子平均平动动能增大，温度升高；绝热膨胀时则相反。

**6.18** 如果氢气和氦气的温度相同，物质的量相同，那么这两种气体分子的平均动能是否相同？平均平动动能是否相同？两种气体的内能是否相同？

**6.19** 根据分子速率分布的特点，对月球几乎没有大气这一事实给出一个合理的解释。解释在地球的大气中为什么几乎没有He。

**6.17** 日冕的温度为 $2 \times 10^6\text{ K}$，求其中电子的方均根速率。星际空间的温度为 $2.7\text{ K}$，其中成分主要是氢原子，求那里氢原子的方均根速率。1994年，科学家曾用激光冷却的方法使一群钠原子几乎停止运动，相应的温度是 $2.4 \times 10^{-11}\text{ K}$，求这些钠原子的方均根速率。
**解题思路**：方均根速率公式。
**解答**：$v_{rms} = \sqrt{3RT/M}$。日冕电子 $M=9.1\times10^{-31}\text{ kg}$，$\approx 9.5\times10^6\text{ m/s}$；星际氢原子 $\approx 259.5\text{ m/s}$；钠原子 $M=0.023\text{ kg/mol}$，$\approx 1.6\times10^{-4}\text{ m/s}$。

**6.18** 某些恒星的温度达到 $10^8\text{ K}$ 的数量级，在这些温度下原子已不存在，只有质子存在。试问：
**解题思路**：平均动能和方均根速率。
**解答**：(1) 质子平均动能 $\frac{3}{2}kT \approx 2.07\times10^{-15}\text{ J} \approx 12.9\text{ keV}$。(2) $v_{rms} = \sqrt{3RT/M} \approx 1.57\times10^6\text{ m/s}$。

**6.19** 气缸内储有单原子分子理想气体，若绝热压缩使容积减小一半，问气体分子的平均速率变为原来的几倍？
**解题思路**：绝热过程方程和平均速率与温度的关系。
**解答**：绝热过程 $TV^{\gamma-1}=C$，单原子 $\gamma=5/3$。$T_2 = T_1 (V_1/V_2)^{2/3} = T_1 2^{2/3}$。$\overline{v} \propto \sqrt{T}$，所以变为 $2^{1/3} \approx 1.26$ 倍。

**6.20** 如图所示，假设 $f(v)$ 为 $N$ 个分子组成的系统的速率分布函数，分别写出图(a)、(b)、(c)中阴影部分面积对应的数学表达式和物理意义。
**解题思路**：速率分布函数积分面积的物理意义。
**解答**：(a) $\int_0^{v_p} f(v)dv$，速率小于最概然速率的分子数比例。(b) $\int_{v_p}^\infty f(v)dv$，速率大于最概然速率的分子数比例。(c) $\int_{v_p}^{v_1} f(v)dv$，速率在 $v_p$ 到 $v_1$ 之间的分子数比例。

**6.21** 导体中自由电子的运动可以看作类似于气体分子的运动，电子的速率分布函数为 $f(v) = 4\pi A v^2$（$0<v<v_{\text{max}}$）。
**解题思路**：归一化条件和平均动能公式。
**解答**：(1) 归一化 $\int_0^{v_{\text{max}}} 4\pi A v^2 dv = 1 \Rightarrow A = \frac{3}{4\pi v_{\text{max}}^3}$。(2) 平均动能 $\overline{\varepsilon} = \int_0^{v_{\text{max}}} \frac{1}{2}mv^2 f(v)dv = \frac{3}{10}mv_{\text{max}}^2$。

**6.22** 一个由 $N$ 个粒子所组成的系统，其中 $dN = Kdv$（$0<v<v_0$）。
**解题思路**：分布函数归一化，平均速率和方均根速率定义。
**解答**：(1) $f(v) = \frac{K}{N}$（$0<v<v_0$）。(2) $K = \frac{N}{v_0}$。(3) 平均速率 $\overline{v} = \frac{v_0}{2}$，方均根速率 $\sqrt{\overline{v^2}} = \frac{\sqrt{3}}{3}v_0$。

**6.23** 根据麦克斯韦速率分布律，求平动动能的最概然值 $\varepsilon_{\text{tp}}$。
**解题思路**：最概然动能的定义，即动能分布函数取极大值。
**解答**：由 $f(\varepsilon) = \frac{2}{\sqrt{\pi}}(kT)^{-3/2}\sqrt{\varepsilon}e^{-\varepsilon/kT}$，令 $df(\varepsilon)/d\varepsilon = 0$，解得 $\varepsilon_{\text{tp}} = \frac{1}{2}kT$。

**6.24** 试问上升到什么高度时，大气压强减至地面的75%？设空气温度为 $0\text{ }^\circ\text{C}$，空气的摩尔质量为 $0.029\text{ kg}\cdot\text{mol}^{-1}$。
**解题思路**：玻尔兹曼分布（气压公式）。
**解答**：$p = p_0 e^{-Mgh/RT}$，$0.75 = e^{-Mgh/RT} \Rightarrow h = -\frac{RT}{Mg}\ln 0.75 \approx 2296\text{ m}$。

**6.25** 拉萨海拔约为 $3600\text{ m}$，气温为 $273\text{ K}$。
**解题思路**：气压公式。
**解答**：(1) $p = p_0 e^{-Mgh/RT} \approx 6.45\times10^4\text{ Pa}$。(2) 每分钟吸入空气质量与密度（正比于压强）有关，$N = N_0 \frac{p_0}{p} = 12 \times \frac{1.013\times10^5}{6.45\times10^4} \approx 18.8$ 次。

**6.26** 设容器内盛有质量为 $m_1$ 和 $m_2$ 的两种不同的单原子分子气体，此混合气体处在平衡态时内能相等，均为 $E$。
**解题思路**：内能相等，理想气体状态方程。
**解答**：(1) 内能相等 $\Rightarrow \nu_1 = \nu_2$。$\overline{v} \propto \sqrt{1/M}$，所以 $\overline{v_1}/\overline{v_2} = \sqrt{M_2/M_1}$。(2) 总压强 $p = \frac{2\nu_1 RT}{V}$，由 $E = \frac{3}{2}\nu_1 RT \Rightarrow \nu_1 RT = \frac{2E}{3}$，所以 $p = \frac{4E}{3V}$。

**6.27** 求氮气在标准状况下的平均碰撞频率。
**解题思路**：平均碰撞频率公式。
**解答**：(1) $\overline{Z} = \sqrt{2}\pi d^2 n \overline{v} \approx 5.4\times10^8\text{ s}^{-1}$。(2) 压强降至 $1.33\times10^{-4}\text{ Pa}$ 时，$n$ 减小，$\overline{Z}' = \overline{Z} \times (p'/p) \approx 0.71\text{ s}^{-1}$。

**6.28** 试求氧气在标准状况下的平均自由程，设氧气分子直径为 $3\times10^{-10}\text{ m}$。
**解题思路**：平均自由程公式。
**解答**：$\overline{\lambda} = \frac{kT}{\sqrt{2}\pi d^2 p} \approx 9.3\times10^{-8}\text{ m}$。

**6.29** 大气层中各种气体的比例会随着高度的变化而略有变化，试问在海拔很高处，氧气所占的比例比海平面处高还是低？为什么？
**解题思路**：玻尔兹曼分布律，重力场中分子数密度随高度的变化。
**解答**：低。因为 $n = n_0 e^{-mgh/kT}$，摩尔质量大的气体衰减更快。氧气（32）比氮气（28）重，所以比例下降更快。

**6.30** 已知某真空设备的真空度可达到 $10^{-15}\text{ atm}$，试问在此压强下温度为 $27\text{ }^\circ\text{C}$ 时，$1\text{ m}^3$ 体积中有多少个气体分子？
**解题思路**：理想气体状态方程。
**解答**：$p = 10^{-15} \times 1.013\times10^5 = 1.013\times10^{-10}\text{ Pa}$，$n = p/kT \approx 2.45\times10^{10}\text{ m}^{-3}$。

**6.31** 迪舍尔采取了另一种办法，即只对小球的热运动进行被动观察，他发现，在室温下小球偏离平衡位置的方均根值 $\sqrt{(\Delta x)^2} \approx 35\text{ nm}$，试问：这个“弹簧”的弹性系数 $k_{\text{sp}}$ 是多少？
**解题思路**：能量均分定理。
**解答**：根据能量均分定理，$\frac{1}{2}k_{\text{sp}}\overline{(\Delta x)^2} = \frac{1}{2}kT$。$k_{\text{sp}} = \frac{kT}{\overline{(\Delta x)^2}} = \frac{1.38\times10^{-23} \times 293}{(3.5\times10^{-8})^2} \approx 3.3\times10^{-6}\text{ N/m}$。

## 思考题解答

### 6.1

**思路**：从分子间隙、分子斥力的特点分析。

> 
> 气体分子间距大，分子间作用力微弱，容易压缩；当被压缩到分子距离很近时，分子间斥力急剧增大，因此不能无限压缩。
> **答**：气体分子之间有很大空隙，分子间相互作用力很小，所以容易被压缩；当压缩到分子间距极小时，分子斥力急剧变大，阻碍进一步压缩，故不能无限压缩。

### 6.2

**思路**：回顾压强公式 $p=\frac13 n m_0\overline{v^2}$ 的推导假设。

1. 理想气体假设：分子视为质点；除碰撞瞬间无相互作用力；碰撞为完全弹性碰撞。
2. 平衡态：容器内各处分子数密度、温度相同，统计平均不随时间变。
3. 统计平均：对大量分子的冲量取统计平均得到压强。
4. 不考虑分子间碰撞：分子之间碰撞只是交换速度，**不改变分子对器壁的碰撞频率与冲量的统计平均效果**，因此可以忽略。

### 6.3

**思路**：平衡态定义；热平衡 vs 力学平衡。
**答**：
平衡态特征：宏观上$(p,V,T)$不随时间变化；微观上分子不停做无规则热运动，分子速度、位置不断变化。
热力学平衡：同时满足力学平衡、热平衡、化学平衡，宏观参量不变，微观仍有热运动；
力学平衡：物体所受合外力为零，物体静止/匀速，无持续热运动。

### 6.4

**思路**：理想气体 $pV=\nu RT$；微观压强 $p=\frac23 n \bar{\varepsilon}_k$。
**答**：
相同点：微观本质都是大量分子撞击器壁产生压强。
不同点：

- $V$减小，$n$增大，分子撞击器壁次数增多，压强变大；
- $T$升高，分子平均平动动能增大，撞击力度变大，压强变大。

### 6.5

**思路**：分子动能、内能公式
(1) $\boldsymbol{\frac12 kT}$：一个分子**每个自由度**的平均动能。
(2) $\boldsymbol{\frac32 kT}$：一个分子的**平均平动动能**（平动3个自由度）。
(3) $\boldsymbol{\frac{i}{2}RT}$：**1mol理想气体**的内能。
(4) $\boldsymbol{\frac{m}{M}\frac{i}{2}RT}$：质量为$m$的理想气体内能，$\nu=\frac{m}{M}$为物质的量。

### 6.6

**思路**：速率分布函数定义 $(f(v))=\frac{\mathrm{d}N}{N\mathrm{d}v}$

1. $(f(v))\mathrm{d}v=\frac{\mathrm{d}N}{N}$：一个分子速率落在$v\sim v+\mathrm{d}v$区间的概率。
2. $N(f(v))\mathrm{d}v=\mathrm{d}N$：速率在$v\sim v+\mathrm{d}v$区间内的分子数目。
3. $\int_0^v f(v')\mathrm{d}v'$：分子速率小于$v$的概率。
4. $\frac{\int_0^v uf(u)\mathrm{d}u}{\int_0^v f(u)\mathrm{d}u}$：速率在$0\sim v$区间内分子的**平均速率**。
5. $\frac{\int_0^\infty \frac12 m_0 v^2 (f(v))\mathrm{d}v}{\int_0^\infty (f(v))\mathrm{d}v}=\int_0^\infty \frac12 m_0 v^2 (f(v))\mathrm{d}v$：一个分子的**平均动能**。

### 6.7

**思路**：理想气体声速 $u=\sqrt{\frac{\gamma RT}{M}}$，$\gamma$为比热容比。
$\mathrm{N_2}$摩尔质量$M$小于空气平均摩尔质量，温度相同时，$M$越小，声速越大，故氮气中声速更大。

### 6.8

**思路**：自由度定义。
自由度：确定分子空间位置所需独立坐标数目。
**刚性分子自由度与温度无关**；高温下分子振动自由度被激发，总自由度随温度变化。

### 6.9

**思路**：理想气体内能 $E=\nu \frac i2 RT$。
理想气体内能只与**物质的量、分子自由度、热力学温度**有关；与压强、体积无关。

### 6.10

**思路**：平均碰撞频率、平均自由程定义。
平均碰撞频率 $\bar Z$：一个分子单位时间内与其他分子平均碰撞次数；
平均自由程 $\bar\lambda$：分子连续两次碰撞之间走过的平均路程；关系：$\bar\lambda=\frac{\bar v}{\bar Z}$。

### 6.11

**思路**：温度是**分子无规则热运动**的平均动能标志，和整体定向运动无关。
容器整体匀速运动只是所有分子叠加一个共同定向速度，**无规则热运动动能不变，温度不变**。

### 6.12

**思路**：麦克斯韦速率分布：温度越高，最概然速率$v_p=\sqrt{\frac{2RT}{M}}$越大，曲线越宽越矮；$M$越大，$v_p$越小，曲线越尖高。
(a) $\boldsymbol{T_2}$温度更高；
(b) $\boldsymbol{M_1}$摩尔质量更大。

### 6.13

(1) 概率：$\boldsymbol{(f(v))\Delta v}$；
(2) 最概然速率$v_p$处，速率区间$\mathrm{d}v$内概率为$f(v_p)\mathrm{d}v$；
(3) 长时间平均，取**统计平均速率**，即麦克斯韦分布的平均速率 $\bar v=\int_0^\infty v (f(v))\mathrm{d}v$。

### 6.14

**思路**：重力势能 $E_p=m_0gh$；分子引力势能。
分子与地球引力势能远小于分子热动能，可以忽略；
分子之间引力相互作用势能，在理想气体模型中忽略，实际气体中不可忽略。

### 6.15

**思路**：$v_p=\sqrt{\frac{2RT}{M}},\bar v=\sqrt{\frac{8RT}{\pi M}},\sqrt{\overline{v^2}}=\sqrt{\frac{3RT}{M}}$

- 最概然速率$v_p$：速率分布概率最大的速率；用于讨论速率分布曲线。
- 平均速率$\bar v$：分子速率统计平均值；用于平均自由程、碰撞频率计算。
- 方均根速率$\sqrt{\overline{v^2}}$：速率平方平均再开方；用于平均平动动能计算。
大小关系：$\boldsymbol{v_p<\bar v<\sqrt{\overline{v^2}}}$。

### 6.16

**答**：分布函数$(f(v))$**不随时间变化**。
思路：虽然单个分子碰撞改变速率，但大量分子的速率分布达到统计稳态，任意速率区间的分子数目保持不变，故$(f(v))$与时间无关。

### 6.17

**思路**：绝热$(Q=0)$；活塞压缩气体，外界对气体做功$(W>0)$，热力学第一定律$\Delta E=Q+W=(W>0)$，内能增加，温度升高。
微观：分子与运动活塞碰撞，分子获得动能，热运动平均动能增大，温度上升。

### 6.18

**思路**：平均平动动能只和温度有关；内能$E=\nu\frac i2 RT$。
$T$相同：**分子平均平动动能相同**；
$\mathrm{H_2}$双原子$(i=5)$，$\mathrm{(\mathrm{He})}$单原子$(i=3)$；物质的量$\nu$相同，**内能不同**；
平均动能（总动能）：$\mathrm{H_2}$分子总平均动能更大。

### 6.19

**思路**：重力场粒子数密度；麦克斯韦分布，轻分子速率大。
$\mathrm{(\mathrm{He})}$摩尔质量很小，分子热运动速率很大，容易逃逸地球引力；月球引力弱，气体分子更容易逃逸，几乎无大气。

### 6.20

**思路**：蒸发是液体分子逸出，只有动能大的分子才能克服势能离开液体；剩下分子平均动能下降，温度降低。
液体中高速分子离开液面，剩下液体分子平均动能减小，宏观温度下降，所以蒸发制冷。

体积 $V=1\times0.4\times0.5=0.2\mathrm{m^3}$
$$
F=N\frac{mv^2}{L}
$$
$$
p=\frac{F}{S_{\text{侧壁}}}=\frac{Nmv^2}{V}
$$
代入
$$
p=\frac{10000\times 0.002\times 1500^2}{0.2}=2.25\times10^7 \ \mathrm{Pa}

**答案**：(1)$\boldsymbol{\dfrac{\bar v_1}{\bar v_2}=\sqrt{\dfrac{m_2}{m_1}}}$；(2)$\boldsymbol{p=\dfrac{4E}{3V}}$

## 6.27

氮气，标准状态：$T=273\mathrm{K},p_1=1.013\times10^5\mathrm{Pa}$，分子直径$d=10^{-10}\mathrm{m}$。
平均自由程：$\bar\lambda=\frac{kT}{\sqrt{2}\pi d^2 p}$；平均碰撞频率$\bar Z=\frac{\bar v}{\bar\lambda}$；$\bar v=\sqrt{\frac{8RT}{\pi M}},M=0.028\mathrm{kg/mol}$。
标准状态：
$$
\bar v=\sqrt{\frac{8\times8.31\times273}{\pi\times0.028}}\approx454\mathrm{m/s}
$$
$$
\bar\lambda_1=\frac{1.38\times10^{-23}\times273}{\sqrt{2}\pi (10^{-10})^2\times1.013\times10^5}\approx8.4\times10^{-7}\mathrm{m}
$$
$$
\bar Z_1=\frac{\bar v}{\bar\lambda_1}=\frac{454}{8.4\times10^{-7}}\approx5.4\times10^8\ \mathrm{s^{-1}}
$$
$T$不变，$p_2=1.33\times10^{-4}\mathrm{Pa}$；$\bar\lambda\propto1/p$
$$
\bar\lambda_2=\bar\lambda_1 \frac{p_1}{p_2}=8.4\times10^{-7}\times\frac{1.013\times10^5}{1.33\times10^{-4}}\approx0.64\mathrm{m}
$$
$$
\bar Z_2=\frac{\bar v}{\bar\lambda_2}=\frac{454}{0.64}\approx709\mathrm{s^{-1}}
$$

## 6.28

## 6.29

**思路**：重力场粒子数密度$n=n_0e^{-Mgh/(RT)}$；摩尔质量越大，随高度衰减越快。
氧气摩尔质量大于氮气，高度越高，氧气分子数密度衰减更快，所以高海拔处氧气比例**低于海平面**。

## 6.31

布朗运动，一维均方位移：$\overline{(\Delta x)^2}=\frac{kT}{k_p}$；$T=293\mathrm{K}$，$\sqrt{\overline{(\Delta x)^2}}=35\mathrm{nm}=35\times10^{-9}\mathrm{m}$

$$
$$
k_p=\frac{kT}{\overline{(\Delta x)^2}}=\frac{1.38\times10^{-23}\times293}{(35\times10^{-9})^2}\approx3.3\times10^{-3}\mathrm{N/m}
$$
**答案**：$\boldsymbol{k_p\approx3.3\times10^{-3}\mathrm{N/m}}$
